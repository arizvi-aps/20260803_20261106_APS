    print("-> Install Microsoft VS Code, or remove the conflicting 'code' entry from PATH.")
    print(f"FAILED: {TEST_NAME}")
else:
    print(f"MICROSOFT VS CODE FOUND: {code_path}")
    try:
        result = subprocess.run([code_path, "--version"], capture_output=True, text=True, shell=True, timeout=60)
        if result.returncode == 0 and result.stdout.strip():
            version_lines = result.stdout.strip().splitlines()
            print(f"VERSION: {version_lines[0]}")
            print(f"PASSED: {TEST_NAME}")
        else:
            print(f"VS CODE FOUND BUT COULD NOT RUN: {result.stderr.strip() or 'no output returned'}")
            print(f"FAILED: {TEST_NAME}")
    except subprocess.TimeoutExpired:
        print("VS CODE FOUND BUT '--version' TIMED OUT AFTER 60 SECONDS")
        print(f"FAILED: {TEST_NAME}")

# PlexOS-1.02D - Package Download & Dependency Validation (Plexos-to-Reporting files)
##Skip this one after the files are all updated.

#Configuration
repo_owner = "pecenak21"
repo_name = "Plexos_to_reporting"
branch = "main"
default_parent_dir = os.path.join(os.path.expanduser("~"), "Downloads")

#Step 0: Ask the user where the repo should be downloaded / updated
print(f"Choose the folder where '{repo_name}' should be downloaded or updated.")
print(f"A '{repo_name}' subfolder will be used inside the folder you pick.")
print("If you have already downloaded it, pick the folder that CONTAINS it and it will be updated in place.")

selected_dir = ""
try:
    import tkinter as tk
    from tkinter import filedialog
    root = tk.Tk()
    root.withdraw()
    root.attributes("-topmost", True)   #keeps the dialog from hiding behind the notebook window
    selected_dir = filedialog.askdirectory(parent=root, title=f"Select a folder for {repo_name}", initialdir=default_parent_dir)
    root.destroy()
except Exception:
    #No GUI available (remote / headless session): fall back to a typed path
    selected_dir = input(f"Paste a folder path (press Enter for default: {default_parent_dir}): ").strip().strip('"').strip("'")

if not selected_dir:
    selected_dir = default_parent_dir
    print(f"NO FOLDER SELECTED. USING DEFAULT: {selected_dir}")

selected_dir = os.path.normpath(selected_dir)

#If the user picked the repo folder itself, use it directly; otherwise create/use a subfolder
if os.path.basename(selected_dir).lower() == repo_name.lower():
    clone_dir = selected_dir
else:
    clone_dir = os.path.join(selected_dir, repo_name)

try:
    os.makedirs(clone_dir, exist_ok=True)
    write_test = os.path.join(clone_dir, ".write_test")
    with open(write_test, "w") as f:
        f.write("ok")
    os.remove(write_test)
except Exception as e:
    print(f"CANNOT WRITE TO FOLDER: {clone_dir}")
    print(f"Reason: {e}")
    print("-> Re-run this cell and pick a folder you have write access to (for example, your Documents or Downloads folder).")
    print("FAILED: PlexOS-1.02D - Package Download & Dependency Validation (Plexos-to-Reporting files)")
    raise RuntimeError("Download folder is not writable")

existing_file_count = sum(len(files) for _, _, files in os.walk(clone_dir))
if existing_file_count > 0:
    print(f"EXISTING LOCAL COPY FOUND ({existing_file_count} files) AT: {clone_dir}")
    print("-> Existing files will be updated to the latest version from GitHub.")
else:
    print(f"NO LOCAL COPY FOUND. DOWNLOADING A NEW COPY TO: {clone_dir}")

#Token: environment variable GITHUB_TOKEN if set, otherwise you are asked for it (input is hidden)
token = os.environ.get("GITHUB_TOKEN") or getpass("Paste your GitHub personal access token: ")
headers = {
    "Authorization": f"Bearer {token.strip()}",
    "Accept": "application/vnd.github+json",
    "X-GitHub-Api-Version": "2022-11-28",
}

#SSL setup for a TLS-inspecting corporate proxy
#Priority 1: CORP_CA_BUNDLE env var pointing to a .pem file
#Priority 2: certifi bundle + every certificate in the Windows ROOT/CA stores (includes the proxy root CA)
verify_path = None
ssl_mode = "default certifi bundle (will fail behind a TLS-inspecting proxy)"
corp_bundle = os.environ.get("CORP_CA_BUNDLE")

if corp_bundle and os.path.isfile(corp_bundle):
    verify_path = corp_bundle
    ssl_mode = f"corporate CA bundle ({corp_bundle})"
elif os.name == "nt":
    with open(certifi.where(), "r", encoding="utf-8") as f:
        pem_parts = [f.read()]
    windows_cert_count = 0
    for store_name in ("ROOT", "CA"):
        for cert_bytes, encoding, trust in ssl.enum_certificates(store_name):
            if encoding == "x509_asn":
                pem_parts.append(ssl.DER_cert_to_PEM_cert(cert_bytes))
                windows_cert_count += 1
    #Saved in the temp folder so it never lands inside the user's repo folder
    verify_path = os.path.join(tempfile.gettempdir(), "combined_ca_bundle.pem")
    with open(verify_path, "w", encoding="utf-8") as f:
        f.write("\n".join(pem_parts))
    ssl_mode = f"certifi + {windows_cert_count} Windows store certificates ({verify_path})"

session = requests.Session()
session.headers.update(headers)
if verify_path:
    session.verify = verify_path
    #requests lets this env var override session.verify, so point it at the same file
    os.environ["REQUESTS_CA_BUNDLE"] = verify_path
print(f"SSL: using {ssl_mode}")

#Step 1: Get the full, current file tree from GitHub via API
api_url = f"https://api.github.com/repos/{repo_owner}/{repo_name}/git/trees/{branch}?recursive=1"
try:
    response = session.get(api_url, timeout=60)
except requests.exceptions.SSLError as e:
    print(f"SSL CERTIFICATE ERROR: {e}")
    print("-> Set CORP_CA_BUNDLE to the path of your organization's root CA .pem file and re-run.")
    print("FAILED: PlexOS-1.02D - Package Download & Dependency Validation (Plexos-to-Reporting files)")
    raise RuntimeError("SSL verification failed")

if response.status_code != 200:
    print(f"FAILED TO REACH GITHUB API: {response.status_code}")
    print(f"GitHub says: {response.text[:300]}")
    if response.status_code in (401, 403):
        print("-> Token rejected or not authorized (check the 'repo' scope, expiry, and SSO authorization).")
    elif response.status_code == 404:
        print("-> Repo/branch not found OR the token has no access to it. Check the owner/repo/branch spelling and the token's 'repo' scope.")
    print("FAILED: PlexOS-1.02D - Package Download & Dependency Validation (Plexos-to-Reporting files)")
    raise RuntimeError(f"GitHub API returned {response.status_code}")

tree_json = response.json()
if tree_json.get("truncated"):
    print("WARNING: GitHub truncated the file tree (repo too large for one call). Some files will NOT be checked.")
tree = tree_json.get("tree", [])
repo_files = [item for item in tree if item["type"] == "blob"]
print(f"FOUND {len(repo_files)} FILES IN GITHUB REPO")

#Step 2: For each repo file, verify local match, size, and overwrite with the latest version
all_ok = True
size_mismatches = []
repo_folder_counts = defaultdict(int)
local_folder_counts = defaultdict(int)

for item in repo_files:
    file_path = item["path"]
    remote_size = item.get("size", 0)
    folder_name = os.path.dirname(file_path) or "(root)"
    repo_folder_counts[folder_name] += 1
    local_file_path = os.path.join(clone_dir, file_path.replace("/", os.sep))
    os.makedirs(os.path.dirname(local_file_path), exist_ok=True)
    existed_before = os.path.isfile(local_file_path)

    #contents API with the raw media type: works for private repos, with the same token
    file_url = f"https://api.github.com/repos/{repo_owner}/{repo_name}/contents/{quote(file_path)}?ref={branch}"
    file_response = session.get(file_url, headers={"Accept": "application/vnd.github.raw+json"}, timeout=120)

    if file_response.status_code == 200:
        with open(local_file_path, "wb") as f:
            f.write(file_response.content)
        status = "UPDATED (existing file replaced)" if existed_before else "ADDED (was missing locally)"
        print(f"{status}: {local_file_path}")
        local_size = os.path.getsize(local_file_path)
        if local_size == remote_size:
            print(f"  SIZE OK: {local_size} bytes")
            local_folder_counts[folder_name] += 1
        else:
            print(f"  SIZE MISMATCH: local={local_size} bytes, remote={remote_size} bytes")
            size_mismatches.append(file_path)
            all_ok = False
    else:
        print(f"FAILED TO DOWNLOAD: {file_path} (status {file_response.status_code})")
        all_ok = False

print("FOLDER FILE COUNT SUMMARY")
for folder in sorted(repo_folder_counts):
    expected = repo_folder_counts[folder]
    actual = local_folder_counts.get(folder, 0)
    match_status = "MATCH" if expected == actual else "MISMATCH"
    print(f"{folder}: GitHub file count: {expected}, Local path file count: {actual} - {match_status}")
    if expected != actual:
        all_ok = False

if size_mismatches:
    print(f"\n{len(size_mismatches)} FILE(S) WITH SIZE MISMATCHES:")
    for f in size_mismatches:
        print(f"  - {f}")

#Step 3: Final verification
print(f"\nLOCAL REPO LOCATION: {clone_dir}")
if all_ok:
    print("PASSED: PlexOS-1.02D - Package Download & Dependency Validation (Plexos-to-Reporting files)")
else:
    print("FAILED: PlexOS-1.02D - Package Download & Dependency Validation (Plexos-to-Reporting files)")

# PlexOS-2.02E Application Error Handling (Input File Errors)
Provide invalid, corrupted, unsupported, inaccessible, or missing input files to the application and execute the reporting workflow to validate input file validation routines, error messaging, and graceful failure handling.

source_zip_path = input("What is the path and file name of the .zip file to corrupt a COPY of? ").strip().strip('"')
output_dir = r"C:\Users\ZN1027\Downloads\SIT Outputs\PlexOS-2.02E Application Error Handling (Input File Errors)"

if not os.path.isfile(source_zip_path):
    print(f"SOURCE FILE NOT FOUND: {source_zip_path}")
    print("No file was copied or modified.")
else:
    os.makedirs(output_dir, exist_ok=True)
    output_zip_path = os.path.join(output_dir, os.path.basename(source_zip_path))
    #Copy the original file into the target folder first. shutil.copyfile
    #opens the SOURCE in read-only mode ("rb") and only ever WRITES to the destination
    shutil.copyfile(source_zip_path, output_zip_path)
    #Aggressively corrupt ONLY the copy that now lives in output_dir
    with open(output_zip_path, "rb+") as f:
        data = bytearray(f.read())
        local_file_header_sig = b"\x50\x4B\x03\x04"
        central_directory_sig = b"\x50\x4B\x01\x02"
        eocd_sig = b"\x50\x4B\x05\x06"
        for sig in (local_file_header_sig, central_directory_sig, eocd_sig):
            search_start = 0
            while True:
                idx = data.find(sig, search_start)
                if idx == -1:
                    break
                data[idx:idx + 4] = b"\xFF\xFF\xFF\xFF"
                search_start = idx + 4
        tail_start = max(0, len(data) - 32)
        for i in range(tail_start, len(data)):
            data[i] = 0xFF
        f.seek(0)
        f.write(bytes(data))
        f.truncate()
    print("Original source file (untouched):")
    print(source_zip_path)
    print("Corrupted copy written to:")
    print(output_zip_path)

#Informational (not part of any pass/fail): does the raw Total System Cost ($000) column reproduce the report values?

for tkey in TSC_CHECK_TABLES:
    if tkey not in extracted or tkey not in std_px_tables:
        continue
    table_cov = cov_df[(cov_df["Table"] == tkey) & cov_df["Status"].astype(str).str.startswith("EXTRACTED")]
    tsc_cov_rows = table_cov["Raw Column"] == TSC_LABEL
    if not tsc_cov_rows.any():
        continue
    table_is_all_tsc = bool(tsc_cov_rows.all())      #every extracted Blueprint row of the section reads the Total System Cost column
    px_data = std_px_tables[tkey]["data"]
    if px_data.shape[1] != len(raw_columns):
        print(f"  Table {tkey}: PLEXOS has {px_data.shape[1]} columns, expected {len(raw_columns)}; Total System Cost check skipped for this table")
        continue
    px_titles = normalize_series(px_data.iloc[:, 0])
    px_pos = {t: p for p, t in enumerate(px_titles)}
    px_month_pos = [i for i, c in enumerate(raw_columns) if c in month_set]
    for key, bvals in extracted[tkey].iterrows():
        is_tsc_row = table_is_all_tsc or ("total system cost" in str(extracted_names[tkey].get(key, "")).lower())
        if (not is_tsc_row) or bvals.isna().all() or key not in px_pos:
            continue
        rv = pd.to_numeric(px_data.iloc[px_pos[key], px_month_pos], errors="coerce").values.astype(float)
        bv = bvals.values.astype(float)
        good = np.isclose(rv, bv, rtol=CHECK_REL_TOL, atol=CHECK_ABS_TOL)
        ratio = float(np.nansum(rv) / np.nansum(bv)) if np.nansum(bv) != 0 else float("nan")

    table3_stack = np.vstack(table3_arrays)
    table3_zero_columns = [value_columns[i] for i in np.flatnonzero(np.all(table3_stack == 0, axis=0))]
    if len(table3_zero_columns) > 0:
        print(f"  WARNING Table 3: {len(table3_zero_columns)} of {n_value_columns} value column(s) are zero for every generator: {table3_zero_columns[:12]}")
#---- Build the paired rows ----
comparison_tables = {}      #sheet name -> {"frame": DataFrame, "diff_matrix": bool array (pair x value column)}
summary_rows = []
excluded_rows = []
only_px_records = []
for report_name in ("Standard Report", "Ratings Report"):
    report_tables = reports[report_name]
    table_numbers = shared_table_numbers_by_report.get(report_name, [])
    is_standard = (report_name == "Standard Report")
    for table_number in table_numbers:
        raw_entry = report_tables["Raw_tables"][table_number]
        px_entry = report_tables["PLEXOS_tables"][table_number]
        raw_frame_for_compare = prepare_compare_frame(raw_entry["data"], value_columns, f"[{report_name}] Table {table_number} raw")
        px_frame_for_compare = prepare_compare_frame(px_entry["data"], value_columns, f"[{report_name}] Table {table_number} PLEXOS")
        if is_standard:
            fragment_source = ROW_COMPARE_EXCLUDE_RAW_ONLY.get(table_number, [])
            exclude_fragments = [re.sub(r"\s+", " ", fragment).strip().lower() for fragment in fragment_source]
            raw_names_for_table = extracted_names.get(table_number, {})
        else:
            exclude_fragments = []
            raw_names_for_table = {}
        table3_values_for_table = table3_px_values if (is_standard and table_number == "3") else None
        frame, diff_matrix, summary_row, table_excluded, only_px_titles = build_table_comparison(
            report_name, table_number, raw_entry["title"], raw_frame_for_compare, px_frame_for_compare, value_columns, raw_names_for_table,
            CHECK_REL_TOL, ROW_COMPARE_ABS_TOL.get(table_number, CHECK_ABS_TOL), exclude_fragments, table3_values_for_table,
            meta_columns, NOT_PRESENT, RAW_NAMES_MAX_SHOWN)
        sheet_prefix = "Std " if is_standard else "Rat "
        sheet_name = (sheet_prefix + str(table_number))[:31]
        comparison_tables[sheet_name] = {"frame": frame, "diff_matrix": diff_matrix}
        summary_rows.append(summary_row)
        excluded_rows.extend(table_excluded)
        only_px_records.extend((report_name, table_number, title) for title in only_px_titles)

#---- ANNUAL TOTALS vs MONTHS: compare the Tot-YYYY / Total columns on their own, and say when the PLEXOS months are empty ----

annual_columns_out = ["Report", "Table Number", "Table Title", "PLEXOS Row Title", "Raw Data Row Title", "Raw Data Name(s)", "Annual Columns Differing",
                      "Annual Percent Error ((PLEXOS - Raw) / Raw)", "Annual Status", "Monthly Status"]
annual_percent_name = annual_columns_out[7]
annual_rows = []
for sheet_name, table in comparison_tables.items():
    pair_frame = table["frame"]
    if len(pair_frame) == 0:
        continue
    px_part = pair_frame.iloc[0::2].reset_index(drop=True)
    raw_part = pair_frame.iloc[1::2].reset_index(drop=True)
    table_number = px_part["Table Number"].iloc[0]
    report_label = px_part["Report"].iloc[0]
    table_title = px_part["Table Title"].iloc[0]
    abs_tol = ROW_COMPARE_ABS_TOL.get(table_number, CHECK_ABS_TOL)
    px_annual = px_part[annual_columns].to_numpy(dtype=float)
    raw_annual = raw_part[annual_columns].to_numpy(dtype=float)
    px_months = px_part[month_only_columns].to_numpy(dtype=float)
    raw_months = raw_part[month_only_columns].to_numpy(dtype=float)
    px_tot = px_part[tot_year_columns].to_numpy(dtype=float)
    raw_tot = raw_part[tot_year_columns].to_numpy(dtype=float)
    for i in range(len(px_part)):
        px_title = px_part.at[i, "Row Title"]
        raw_title = raw_part.at[i, "Row Title"]
        record = {"Report": report_label, "Table Number": table_number, "Table Title": table_title, "PLEXOS Row Title": px_title, "Raw Data Row Title": raw_title,
                  "Raw Data Name(s)": raw_part.at[i, "Raw Data Name(s)"], "Annual Columns Differing": np.nan, annual_percent_name: np.nan,
                  "Annual Status": "", "Monthly Status": ""}
        if px_title == NOT_PRESENT or raw_title == NOT_PRESENT:
            record["Annual Status"] = "Only in Raw Data" if px_title == NOT_PRESENT else "Only in PLEXOS"
            record["Monthly Status"] = record["Annual Status"]
            annual_rows.append(record)
            continue
        if np.isnan(raw_annual[i]).all() and np.isnan(raw_months[i]).all():
            record["Annual Status"] = "Raw identifier only"
            record["Monthly Status"] = "Raw identifier only"
            annual_rows.append(record)
            continue
        annual_close = np.isclose(np.nan_to_num(px_annual[i]), np.nan_to_num(raw_annual[i]), rtol=CHECK_REL_TOL, atol=abs_tol)
        differing_count = int((~annual_close).sum())
        record["Annual Columns Differing"] = differing_count
        record["Annual Status"] = "ANNUAL MATCH" if differing_count == 0 else "ANNUAL DIFFERS"
        px_sum = float(np.nansum(px_tot[i]))
        raw_sum = float(np.nansum(raw_tot[i]))
        if abs(raw_sum) > 1e-9:
            record[annual_percent_name] = (px_sum - raw_sum) / raw_sum
        elif abs(px_sum) <= 1e-9:
            record[annual_percent_name] = 0.0
        px_months_empty = bool((np.nan_to_num(px_months[i]) == 0).all()) and bool((np.nan_to_num(raw_months[i]) != 0).any())
        if px_months_empty:
            record["Monthly Status"] = "PLEXOS MONTHS EMPTY"
        else:
            month_close = np.isclose(np.nan_to_num(px_months[i]), np.nan_to_num(raw_months[i]), rtol=CHECK_REL_TOL, atol=abs_tol)
            if bool(month_close.all()):

            else:
                record["Monthly Status"] = f"MONTHS DIFFER ({int((~month_close).sum())} of {len(month_only_columns)})"
        annual_rows.append(record)
annual_results = pd.DataFrame(annual_rows, columns=annual_columns_out)

                            print(f"  [{rating_key}] ':{rating_definition['alias']}' basis check: {rating_used_matching} of {rating_rows_checked} rows agree with {rating_describe_collections(rating_measure['collections'])} (used); {rating_alternative_matching} of {rating_rows_checked} would agree with {rating_describe_collections(rating_measure['alternative'])}")
                            if rating_alternative_matching > rating_used_matching:
                                print(f"  WARNING [{rating_key}]: more ':{rating_definition['alias']}' rows agree with the alternative basis than with the one used: change the 'collections' of '{rating_definition['alias'].lower()}' in RATING_ALIAS_MEASURES")
                        rating_section_frames.append(rating_pairs_frame)
                        rating_section_diffs.append(rating_diff_matrix)
                        rating_excluded_rows.extend(rating_excluded_here)
                        rating_summary_rows.append(rating_section_row)
                    except Exception as rating_definition_error:
                        print(f"  PROBLEM [{rating_key}]: the ':{rating_definition['alias']}' rows of the Timeslice '{rating_raw_timeslice}' could not be compared ({type(rating_definition_error).__name__}: {str(rating_definition_error)[:150]}); the other rows are not affected")
                        rating_not_compared_rows.append({"Report": "Ratings Report", "Table Number": rating_key, "Row Title / Section": f"(':{rating_definition['alias']}' rows of '{rating_section['title']}')", "Reason": f"comparison failed: {type(rating_definition_error).__name__}: {str(rating_definition_error)[:150]}"})
                if len(rating_section_frames) > 0:
                    rating_section_tables[f"Rat {rating_key}"] = {"frame": pd.concat(rating_section_frames, ignore_index=True), "diff_matrix": np.vstack(rating_section_diffs)}
                    rating_pair_frames.append(rating_section_tables[f"Rat {rating_key}"]["frame"])
            except Exception as rating_error:
                print(f"  PROBLEM [{rating_key}]: the Timeslice '{rating_raw_timeslice}' could not be compared ({type(rating_error).__name__}: {str(rating_error)[:150]}); the rest of the test case is not affected")
                rating_not_compared_rows.append({"Report": "Ratings Report", "Table Number": rating_key, "Row Title / Section": f"(whole section '{rating_section['title']}')", "Reason": f"comparison failed: {type(rating_error).__name__}: {str(rating_error)[:150]}"})
                continue
            if len(rating_section["missing_columns"]) > 0:
                print(f"  WARNING [{rating_key}]: the Ratings Report section has no column for {len(rating_section['missing_columns'])} of {len(rating_value_columns)} value column(s): {rating_section['missing_columns'][:8]}")
    for rating_ts_key, rating_ts_group in rating_raw_timeslices_by_key.items():
        if rating_ts_key not in rating_sections:
            rating_not_compared_rows.append({"Report": "Ratings Report", "Table Number": rating_ts_key, "Row Title / Section": f"(raw Timeslice {rating_ts_group})", "Reason": "the Ratings Report has no section for this raw Timeslice"})
            print(f"  NOTE: the raw Timeslice {rating_ts_group} has no section in the Ratings Report: not compared")
finally:
    #load_raw_frames_cached keeps ONE prepared copy in the session cache: prepare it for the Standard Report Timeslice again, so every other test case finds the cache as it was
    with contextlib.redirect_stdout(io.StringIO()):
        load_raw_frames_cached(raw_data_path=raw_data_path, standard_timeslice=STANDARD_TIMESLICE, excel_serial_min=EXCEL_SERIAL_MIN, excel_serial_max=EXCEL_SERIAL_MAX)
rating_summary_columns = ["Report", "Table Number", "Table Title", "Raw Timeslice", "Row Pairs", "MATCH", "DIFFERS", "Only in PLEXOS", "Only in Raw Data", "Excluded (removed earlier)", "Largest Absolute Difference", "Ratings Columns Missing", "Ratings Rows Not Compared",
                          "Compared Rows", "Raw Collections Used", "Rows Matching The Alternative Basis"]
rating_summary = pd.DataFrame(rating_summary_rows, columns=rating_summary_columns)
rating_all_pairs = pd.concat(rating_pair_frames, ignore_index=True) if len(rating_pair_frames) > 0 else pd.DataFrame(columns=meta_columns + list(rating_value_columns))
rating_excluded = pd.DataFrame(rating_excluded_rows, columns=["Report", "Table Number", "Side", "Row Title", "Raw Data Name(s)", "Reason"])
rating_not_compared = pd.DataFrame(rating_not_compared_rows, columns=["Report", "Table Number", "Row Title / Section", "Reason"])
rating_pairs_path = save_results_csv(rating_all_pairs, results_folder, "PlexOS_3_02_Ratings_Generation_Comparison.csv")
rating_summary_path = save_results_csv(rating_summary, results_folder, "PlexOS_3_02_Ratings_Generation_Summary.csv")
rating_workbook_path = results_folder / "PlexOS_3_02_Ratings_Generation_Comparison.xlsx"
write_rating_generation_workbook(rating_workbook_path, rating_summary, rating_section_tables, rating_excluded, rating_not_compared, meta_columns, rating_value_columns, meta_widths, fill_head, fill_band, fill_diff)
print("\nRATINGS REPORT COMPARISON (PLEXOS row directly above the matching Raw Data row; informational, not part of PASS/FAIL)")
print(rating_summary.drop(columns=["Table Title", "Raw Collections Used"]).to_string(index=False))
for _, rating_summary_row in rating_summary.iterrows():
    rating_section_pairs = rating_section_tables[f"Rat {rating_summary_row['Table Number']}"]["frame"]
    rating_alias_text = re.search(r"'(:[^']+)' rows", rating_summary_row["Compared Rows"]).group(1)
    rating_px_rows = rating_section_pairs[(rating_section_pairs["Source"] == "PLEXOS") & (rating_section_pairs["Pair Status"] != "MATCH") & (rating_section_pairs["Table Title"] == rating_summary_row["Table Title"])]
    if len(rating_px_rows) > 0:
        print(f"  [{rating_summary_row['Table Number']}] {rating_alias_text} rows: {len(rating_px_rows)} row pair(s) are not MATCH:")
        for _, rating_not_match_row in rating_px_rows.head(15).iterrows():
            print(f"    '{rating_not_match_row['Row Title']}': {rating_not_match_row['Pair Status']} (raw name(s): {rating_not_match_row['Raw Data Name(s)'] if rating_not_match_row['Raw Data Name(s)'] else 'n/a'})")
        if len(rating_px_rows) > 15:
            print(f"    ... and {len(rating_px_rows) - 15} more in the comparison file")
    else:
        print(f"  [{rating_summary_row['Table Number']}] {rating_alias_text} rows: every compared row pair is a MATCH ({int(rating_summary_row['MATCH'])} of {int(rating_summary_row['Row Pairs'])})")
print(f"  Ratings Report rows not compared (label not ending in a compared suffix) or sections without a counterpart: {len(rating_not_compared)} (sheet 'Not compared'); rows left out because an earlier filter removed them: {len(rating_excluded)} (sheet 'Excluded')")
print(f"  Rows written: {len(rating_all_pairs)} ({len(rating_all_pairs) // 2} pairs across {len(rating_section_tables)} Timeslice(s))")
print(f"  CSV:     {rating_pairs_path}")
print(f"  Summary: {rating_summary_path}")
print(f"  Excel:   {rating_workbook_path}")
print("Ratings Report comparison complete.")

                print(f"  PROBLEM [{rating_key}]: the Timeslice '{rating_raw_timeslice}' could not be compared ({type(rating_error).__name__}: {str(rating_error)[:150]}); the rest of the test case is not affected")
                rating_not_compared_rows.append({"Report": "Ratings Report", "Table Number": rating_key, "Row Title / Section": f"(whole section '{rating_section['title']}')", "Reason": f"comparison failed: {type(rating_error).__name__}: {str(rating_error)[:150]}"})
                continue

            for occurrence in range(min(len(raw_positions[key]), len(px_positions[key]))):
                raw_series_full = raw_matrix[raw_row_of[raw_positions[key][occurrence]]]
                if report_name == "Standard Report" and table_number == "3" and key in table3_px_values and occurrence < len(table3_px_values[key]):
                    px_series_full = np.asarray(table3_px_values[key][occurrence], dtype=float)        #Primary + Secondary + Topping (matches the raw Offtake definition)
                else:
                    px_series_full = px_matrix[px_row_of[px_positions[key][occurrence]]]
                raw_count_numeric, raw_stats = summarize_series(raw_series_full)
                px_count_numeric, px_stats = summarize_series(px_series_full)
                row_title = px_display.get(key, raw_display.get(key, key))
                if raw_count_numeric == 0:
                    #structurally blank raw field: no raw measure exists, so there is nothing to reconcile
                    magnitude_na_rows.append({"Report": report_name, "Table Number": table_number, "Row Title": row_title, "Reason": "Structurally blank raw row (no raw measure): excluded as not applicable"})
                    continue
                result_row = {"PlexOS Step": "PlexOS-3.03C", "Report": report_name, "Table Number": table_number, "Table Title": table_title, "Row Title": row_title,
                              "Raw Data Name(s)": raw_name_lookup.get(key, ""), "Occurrence": occurrence + 1, "Raw Numeric Count": raw_count_numeric, "PLEXOS Numeric Count": px_count_numeric}
                stat_fields, all_stats_match = compute_magnitude_row(raw_stats, px_stats, magnitude_stats, ZERO_THRESHOLD)
                result_row.update(stat_fields)
                #Monthly-series check (informational): the power-of-ten test only compares orders of magnitude, so raw 9.9M vs PLEXOS 3.2M passes it. This compares the 72 monthly values themselves.
                series_names_count = max(1, len([part for part in str(raw_name_lookup.get(key, "")).split(" + ") if part.strip()]))
                series_max_pct, series_months_differing, series_label = calculate_series_check(raw_series_full, px_series_full, series_rounding_per_month * series_names_count, ZERO_THRESHOLD, SERIES_REL_TOL)
                result_row["Max Monthly Difference (% of row level)"] = series_max_pct
                result_row["Months Differing"] = series_months_differing
                result_row["Series Check"] = series_label
                result_row["Row Status"] = "PASS" if all_stats_match else "FAIL"
                result_row["Row Status Pass"] = 1 if all_stats_match else 0
                magnitude_rows.append(result_row)
        #Mapping audit for this section (Standard Report): for EVERY PLEXOS row, is the raw row the comparison used the right one? When not, every raw resource is tried and the best match is suggested as a maptable_ fix.
        if report_name == "Standard Report" and len(monthly_columns) > 0:
            mapping_audit_rows.extend(audit_section_row_mapping(table_number, table_title, raw_df, px_df, raw_positions, px_positions, px_display, raw_name_lookup, monthly_columns, table3_px_values,
                                                                series_sources, raw_parts, RAW_ROUNDING_STEP, AUDIT_REL_TOL, ZERO_THRESHOLD))

mapping_audit_path = save_results_csv(mapping_audit, results_folder, "PlexOS_3_03C_Mapping_Audit.csv")

# PlexOS- 3.03D Generation, Cost, and Cost-per-MWh Abnormality Screening
Verify that generation, cost, and derived cost-per-MWh results produced by the PlexOS reporting process remain within an acceptable statistical range when compared to the approved RT Sim baseline data. This validation identifies anomalous report rows by comparing PlexOS averages against the corresponding RT Sim mean and standard deviation for selected generation and cost report sections. Results exceeding the defined statistical threshold are identified and reported for review.

#PlexOS-3.03D Generation, Cost, and Cost-per-MWh Abnormality Screening
TC_NAME = "PlexOS-3.03D Generation, Cost, and Cost-per-MWh Abnormality Screening"
NO_COUNTERPART = "No solution.zip counterpart"
RATINGS_NO_RAW_REASON = "Rating-period data is not in the raw data yet, so the Ratings Report has no raw counterpart"
SD_FLAG_THRESHOLD = 3          #distance (in raw standard deviations) above which a row is flagged as an abnormality
DOUBLE_COUNT_RATIO_LOW, DOUBLE_COUNT_RATIO_HIGH = 1.9, 2.1      #a Raw / PLEXOS mean ratio in this band is the signature of raw values counted twice
DOUBLE_COUNT_WARN_SHARE = 0.05                                  #warn when more than this share of the screened rows sits in that band
ZERO_THRESHOLD = 1e-9
#Standard Report sections in scope for the entire SIT process = the sections used in the data model (all other sections are excluded everywhere).
IN_SCOPE_TABLES = ["1", "2", "3", "5", "6", "7", "8", "10", "11", "21", "22", "23", "26", "31",
                   "33c", "34c", "56a", "57a", "80", "87"]
GENERATION_TABLES = ["2", "33c", "34c", "56a"]        #Steps 2: required to be evaluated for generation abnormalities
COST_TABLES = ["5", "6", "7"]                         #Step 3: required to be evaluated for cost abnormalities

#---- Step 1 result ----
step1_checks = {
    "All required input files found": True,
    "Raw data normalized (Standard sections rebuilt from Blueprint_std)": len(reports["Standard Report"]["Raw_tables"]) > 0,
    "PLEXOS Standard Report parsed into sections": len(reports["Standard Report"]["PLEXOS_tables"]) > 0,
    "PLEXOS Ratings Report parsed into sections": len(reports["Ratings Report"]["PLEXOS_tables"]) > 0,
    f"Raw measures use only Timeslice == '{STANDARD_TIMESLICE}' rows ({len(timeslice_check)} sheet / measure combination(s) checked)": timeslice_ok,

print_status_checks(step1_checks, status_text)
if len(raw_ratings_tables) > 0:
    print(f"  Ratings Report raw counterpart: {len(raw_ratings_tables)} Timeslice section(s) built from the raw Generator sheet; screened like the Standard Report sections")

#Screening engine. ONE pass builds a list of "jobs" (each job = one matched row occurrence with its raw and output monthly series) for
#(a) EVERY in-scope Standard Report section (the 20 sections used in the data model: generation tables 2, 33c, 34c, 56a; cost tables 5, 6, 7; and every other in-scope section), and
#(b) cost-per-MWh (raw Fuel Cost $000 / Generation GWh vs output Table 6 Fuel Cost / Table 2 Generation, month by month).
#Every job is then screened the same way: raw row mean and standard deviation across the 72 monthly values, distance = ABS(output mean - raw mean) / raw standard deviation, flagged when the distance is greater than 3.
#A raw standard deviation of 0 is not divided: the row is flagged only if the output mean differs from the raw mean, and it is reported separately.
#Rows are matched on the row title (case-insensitive, extra spaces ignored) and occurrence. Retired resources, Total / blank / sub-rows and raw rows that do not belong in a table are already removed above.
#(c) the Ratings Report: every ':MW' row (raw Generation) and every ':MBTU' row (raw Offtake of Generator.Fuels, Start Fuels excluded) of every rating period is screened the same way, against the raw data of that Timeslice (category 'Rating').
period_pattern = re.compile(r"^(Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec)-\d{4}$")
total_pattern = re.compile(r"^Tot-\d{4}$|^Total$")
reports_without_raw = get_reports_without_raw(reports)
standard_raw_tables = reports["Standard Report"]["Raw_tables"]
standard_px_tables = reports["Standard Report"]["PLEXOS_tables"]
section_log, jobs, section_blocks = [], [], {}          #one row per section considered / one entry per matched row occurrence / numeric monthly blocks and row positions of each section (reused for cost-per-MWh)
monthly_columns_used, excluded_columns_log = [], []     #(report, table, columns actually used / Tot-YYYY and Total columns excluded) for Steps 4 and 5
unmatched_rows = []                                     #rows present on one side only (presence is validated in PlexOS-3.03B; listed here for context)
table3_rows_rebuilt = 0                                 #Table 3 PLEXOS rows screened as Primary + Secondary + Topping
for table_number in SCREENED_TABLES:
    category = table_category(table_number)
    raw_entry, px_entry = standard_raw_tables.get(table_number), standard_px_tables.get(table_number)
    table_title = raw_entry["title"] if raw_entry is not None else (px_entry["title"] if px_entry is not None else blueprint_titles.get(table_number, ""))
    if raw_entry is None or px_entry is None:
        section_log.append(describe_missing_section(table_number, table_title, category, raw_entry, px_entry, NO_COUNTERPART))
        continue
    section_jobs, section_block, section_unmatched, section_excluded_cols, section_table3_rebuilt, section_record = build_section_row_jobs(

    jobs.extend(section_jobs)
    section_blocks[table_number] = section_block
    unmatched_rows.extend(section_unmatched)
    table3_rows_rebuilt += section_table3_rebuilt
    monthly_columns_used.append(("Standard Report", table_number, section_block["monthly_cols"]))
    excluded_columns_log.append(("Standard Report", table_number, section_excluded_cols))
    section_log.append(section_record)
#Ratings Report: every rating period that has a raw counterpart is screened, one kind of row at a time (':MW' = Generation, ':MBTU' = Offtake of Generator.Fuels): Table Number 'atc:MW', 'atc:MBTU', 'offpeak:MW', ... A rating period without a raw counterpart is listed as 'No solution.zip counterpart'.
rating_jobs, rating_rows_screened = [], 0
for rating_key in sorted(set(reports["Ratings Report"]["Raw_tables"]) | set(reports["Ratings Report"]["PLEXOS_tables"]), key=str):
    rating_raw_entry, rating_px_entry = reports["Ratings Report"]["Raw_tables"].get(rating_key), reports["Ratings Report"]["PLEXOS_tables"].get(rating_key)
    if rating_raw_entry is None:
        section_log.append(build_section_log_record("Ratings Report", rating_key, rating_px_entry["title"], "Rating period", "N/A - " + NO_COUNTERPART, 0, rating_no_raw_notes.get(rating_key, rating_no_raw_default) if len(raw_ratings_tables) > 0 else RATINGS_NO_RAW_REASON))
        continue
    if rating_px_entry is None:
        section_log.append(build_section_log_record("Ratings Report", rating_key, rating_raw_entry["title"], "Rating", "FAIL", 0, "the rating period is missing from the PLEXOS Ratings Report"))
        continue
    for rating_definition in rating_definitions:
        rating_table_label = f"{rating_key}:{rating_definition['alias']}"
        rating_table_title = f"{rating_px_entry['title']} :{rating_definition['alias']} rows (raw {rating_definition['measure']['label']}: {rating_definition['measure']['column']} x {RATING_UNIT_FACTOR:g})"
        rating_section_jobs, rating_section_unmatched, rating_section_excluded_cols, rating_section_monthly_cols, rating_section_record = build_rating_section_row_jobs(
            "Ratings Report", rating_key, rating_table_label, rating_table_title, "Rating", rating_definition["alias"], rating_raw_entry["data"], rating_px_entry["data"], month_labels, total_pattern, rating_raw_name_lookup.get(rating_key, {}))
        rating_jobs.extend(rating_section_jobs)
        rating_rows_screened += len(rating_section_jobs)
        unmatched_rows.extend(rating_section_unmatched)
        monthly_columns_used.append(("Ratings Report", rating_table_label, rating_section_monthly_cols))
        excluded_columns_log.append(("Ratings Report", rating_table_label, rating_section_excluded_cols))
        section_log.append(rating_section_record)
#Cost-per-MWh jobs: raw Fuel Cost ($000) / Generation (GWh) per resource-month, output Table 6 Fuel Cost / Table 2 Generation per resource-month
cpm_available = (CPM_GENERATION_TABLE in section_blocks) and (CPM_COST_TABLE in section_blocks)

jobs.extend(cpm_jobs)
jobs.extend(rating_jobs)          #the Ratings rows come last, so the Standard and cost-per-MWh rows keep their position in every export
section_log.append(cpm_record)

screen_columns = ["Category", "Report", "Table Number", "Table Title", "Row Title", "Raw Data Name(s)", "Occurrence", "Raw Numeric Count", "PLEXOS Numeric Count",
                  "Raw Row Average", "PLEXOS Row Average", "Raw Row Standard Deviation", "Row Standard Deviations From Raw Average", "Zero Variance Row",
                  "Row Outside 3 Raw Standard Deviations", "Row Unable to Evaluate", "Screening Note",
                  "Raw Table Sum", "PLEXOS Table Sum", "Raw Table Average", "PLEXOS Table Average", "Raw Table Standard Deviation",
                  "Table Average Outside 3 Raw Standard Deviations", "Raw / PLEXOS Mean Ratio"]
all_results = build_results_frame(result_rows, table_accumulators, SD_FLAG_THRESHOLD)
row_results, cpm_results = split_results_by_job_type(all_results, screen_columns)
zero_variance_results = round_numeric_columns(pd.DataFrame(zero_variance_rows))
section_results = pd.DataFrame(section_log)
unmatched_results = pd.DataFrame(unmatched_rows, columns=["Report", "Table Number", "Row Title", "Side"])
#Which of the 20 in-scope sections made it into the screening (and therefore into the _ALL files)

    print(f"  PROBLEM: {reference_problem}")

#Build one histogram and the box-and-whisker data for every mapped section. POPULATION = the matched row occurrences (the rows that exist on BOTH sides, paired title by title and

    for table_number, raw_entry, px_entry, section_rating_key in section_iterations:

        px_positions, px_display = index_row_positions(px_df)

        raw_name_lookup = globals().get("extracted_names", {}).get(table_number, {}) if report_name == "Standard Report" else rating_raw_name_lookup.get(section_rating_key, {})      #Ratings: the raw units behind each ':MW' / ':MBTU' row
        excluded_row_rows.extend(find_unpaired_rows(report_name, table_number, table_title, raw_positions, px_positions, raw_display, px_display, raw_name_lookup))
        if len(common_keys) > 0 and len(monthly_cols) > 0:
            raw_matrix, px_matrix, raw_row_of, px_row_of = build_matched_matrices(raw_df, px_df, monthly_cols, raw_positions, px_positions, common_keys)
        else:
            common_keys, raw_matrix, px_matrix, raw_row_of, px_row_of = [], None, None, {}, {}
        raw_chunks, px_chunks, row_records, row_box_jobs, blank_rows = pair_matched_rows(report_name, table_number, table_title, common_keys, raw_positions, px_positions, raw_matrix, px_matrix, raw_row_of, px_row_of,

                         "Raw Non-Zero Count", "PLEXOS Non-Zero Count", "Raw Mean", "Raw Median", "PLEXOS Mean", "PLEXOS Median", "Mean Ratio (Raw / PLEXOS)", "Bins", "Bin Min", "Bin Max",
                         "Raw Bar Total", "PLEXOS Bar Total", "Plotted Bin Edges Identical", "Plotted Bar Totals Match Counts", "Markers Drawn", "Histogram Path"]

The lower / upper whisker is the smallest / largest value inside the fences, but never inside the box (if no value lies between the box edge and the fence, the whisker ends at the box edge, as in matplotlib). for problem in h9_problems:

record_step_result("H9", len(section_long) > 0 and len(h9_problems) == 0, hist_step_results, HIST_TC_NAME, status_text)
#Step H10: the population and the statistics of this block agree with the PlexOS-3.03C results, row by row (same rows, same minimum / maximum / mean / median on the raw side and on the PLEXOS side)
print("\n----- Step H10: Agreement with PlexOS-3.03C -----")
h10_problems = []
if reference_results is None or reference_compare.empty:
    if reference_results is None:
        h10_problems.append("no usable PlexOS-3.03C results to compare with (run the PlexOS-3.03C script first, in the same session)" + (f": {reference_problem}" if reference_problem != "" else ""))
    else:
        h10_problems.append(f"this block has no matched row to compare: no section has both a raw table and a PLEXOS table with matching row titles, although the 3.03C results hold {len(reference_results)} row(s) "
                            "(the raw tables of this session are probably not the ones PlexOS-3.03C built: run the corrected 3.03C script, then this block, in the same session)")
    h10_pass = False
else:
    h10_problems, h10_has_rows_in_both = summarize_reference_comparison(reference_compare, RECONCILE_RTOL, DOUBLE_COUNT_BAND)
    h10_pass = bool(len(h10_problems) == 0 and h10_has_rows_in_both)

    step2_problem_count += (0 if run["Successful"] else 1) + (0 if len(step_mismatches) == 0 else 1)
    step2_view.append({"Test ID": run["Test ID"], "Steps PASS / total": f"{sum(1 for v in run['Steps'].values() if v == 'PASS')}/{len(run['Steps'])}", "Approved": "yes" if approved else "not yet",
                       "Steps As Approved": "n/a" if not approved else status_text[len(step_mismatches) == 0], "Result": status_text[run["Successful"] and len(step_mismatches) == 0]})
missing_approved_runs = [i for i in (baseline["tests"].keys() if baseline is not None else []) if i not in runs_by_id]
for missing_id in missing_approved_runs:
    check_rows.append({"Category": "SIT test execution", "Check": missing_id, "Description": "approved test case has results", "Passed": False, "Detail": "no Step Summary found in the results folder"})
    exception_rows.append({"Step": 2, "Category": "SIT test execution", "Check": missing_id, "Problem": "the approved test case has no results in the results folder (run it first)"})
step2_pass = bool(len(test_runs) > 0 and step2_problem_count == 0 and len(missing_approved_runs) == 0 and (baseline is not None or baseline_pending))
print(f"Step 2 - Approved SIT Test Cases Executed Successfully ({len(test_runs)} test case(s) with results; {len(baseline['tests']) if baseline is not None else 0} approved)")
if len(step2_view) > 0:
    print(pd.DataFrame(step2_view).to_string(index=False))
for run in test_runs:
    if run["Why Not Successful"] != "":
        print(f"  PROBLEM {run['Test ID']}: {run['Why Not Successful']}")
for missing_id in missing_approved_runs:
    print(f"  PROBLEM {missing_id}: approved but no results found in the results folder")
record_step_result(2, step2_pass, step_results, TC_NAME, status_text)
#---------------------------------------------------------------------
#Step 3: Verify previously resolved defects do not reoccur. The automatic guard (one per test case) checks that no step of the test case reports FAIL; RESOLVED_DEFECTS (optional, at the top)
#adds project-specific checks. A defect that cannot be checked counts as UNVERIFIED (not passed), never as silently fixed.
#---------------------------------------------------------------------
print("\n----- Step 3: Previously resolved defects do not reoccur -----")
defects = [{"Defect ID": f"{AUTO_DEFECT_ID_PREFIX}-{n:02d}", "Description": f"AUTO GUARD: no step of {tid} reports FAIL", "Test ID": tid, "Check Type": "no_fail_step", "File": "", "Filter": "", "Step": "", "Expected": ""}
           for n, tid in enumerate(ordered_test_ids, start=1)]
for custom_defect in RESOLVED_DEFECTS:
    defects.append({key: str(custom_defect.get(key, "")).strip() for key in ("Defect ID", "Description", "Test ID", "Check Type", "File", "Filter", "Step", "Expected")})
defect_results = []
for defect in defects:
    run = runs_by_id.get(defect["Test ID"])
    outcome = {"Defect ID": defect["Defect ID"], "Description": defect["Description"], "Test ID": defect["Test ID"], "Check Type": defect["Check Type"], "Result": "", "Detail": ""}
    if defect["Check Type"] in ("no_fail_step", "step_status") and run is None:
        outcome["Result"], outcome["Detail"] = "UNVERIFIED", f"no results for {defect['Test ID']} in the results folder"
    elif defect["Check Type"] == "no_fail_step":
        failing = [k for k, v in run["Steps"].items() if v == "FAIL"]
        outcome["Result"] = "NOT REOCCURRED" if len(failing) == 0 else "REOCCURRED"
        outcome["Detail"] = "no step reports FAIL" if len(failing) == 0 else f"step(s) {failing} report FAIL"
    elif defect["Check Type"] == "step_status":
        expected_step_value = defect["Expected"] if defect["Expected"] != "" else "PASS"
        current_step_value = run["Steps"].get(defect["Step"], "(not reported)")
        outcome["Result"] = "NOT REOCCURRED" if current_step_value == expected_step_value else "REOCCURRED"
        outcome["Detail"] = f"step {defect['Step']} reports {current_step_value} (expected {expected_step_value})"
    elif defect["Check Type"] in ("no_rows", "row_count"):
        matches = [p for rel, p in result_files.items() if fnmatch.fnmatch(p.name.lower(), defect["File"].lower())]
        if defect["File"] == "" or len(matches) == 0:
            outcome["Result"], outcome["Detail"] = "UNVERIFIED", f"the result file '{defect['File']}' is not in the results folder"
        elif defect["Check Type"] == "row_count" and not re.match(r"^\d+$", defect["Expected"]):
            outcome["Result"], outcome["Detail"] = "UNVERIFIED", "Check Type row_count needs a whole-number Expected"
        else:
            try:
                data_frame = pd.read_csv(max(matches, key=lambda f: f.stat().st_mtime))
                matched_rows = int(len(data_frame.query(defect["Filter"], engine="python"))) if defect["Filter"] != "" else int(len(data_frame))
                if defect["Check Type"] == "no_rows":
                    outcome["Result"] = "NOT REOCCURRED" if matched_rows == 0 else "REOCCURRED"
                    outcome["Detail"] = f"{matched_rows} row(s) match the defect symptom (expected 0)"
                else:
                    outcome["Result"] = "NOT REOCCURRED" if matched_rows == int(defect["Expected"]) else "REOCCURRED"
                    outcome["Detail"] = f"{matched_rows} row(s) match (expected {defect['Expected']})"
            except Exception as exc:
                outcome["Result"], outcome["Detail"] = "UNVERIFIED", f"the Filter could not be evaluated: {type(exc).__name__}: {str(exc)[:110]}"
    else:
        outcome["Result"], outcome["Detail"] = "UNVERIFIED", f"unknown Check Type '{defect['Check Type']}'"
    defect_results.append(outcome)
    check_rows.append({"Category": "Defect not reoccurred", "Check": defect["Defect ID"], "Description": defect["Description"], "Passed": outcome["Result"] == "NOT REOCCURRED", "Detail": outcome["Detail"]})
    if outcome["Result"] != "NOT REOCCURRED":
        exception_rows.append({"Step": 3, "Category": "Defect not reoccurred", "Check": defect["Defect ID"], "Problem": f"{outcome['Result']}: {outcome['Detail']}"})
defect_frame = pd.DataFrame(defect_results, columns=["Defect ID", "Description", "Test ID", "Check Type", "Result", "Detail"])
only_auto_guards = len(RESOLVED_DEFECTS) == 0
step3_pass = bool(len(defect_frame) > 0 and (defect_frame["Result"] == "NOT REOCCURRED").all())
print(f"Step 3 - Resolved Defects ({len(defect_frame)} defect check(s); {int((defect_frame['Result'] == 'NOT REOCCURRED').sum())} not reoccurred, {int((defect_frame['Result'] == 'REOCCURRED').sum())} reoccurred, {int((defect_frame['Result'] == 'UNVERIFIED').sum())} unverified)")
if len(defect_frame) > 0:
    print(defect_frame[["Defect ID", "Test ID", "Check Type", "Result", "Detail"]].assign(Detail=defect_frame["Detail"].str.slice(0, 80)).to_string(index=False))
if only_auto_guards:
    print("  NOTE: only the automatic guards were checked (RESOLVED_DEFECTS is empty): this confirms that no step of any test case reports FAIL.")
record_step_result(3, step3_pass, step_results, TC_NAME, status_text)

#Step 4: Verify the historical comparison passes at the approved tolerance, using the approved input versions.
#(a) Versions: raw data, section-mapping template and pinned files are byte-identical to the approved versions (SHA-256).
#(b) Historical comparison: every approved result file is compared with the file the earlier test case just produced: same columns, same row count, text equal, numbers within
#|new - approved| <= abs_tol + rel_tol x |approved|. Run-to-run noise columns are ignored. Differences are exported.
#When no approved baseline exists yet (and ESTABLISH_BASELINE_IF_MISSING = True) it is written here from this run's results, provided Steps 2 and 3 passed.
#---------------------------------------------------------------------
print("\n----- Step 4: Historical comparison at the approved tolerance (approved versions) -----")
baseline_established_now = False
if baseline is None and baseline_pending:
    if step_results[2] and step_results[3]:
        try:
            (baseline_dir / "files").mkdir(parents=True, exist_ok=True)
            for relative_name, source_csv in result_files.items():
                destination = baseline_dir / "files" / relative_name
                destination.parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(source_csv, destination)
            baseline = {"approved_on": run_started.strftime("%Y-%m-%d %H:%M:%S"), "approved_by": get_current_user_name(), "established_from_run": run_id, "rel_tol": REL_TOL, "abs_tol": ABS_TOL,
                        "versions": {k: v for k, v in current_versions.items()}, "files": sorted(result_files.keys()),
                        "tests": {r["Test ID"]: {"expected_status": "PASS", "expected_steps": r["Steps"]} for r in test_runs}}
            write_json_file(baseline_json_path, baseline)
            baseline_established_now = True
            print(f"  APPROVED BASELINE ESTABLISHED from this run: {baseline_json_path} ({len(baseline['tests'])} test case(s), {len(baseline['files'])} result file(s)).")
            print("  The comparisons below start with the next run (this run IS the baseline).")
        except Exception as exc:
            print(f"  PROBLEM: the approved baseline could not be written: {type(exc).__name__}: {exc}")
            exception_rows.append({"Step": 4, "Category": "Baseline", "Check": "establish baseline", "Problem": f"{type(exc).__name__}: {exc}"})
    else:
        print("  The approved baseline was NOT established because Step 2 or Step 3 did not pass.")
        exception_rows.append({"Step": 4, "Category": "Baseline", "Check": "establish baseline", "Problem": "not established: Step 2 or Step 3 did not pass"})
version_rows = []
versions_ok = True
comparison_ok = False
diff_rows = []
file_compare_rows = []
if baseline is not None and not baseline_established_now:
    approved_versions = baseline["versions"]
    for role_name in sorted(set(list(approved_versions.keys()) + list(current_versions.keys()))):
        approved_record = approved_versions.get(role_name)
        current_record = current_versions.get(role_name)
        if approved_record is None or not approved_record.get("compared", True):
            version_rows.append({"Role": role_name, "Approved File": approved_record["file_name"] if approved_record else "(not in the baseline)", "Current File": current_record["file_name"] if current_record else "",
                                 "Approved SHA-256": approved_record["sha256"] if approved_record else "", "Current SHA-256": current_record["sha256"] if current_record else "", "Result": "RECORDED ONLY"})
            continue
        same = bool(current_record is not None and current_record["sha256"] == approved_record["sha256"])
        versions_ok = versions_ok and same
        version_rows.append({"Role": role_name, "Approved File": approved_record["file_name"], "Current File": current_record["file_name"] if current_record else "(not supplied)",
                             "Approved SHA-256": approved_record["sha256"], "Current SHA-256": current_record["sha256"] if current_record else "", "Result": "SAME VERSION" if same else "DIFFERENT VERSION"})
        check_rows.append({"Category": "Approved versions", "Check": role_name, "Description": "same SHA-256 as the approved version", "Passed": same,
                           "Detail": "" if same else f"approved {approved_record['sha256'][:16]}... ({approved_record['file_name']}), current {(current_record['sha256'][:16] + '... (' + current_record['file_name'] + ')') if current_record else 'not supplied'}"})
        if not same:
            exception_rows.append({"Step": 4, "Category": "Approved versions", "Check": role_name, "Problem": "not the approved version: the regression comparison is only valid against the approved inputs"})
    if not (("raw_data_workbook" in approved_versions) and ("section_mapping_template" in approved_versions)):
        versions_ok = False
        print("  PROBLEM: the approved baseline does not record the raw data and the section-mapping template versions")
    for relative_name in baseline["files"]:
        approved_file = baseline_dir / "files" / relative_name
        current_file = result_files.get(relative_name)
        record = {"File": relative_name, "Approved Rows": "", "Current Rows": "", "Cells Compared": 0, "Cells Different": 0, "Max Abs Difference": 0.0, "Result": "", "Detail": ""}
        if not approved_file.is_file():
            record["Result"], record["Detail"] = "FAIL", "the approved copy of this file is missing from the approved baseline folder"
        elif current_file is None:
            record["Result"], record["Detail"] = "FAIL", "the current run did not produce this file"
        else:
            try:
                approved_frame = pd.read_csv(approved_file, dtype=str, keep_default_na=False)
                current_frame = pd.read_csv(current_file, dtype=str, keep_default_na=False)
                approved_frame = approved_frame.drop(columns=[c for c in IGNORE_COLUMNS if c in approved_frame.columns])
                current_frame = current_frame.drop(columns=[c for c in IGNORE_COLUMNS if c in current_frame.columns])
                record["Approved Rows"], record["Current Rows"] = len(approved_frame), len(current_frame)
                if list(approved_frame.columns) != list(current_frame.columns):
                    only_approved = [c for c in approved_frame.columns if c not in current_frame.columns]
                    only_current = [c for c in current_frame.columns if c not in approved_frame.columns]
                    record["Result"] = "FAIL"
                    record["Detail"] = "columns differ" + (f"; missing now: {only_approved[:6]}" if only_approved else "") + (f"; new: {only_current[:6]}" if only_current else "") + ("" if (only_approved or only_current) else "; same columns in a different order")
                elif len(approved_frame) != len(current_frame):
                    record["Result"], record["Detail"] = "FAIL", f"row count differs: approved {len(approved_frame)}, now {len(current_frame)}"
                else:
                    cells_different, max_abs, file_diff_rows = 0, 0.0, 0
                    for column in approved_frame.columns:

                        current_text = current_frame[column].str.strip()
                        approved_number = pd.to_numeric(approved_text, errors="coerce")
                        current_number = pd.to_numeric(current_text, errors="coerce")
                        both_numeric = approved_number.notna() & current_number.notna()
                        with np.errstate(invalid="ignore"):
                            abs_difference = (current_number - approved_number).abs()
                            within = abs_difference <= (approved_tolerance_abs + approved_tolerance_rel * approved_number.abs())
                        differs = (approved_text != current_text) & ~(both_numeric & within)
                        record["Cells Compared"] += int(len(approved_frame))
                        if both_numeric.any():
                            finite_differences = abs_difference[both_numeric & np.isfinite(abs_difference)]
                            if len(finite_differences) > 0:
                                max_abs = max(max_abs, float(finite_differences.max()))
                        cells_different += int(differs.sum())
                        for row_position in np.where(differs.to_numpy())[0]:
                            if file_diff_rows < MAX_DIFF_ROWS_PER_FILE and len(diff_rows) < MAX_DIFF_ROWS_TOTAL:
                                diff_rows.append({"File": relative_name, "Data Row": int(row_position) + 1, "Column": column, "Approved Value": approved_text.iloc[row_position],
                                                  "Current Value": current_text.iloc[row_position], "Absolute Difference": float(abs_difference.iloc[row_position]) if bool(both_numeric.iloc[row_position]) else np.nan})
                                file_diff_rows += 1
                    record["Cells Different"], record["Max Abs Difference"] = cells_different, round(max_abs, 12)
                    record["Result"] = "PASS" if cells_different == 0 else "FAIL"
                    record["Detail"] = (f"all {record['Cells Compared']} cell(s) within the approved tolerance (rel {approved_tolerance_rel}, abs {approved_tolerance_abs})" if cells_different == 0
                                        else f"{cells_different} of {record['Cells Compared']} cell(s) outside the approved tolerance (rel {approved_tolerance_rel}, abs {approved_tolerance_abs}); largest numeric difference {max_abs:.6g}")
            except Exception as exc:
                record["Result"], record["Detail"] = "FAIL", f"could not be compared: {type(exc).__name__}: {str(exc)[:120]}"
        file_compare_rows.append(record)
        check_rows.append({"Category": "Historical comparison", "Check": relative_name, "Description": "matches the approved result at the approved tolerance", "Passed": record["Result"] == "PASS", "Detail": record["Detail"]})
        if record["Result"] != "PASS":
            exception_rows.append({"Step": 4, "Category": "Historical comparison", "Check": relative_name, "Problem": record["Detail"]})
version_frame = pd.DataFrame(version_rows, columns=["Role", "Approved File", "Current File", "Approved SHA-256", "Current SHA-256", "Result"])
file_compare_frame = pd.DataFrame(file_compare_rows, columns=["File", "Approved Rows", "Current Rows", "Cells Compared", "Cells Different", "Max Abs Difference", "Result", "Detail"])
diff_frame = pd.DataFrame(diff_rows, columns=["File", "Data Row", "Column", "Approved Value", "Current Value", "Absolute Difference"])
print("  (a) Approved versions:")
if len(version_frame) > 0:
    print(version_frame[["Role", "Approved File", "Current File", "Result"]].to_string(index=False))
else:
    print("      " + ("none to compare: this run established the approved baseline" if baseline_established_now else "no approved baseline, so no approved versions to compare with"))
print(f"  (b) Historical comparison at the approved tolerance (rel {approved_tolerance_rel}, abs {approved_tolerance_abs}; ignored columns: {IGNORE_COLUMNS}):")
if len(file_compare_frame) > 0:
    comparison_ok = bool((file_compare_frame["Result"] == "PASS").all())
    print(file_compare_frame[["File", "Approved Rows", "Current Rows", "Cells Different", "Max Abs Difference", "Result"]].to_string(index=False))
    for _, compare_row in file_compare_frame[file_compare_frame["Result"] != "PASS"].iterrows():
        print(f"      PROBLEM {compare_row['File']}: {compare_row['Detail']}")
elif baseline_established_now:
    print("      none to compare: this run established the approved baseline")
else:
    print("      nothing to compare: there is no approved baseline")
step4_pass = bool(baseline_established_now or (baseline is not None and versions_ok and comparison_ok))
record_step_result(4, step4_pass, step_results, TC_NAME, status_text)
#pass rate (reported as Step 6; calculated now so the archive can record it)
checks_frame = pd.DataFrame(check_rows, columns=["Category", "Check", "Description", "Passed", "Detail"])
checks_frame["Passed"] = checks_frame["Passed"].astype(bool)
total_checks = len(checks_frame)
passed_checks = int(checks_frame["Passed"].sum()) if total_checks > 0 else 0
pass_rate = (100.0 * passed_checks / total_checks) if total_checks > 0 else 0.0
category_rates = checks_frame.groupby("Category").agg(Checks=("Passed", "size"), Passed=("Passed", "sum")).reset_index() if total_checks > 0 else pd.DataFrame(columns=["Category", "Checks", "Passed"])
if total_checks > 0:
    category_rates["Failed"] = category_rates["Checks"] - category_rates["Passed"]
    category_rates["Pass Rate %"] = (100.0 * category_rates["Passed"] / category_rates["Checks"]).round(2)
step6_pass = bool((baseline is not None or baseline_established_now) and total_checks > 0 and passed_checks == total_checks)
#---------------------------------------------------------------------
#Step 5: Verify the regression results are archived for auditing: everything this run used and produced is written to a zip archive in a new, uniquely named folder, a SHA-256 manifest

print("\n----- Step 5: Regression results archived for auditing -----")
save_results_csv(checks_frame, run_dir, "PlexOS_3_09_Regression_Checks.csv")
save_results_csv(category_rates, run_dir, "PlexOS_3_09_Category_Pass_Rates.csv")
save_results_csv(pd.DataFrame([{"Test ID": r["Test ID"], "Summary File": r["Summary File"], "Modified": r["Modified"].strftime("%Y-%m-%d %H:%M:%S"), "Successful": r["Successful"], "Why Not Successful": r["Why Not Successful"], "Steps": json.dumps(r["Steps"])} for r in test_runs],
                              columns=["Test ID", "Summary File", "Modified", "Successful", "Why Not Successful", "Steps"]), run_dir, "PlexOS_3_09_Test_Executions.csv")
save_results_csv(defect_frame, run_dir, "PlexOS_3_09_Defect_Checks.csv")

def_user = get_current_user_name()
def_host, def_platform = get_host_and_platform()
run_metadata = {"test_case": TC_NAME, "run_id": run_id, "started": run_started.strftime("%Y-%m-%d %H:%M:%S"), "user": def_user, "host": def_host, "platform": def_platform, "python": sys.version.split()[0],
                "pandas": pd.__version__, "approved_baseline_used": bool(baseline is not None and not baseline_established_now), "baseline_established_this_run": baseline_established_now,
                "approved_baseline_file": str(baseline_json_path), "approved_by": baseline.get("approved_by", "") if baseline else "", "approved_on": baseline.get("approved_on", "") if baseline else "",

                      "regression_pass_rate_percent": round(pass_rate, 4), "checks_total": total_checks, "checks_passed": passed_checks}

for relative_name, source_csv in result_files.items():

archive_dir = archive_root / f"Regression_{run_id}"

member_hashes = {}
try:
    archive_dir.mkdir(parents=True, exist_ok=False)
    with zipfile.ZipFile(archive_zip_path, "w", compression=zipfile.ZIP_DEFLATED) as archive_zip:
        for source_file in sorted(p for p in run_dir.rglob("*") if p.is_file()):
            member_name = "run/" + source_file.relative_to(run_dir).as_posix()
            data_bytes = source_file.read_bytes()
            archive_zip.writestr(member_name, data_bytes)
            member_hashes[member_name] = (hashlib.sha256(data_bytes).hexdigest(), len(data_bytes))
        for member_name, source_file in supporting_files.items():
            data_bytes = source_file.read_bytes()
            archive_zip.writestr(member_name, data_bytes)
            member_hashes[member_name] = (hashlib.sha256(data_bytes).hexdigest(), len(data_bytes))
    save_results_csv(pd.DataFrame([{"Archive Member": name, "Bytes": size, "SHA-256": digest} for name, (digest, size) in member_hashes.items()]), archive_dir, archive_manifest_path.name)
except Exception as exc:
    archive_problem = f"{type(exc).__name__}: {exc}"
step5_checks = {"Archive location is new (an earlier archive is never overwritten)": archive_problem == "" and archive_zip_path.is_file()}
verify_ok, verify_count, verify_bad = False, 0, []
if archive_zip_path.is_file():
    with zipfile.ZipFile(archive_zip_path, "r") as archive_read:
        corrupt_member = archive_read.testzip()
        for member_name, (expected_digest, expected_size) in member_hashes.items():
            try:
                read_bytes = archive_read.read(member_name)
            except KeyError:
                verify_bad.append(member_name + " (missing from the archive)")
                continue
            verify_count += 1
            if hashlib.sha256(read_bytes).hexdigest() != expected_digest or len(read_bytes) != expected_size:
                verify_bad.append(member_name + " (hash differs)")
    verify_ok = corrupt_member is None and len(verify_bad) == 0 and verify_count == len(member_hashes) and len(member_hashes) > 0
required_members = ["run/PlexOS_3_09_Run_Metadata.json", "run/PlexOS_3_09_Run_Status.json", "run/PlexOS_3_09_Regression_Checks.csv", "run/PlexOS_3_09_Version_Checks.csv", "run/PlexOS_3_09_Test_Executions.csv"]
step5_checks[f"Archive written ({archive_zip_path.name}, {len(member_hashes)} file(s))"] = archive_zip_path.is_file() and len(member_hashes) > 0
step5_checks[f"Archive read back and every file verified against its SHA-256 ({verify_count} of {len(member_hashes)})"] = verify_ok
step5_checks["Archive holds the run metadata, status, regression checks, version record and test executions"] = all(name in member_hashes for name in required_members)

            for cell in row:
                cell_text = cell.strip()
                if cell_text:
                    first_cell_text = cell_text
                    break
            if not first_cell_text.startswith("("):
                continue                    #a section header always starts with "(": every other row is skipped without building its joined text
            row_text = " ".join(str(cell).strip() for cell in row if str(cell).strip())
            header_match = header_re.match(row_text)

                key = header_match.group(1).lower()

                    standard_duplicates.append(key)
                else:

if ratings_report_path.is_file():

                cell_text = cell.strip()
                if cell_text:
                    non_blank_count += 1
                    if non_blank_count > 1:
                        break               #a section header has exactly one non-blank cell: a second one ends the examination of the row
                    only_text = cell_text
            if non_blank_count == 1:
                key_text = re.sub(r"\([^)]*\)", "", only_text.lower().strip())
                key_text = re.sub(rf"\b{day_token}\s*-\s*{day_token}\b", "", key_text)

    header = blueprint_row["Header"]
    header_match = re.match(r"^\(\s*(\w+)\s*\)\s*(.*)$", header)
    if not header_match:
        continue
    section_key = header_match.group(1).lower()
    entry = section_mapping.setdefault(section_key, {"title": header, "mapped": False, "reasons": []})
    class_text, group_text, property_text = blueprint_row["Class"], blueprint_row["Group"], blueprint_row["Property"]
    pattern_text, unit_text = blueprint_row["Pattern"].lower(), blueprint_row["Unit"]
    class_l, group_l = class_text.lower(), group_text.lower()
    reason = ""
    if section_key in SKIP_TABLES:
        reason = f"custom section: {SKIP_TABLES[section_key]}"
    elif pattern_text.startswith("daily"):
        reason = "daily pattern: the raw data is monthly only"
    else:
        effective_class, effective_property = class_l, property_text
        if (section_key, class_l, property_text) in CUSTOM_OVERRIDES:
            effective_class, effective_property = CUSTOM_OVERRIDES[(section_key, class_l, property_text)]
        if effective_class not in CLASS_TO_RAW:
            reason = f"class '{class_text}' is not in the raw data"
        else:
            sheet_name, collection_name, identifier_column = CLASS_TO_RAW[effective_class]
            if group_l not in ("all", "") and (class_l, group_l) not in known_groups and ("generator", group_l) not in known_groups:
                reason = f"group '{group_text}' is not defined on the Groups sheet"
            elif (effective_class, effective_property) in NO_RAW_PROPERTIES:
                reason = f"no raw measure for '{effective_property}'"
            else:
                raw_column = PROPERTY_TO_RAW_COLUMN.get((sheet_name, collection_name), {}).get(effective_property)
                if raw_column is None:
                    reason = f"no raw column for '{effective_property}' on {collection_name}"
                else:
                    needed_columns = [raw_column]               #the raw column itself, including Total System Cost ($000) for section 5 (never rebuilt from cost components); the frames hold the All Periods rows only
                    columns_ok = True
                    for needed in needed_columns:
                        cache_key = (sheet_name, collection_name, needed)
                        if cache_key not in raw_column_cache:
                            subset_key = (sheet_name, collection_name)

                            subset = raw_subset_cache[subset_key]

                        columns_ok = columns_ok and raw_column_cache[cache_key]
                    if not columns_ok:
                        reason = f"the raw column(s) {needed_columns} hold no data for {collection_name} in the '{STANDARD_TIMESLICE}' rows"
                    else:
                        unit_match = re.search(r"\(([^)]*)\)\s*$", raw_column)
                        raw_unit = unit_match.group(1).strip().lower() if unit_match else "-"
                        raw_unit = "hours" if raw_unit == "h" else raw_unit
                        header_unit = re.search(r"\(([^()]*)\)\s*$", header.strip())
                        target = unit_text if unit_text != "" else (header_unit.group(1).strip() if header_unit else raw_unit)
                        target_l = target.lower().strip()
                        if target_l.startswith("1000·"):
                            target_l = target_l[len("1000·"):]
                        target_l = "hours" if target_l == "h" else target_l
                        if target_l != raw_unit and (raw_unit, target_l) not in unit_rules:
                            reason = f"UnitConversion has no rule {raw_unit} -> {target}"
    if reason == "":
        entry["mapped"] = True
    elif entry["reasons"].count(reason) == 0:
        entry["reasons"].append(reason)

    for step_key, step_value in run["Steps"].items():
        if step_value == "FAIL":
            check_text = run["Step Checks"].get(step_key, "")
            failed_validations.append({"Category": "SIT test step", "Test ID": run["Test ID"], "Step": step_key, "Section": "", "Problem": f"step {step_key} failed" + (f": {check_text}" if check_text else "")})
for _, section_row in failed_sections.iterrows():
    failed_validations.append({"Category": "Section", "Test ID": section_row["Note"].replace("failed in: ", "") if section_row["Note"].startswith("failed in: ") else "", "Step": "",
                               "Section": f"{section_row['Report']} / {section_row['Section']}", "Problem": section_row["Note"] or "failed validation"})
if COVERAGE_GAPS_FAIL:
    for _, gap_row in coverage_gaps.iterrows():
        failed_validations.append({"Category": "Coverage gap", "Test ID": "", "Step": "", "Section": f"{gap_row['Report']} / {gap_row['Section']}", "Problem": gap_row["Note"]})
print("\n----- Step 4: Exception report -----")
exception_columns = ["Exception Id", "Category", "Test ID", "Step", "Section", "Problem"]
exception_df = pd.DataFrame([{"Exception Id": f"EX-{i + 1:04d}", **item} for i, item in enumerate(failed_validations)], columns=exception_columns)
exception_path = save_results_csv(exception_df, run_dir, "PlexOS_3_10_Exception_Report.csv")
exception_detail_path = save_results_csv(pd.DataFrame(exception_detail_rows, columns=["Test ID", "Source File", "Row", "Report", "Section", "Detail"]), run_dir, "PlexOS_3_10_Exception_Detail.csv")
step4_checks = {}
step4_checks["Exception report written (also when there are no failures)"] = exception_path.is_file()
try:
    readback_exceptions = pd.read_csv(exception_path, dtype=str, keep_default_na=False)
except pd.errors.EmptyDataError:
    readback_exceptions = pd.DataFrame(columns=exception_columns)
step4_checks[f"Read back: every failed validation is in the file ({len(readback_exceptions)} of {len(failed_validations)})"] = len(readback_exceptions) == len(failed_validations)
#independent re-check from the raw results: every failed test case, failed step, failed section and unusable report must appear
missing_from_report = []
for run in test_runs:
    if not run["Successful"] or run in stale_runs:
        if len(readback_exceptions[(readback_exceptions["Test ID"] == run["Test ID"]) & (readback_exceptions["Category"].isin(["SIT test case", "Automated execution"]))]) == 0:
            missing_from_report.append(f"test case {run['Test ID']}")
    for step_key, step_value in run["Steps"].items():
        if step_value == "FAIL":
            if len(readback_exceptions[(readback_exceptions["Test ID"] == run["Test ID"]) & (readback_exceptions["Step"] == step_key)]) == 0:
                missing_from_report.append(f"{run['Test ID']} step {step_key}")
for _, section_row in audit_df[audit_df["Audit Status"] == "FAIL"].iterrows():
    if len(readback_exceptions[readback_exceptions["Section"] == f"{section_row['Report']} / {section_row['Section']}"]) == 0:
        missing_from_report.append(f"section {section_row['Report']} / {section_row['Section']}")
if (not reports_ok) and len(readback_exceptions[readback_exceptions["Category"] == "Report datasets"]) == 0:
    missing_from_report.append("the missing report dataset(s)")
step4_checks["Independent re-check: no failed test case, test step, section or report problem is missing from the report"] = len(missing_from_report) == 0
step4_checks["Every exception names its category and the failed item"] = bool(len(readback_exceptions) == 0 or ((readback_exceptions["Category"].str.strip() != "").all() and (readback_exceptions["Problem"].str.strip() != "").all()))
step4_checks[f"Detail file written ({len(exception_detail_rows)} row(s) from the test cases' own exception files)"] = exception_detail_path.is_file()
print(f"  Failed validations found: {len(failed_validations)} ({', '.join(f'{k} {int(v)}' for k, v in exception_df['Category'].value_counts().items()) if len(exception_df) else 'none'})")
print_status_checks(step4_checks, status_text)
for item in missing_from_report[:10]:
    print(f"  PROBLEM: not in the exception report: {item}")
for _, exception_row in exception_df.head(20).iterrows():
    print(f"    {exception_row['Exception Id']} [{exception_row['Category']}] {exception_row['Test ID']} {('step ' + exception_row['Step']) if exception_row['Step'] else ''} {exception_row['Section']}: {exception_row['Problem'][:110]}")
if len(exception_df) > 20:
    print(f"    ... and {len(exception_df) - 20} more (see {exception_path.name})")
record_step_result(4, all(step4_checks.values()), step_results, TC_NAME, status_text)
#---------------------------------------------------------------------
#Step 5: Verify the test execution status is returned as Pass or Fail. The execution status is PASS only when the reports exist, every test case has fresh results and passed, every
#test step passed, no section failed or is missing and no coverage gap exists. It is recomputed independently, written to a status file and the JSON, read back, printed and returned as
#the process exit code (0 = PASS, 1 = FAIL). A run with failed validations returns FAIL and still proves the automation works.
#---------------------------------------------------------------------
print("\n----- Step 5: Test execution status returned as Pass or Fail -----")
report_hashes_after = {report_label: hash_file_sha256(report_path) for report_label, report_path in report_files if report_path.is_file()}
reports_unchanged = bool(len(report_hashes_before) == 2 and report_hashes_before == report_hashes_after)
if reports_ok and not reports_unchanged:
    failed_validations.append({"Category": "Automated execution", "Test ID": "", "Step": "", "Section": "", "Problem": "the Standard / Ratings reports changed while this test case was running, so the audit cannot be trusted"})
    exception_df = pd.DataFrame([{"Exception Id": f"EX-{i + 1:04d}", **item} for i, item in enumerate(failed_validations)], columns=exception_columns)
    save_results_csv(exception_df, run_dir, "PlexOS_3_10_Exception_Report.csv")
execution_status_from_exceptions = "PASS" if len(failed_validations) == 0 else "FAIL"
independent_failure = (not reports_ok) or len(test_runs) == 0 or any(not r["Successful"] for r in test_runs) or len(stale_runs) > 0 or any(v == "FAIL" for r in test_runs for v in r["Steps"].values()) \
    or (len(failed_sections) > 0) or (COVERAGE_GAPS_FAIL and len(coverage_gaps) > 0) or (len(set(standard_duplicates)) > 0) or (reports_ok and not reports_unchanged)
execution_status_independent = "FAIL" if independent_failure else "PASS"
status_file = run_dir / "PlexOS_3_10_Status.txt"
status_file.write_text(execution_status_from_exceptions + "\n", encoding="utf-8")
status_json_path = write_json_file(run_dir / "PlexOS_3_10_Status.json", {"test_case": TC_NAME, "run_id": run_id, "execution_status": execution_status_from_exceptions, "failed_validations": len(failed_validations),
                                                                         "exit_code": 0 if execution_status_from_exceptions == "PASS" else 1})
read_status = status_file.read_text(encoding="utf-8").strip()
read_json_status = json.loads(status_json_path.read_text(encoding="utf-8"))["execution_status"]
exit_code_map = {"PASS": 0, "FAIL": 1}
step5_checks = {}
step5_checks[f"The status is exactly Pass or Fail ('{execution_status_from_exceptions}')"] = execution_status_from_exceptions in ("PASS", "FAIL")
step5_checks["The status agrees with the independent recomputation from the raw run results"] = execution_status_from_exceptions == execution_status_independent
step5_checks["Status file and JSON written and read back with the same status"] = read_status == execution_status_from_exceptions and read_json_status == execution_status_from_exceptions
step5_checks[f"Exit code mapping defined (PASS -> {exit_code_map['PASS']}, FAIL -> {exit_code_map['FAIL']}); this run will exit with {exit_code_map[execution_status_from_exceptions]}"] = set(exit_code_map) == {"PASS", "FAIL"}
step5_checks["The Standard and Ratings reports were unchanged during the run (SHA-256 before and after)"] = reports_unchanged
print_status_checks(step5_checks, status_text)
print(f"  TEST EXECUTION STATUS: {'Pass' if execution_status_from_exceptions == 'PASS' else 'Fail'}" + ("" if execution_status_from_exceptions == "PASS" else f" ({len(failed_validations)} failed validation(s); see the exception report)"))
record_step_result(5, all(step5_checks.values()), step_results, TC_NAME, status_text)
#---------------------------------------------------------------------
#Step 6: Verify the validation results are stored in a machine-readable format: a JSON document with the whole run (steps, status, test cases, audit, exceptions, result-file
#inventory with SHA-256) plus CSV files for the audit, the exceptions, the test executions and an index of every result file. Each is read back and checked.
#---------------------------------------------------------------------
print("\n----- Step 6: Results stored in a machine-readable format -----")
executions_df = pd.DataFrame([{"Test ID": r["Test ID"], "Summary File": r["Summary File"], "Results Written": r["Modified"].strftime("%Y-%m-%d %H:%M:%S") if r["Modified"] is not None else "", "Newer Than Reports": r.get("Newer Than Reports", ""),
                               "Complete": r["Complete"], "Successful": r["Successful"], "Why Not Successful": r["Why Not Successful"], "Steps": json.dumps(r["Steps"])} for r in test_runs],
                             columns=["Test ID", "Summary File", "Results Written", "Newer Than Reports", "Complete", "Successful", "Why Not Successful", "Steps"])

index_path = save_results_csv(pd.DataFrame(files_inventory, columns=["Test ID", "File", "Bytes", "SHA-256"]), run_dir, "PlexOS_3_10_Results_Index.csv")
provisional_steps = {str(k): status_text[v] for k, v in step_results.items()}
provisional_steps["6"] = "IN PROGRESS"
result_document = {"test_case": TC_NAME, "run_id": run_id, "started": run_started.strftime("%Y-%m-%d %H:%M:%S"), "execution_status": execution_status_from_exceptions,
                   "steps": provisional_steps, "reports": {"standard": str(standard_report_path), "ratings": str(ratings_report_path), "sha256": report_hashes_before, "generated": report_times},
                   "expected_standard_sections": list(EXPECTED_SECTIONS.keys()),
                   "counts": {"standard_sections": len(standard_audit), "rating_periods": len(rating_audit), "failed_validations": len(failed_validations), "test_cases": len(test_runs)},
                   "test_cases": [{"test_id": r["Test ID"], "successful": r["Successful"], "complete": r["Complete"], "steps": r["Steps"], "results_written": r["Modified"].strftime("%Y-%m-%d %H:%M:%S") if r["Modified"] is not None else ""} for r in test_runs],
                   "audit": audit_df.to_dict(orient="records"), "exceptions": exception_df.to_dict(orient="records"), "result_files": files_inventory}

step6_checks = {}
parsed_json = None
try:
    parsed_json = json.loads(json_path.read_text(encoding="utf-8"))
except Exception as exc:
    step6_checks[f"JSON results parse: {type(exc).__name__}"] = False
if parsed_json is not None:
    required_keys = ["test_case", "execution_status", "steps", "test_cases", "audit", "exceptions", "counts", "result_files"]
    step6_checks["JSON results written and parsed back with every required section"] = all(k in parsed_json for k in required_keys)
    step6_checks[f"JSON counts agree with the CSV files (audit {len(parsed_json['audit'])} of {len(audit_df)}, exceptions {len(parsed_json['exceptions'])} of {len(exception_df)})"] = len(parsed_json["audit"]) == len(audit_df) and len(parsed_json["exceptions"]) == len(exception_df)

step_summary = pd.DataFrame([

write_json_file(status_json_path, {"test_case": TC_NAME, "run_id": run_id, "execution_status": execution_status_from_exceptions, "test_case_verdict": "PASS" if test_case_pass else "FAIL",
                                   "failed_validations": len(failed_validations), "exit_code": returned_exit_code})
for copy_name in ("PlexOS_3_10_Audit_Report.csv", "PlexOS_3_10_Exception_Report.csv", "PlexOS_3_10_Exception_Detail.csv", "PlexOS_3_10_Results.json", "PlexOS_3_10_Status.json",

```python
#PlexOS-4.01 Standard Report Section Mapping Validation   +   PlexOS-4.04 Ratings Report Section Mapping Validation
TEST_CASES = {"std": ("PlexOS-4.01", "Standard Report Section Mapping Validation"), "rat": ("PlexOS-4.04", "Ratings Report Section Mapping Validation")}
status_text = {True: "PASS", False: "FAIL"}
require_global_names(("plexos4_inputs", "plexos4_results_folder", "RUN_REPORTS"), "PlexOS-4.01 / PlexOS-4.04", "the PLEXOS-4 INPUTS block")
FILE_PATHS = plexos4_inputs                      #the answers above, by name
FILE_NAMES = {"std_P": "PLEXOS Standard Report (the file that gets its first column changed)", "std_C": "RT Sim Standard Report (KeyedData baseline: the labels to copy)",
              "rat_P": "PLEXOS Ratings Report (the file that gets its first column changed)", "rat_C": "RT Sim Ratings Report (KeyedData baseline: the labels to copy)"}

#1. SETTINGS
OUTPUT_FOLDER = str(plexos4_results_folder)               #the results folder answered above: every output of every PlexOS-4 test is written here
OUTPUT_NAMES = {"std": "Standard_Report_PLEXOS_RTSimLabels_mon.csv",      #THE ONE OUTPUT CSV of the Standard Report
                "rat": "Ratings_Report_PLEXOS_RTSimLabels_rat.csv"}       #THE ONE OUTPUT CSV of the Ratings Report
VERIFY_NAME = "Label_Verification_Standard_and_Ratings.xlsx"              #the verification workbook (both reports)

#Blank lines written at the very top of each output file, BEFORE the first section ("( 1 ) System Summary" for the Standard report).
#The OpRep_01 macro only reads the values after the 27th row into KeyedData, so the Standard report gets 27.
#The Ratings report is not changed (0). Set a report to 0 to add nothing to it.
TOP_BLANK_LINES = {"std": 27, "rat": 0}

TABLES_TO_CONVERT = []           #Standard: [] = every table. Or a list of table numbers, e.g. ["1", "2", "3"]
RATING_SECTIONS_TO_CONVERT = []  #Ratings: [] = every section. Or a list of section names (PLEXOS or RT Sim spelling), e.g. ["All Periods", "ATC Mon-Sun"]
TABLES_TO_VERIFY = ["1", "2", "3", "5", "6", "7", "8", "10", "11", "21", "22", "23", "26", "31", "33c", "34c", "56a", "57a", "80", "87"]   #Standard: the sections the test case validates ([] = every table)
RATING_SECTIONS_TO_VERIFY = []   #Ratings: [] = every section

CHANGE_TABLE_TITLES = True       #' ( 1) System Summary' instead of '( 1 ) System Summary'; Ratings: ' ATC Mon-Sun' instead of 'All Periods'
CHANGE_GROUP_HEADINGS = True     #effluent type / contract name heading lines
CHANGE_HEADER_FIRST_CELL = True  #the first cell of every month-header row (RT Sim: one space)
AUTO_DETECT_ROLES = True         #swap P and C automatically when the file given as PLEXOS looks like an RT Sim file and the other way round
VERIFY_OTHER_CELLS_UNCHANGED = True   #every cell outside the changed label cells is identical (counts only; no individual value is reported)
FUZZY_THRESHOLD = 0.94           #same rule as the earlier conversion script
MAX_RT_ONLY_ROWS_IN_WORKBOOK = 30000   #the 'RT Sim only' sheets stop at this many rows per report (the Ratings report has thousands)

#Section-mapping test settings (PlexOS-4.01 / PlexOS-4.04)
REQUIRE_SAME_COLUMN_COUNT = True      #a KeyedData structure is only valid when its month-header row has the same number of value columns as the PLEXOS section (the macro reads the values by position)
TITLE_SIMILARITY_MIN = 0.85           #Standard: a section's title may differ from the KeyedData title (units / wording) as long as the two are at least this similar (1.0 = identical after normalisation)
RATING_APPROVED_METHODS = ("exact name", "alias (rating_section_aliases)", "name prefix", "alias + name prefix")      #Ratings: a section is mapped to an APPROVED KeyedData category by these methods; an approximate name similarity needs a human review and fails

#2. MAPPING TABLES (from your original code)
alias_map = {"microgrid tsmc": "tsmc microgrid",
             "microgrid xhlf": "xhlf microgrid 1",
             "agave pvs solar": "agave pvs sol",
             "ironwood pvs-solar": "ironwood pvs sol",
             "pioneer pvs-solar": "pioneer pvs sol",
             "pioneer pvs-bess": "pioneer pvs bat",
             "co bar c pvs-solar": "co bar c pvs sol",
             "co bar c pvs-bess": "co bar c pvs bat",
             "total effluents lb so2": "total effluents lb sox",
             "total effluents lb nox": "total effluents lb nox",
             "total effluents lb co2": "total effluents lb cox",
             "system native load": "system native energy",
             "shortfall and spin violations": "shortfall spin violations",
             "thermal unit starts": "thermal unit starts",
             "thermal unit hours online hours": "thermal unit hours online",
             "thermal unit capacity factor": "thermal unit capacity factor",
             "thermal comb cycle ct starts": "thermal comb cycle ct starts",
             "comb cycle mode hours online hours": "comb cycle mode hours online",
             "supplemental dump value": "supplemental dump value",
             "buy fix 7x22 (6d)": "buy fix 7-22 (6d)"}

#EXPLICIT OVERRIDES
#(table number, PLEXOS label)
#RT Sim label   (Standard report only)
#DOUBLE CHECK THESE WITH JIM
explicit_overrides = {}
for area_word in ["Zone", "Region"]:
    explicit_overrides[("1", f"{area_word}-all-Native Load-Total")] = "System Native Energy (MWH)"
    explicit_overrides[("1", f"{area_word}-all-Load-Total")] = "Total Net DLC Load (MWH)"
    explicit_overrides[("1", f"{area_word}-all-Dump Energy-Total")] = "Dump Energy (MWH)"
    explicit_overrides[("1", f"{area_word}-all-Cost of Dump Energy-Total")] = "Supplemental Dump Value ($)"
    explicit_overrides[("30", f"{area_word}-all-Dump Energy Hours-Total")] = "Dump hours"
    explicit_overrides[("30", f"{area_word}-all-Dump Energy-Total")] = "Dump energy (MWH)"
    explicit_overrides[("31", f"{area_word}-all-Native Load-Total")] = "System native energy (MWH)"
    explicit_overrides[("31", f"{area_word}-all-Load-Total")] = "System Native Peak Demand (MW)"
explicit_overrides[("1", "Generator-CC-CT-ST_Coal-Oil-UR-Generation-Total")] = "Thermal Generation (MWH)"
explicit_overrides[("1", "Battery-all-Generation-Total")] = "Hydro/Battery Discharge (MWH)"
explicit_overrides[("1", "Generator-wind-pv_sun-Generation-Total")] = "Wind/Solar Generation (MWH)"
explicit_overrides[("1", "Reserve-all-Shortage-Total")] = "Shortfall Energy (MWH)"
explicit_overrides[("1", "Battery-all-Load-Total")] = "Hydro/Battery Charge (MWH)"
explicit_overrides[("1", "Generator-CC-CT-ST_Coal-Oil-UR-FO&M Cost-Total")] = "Thermal Fixed+Capital Cost ($)"
explicit_overrides[("1", "Generator-CC-CT-ST_Coal-Oil-UR-VO&M Cost-Total")] = "Thermal Variable Cost ($)"
explicit_overrides[("1", "Generator-CC-CT-ST_Coal-Oil-UR-Total System Cost ($000)-Total")] = "Thermal Total Cost ($)"
explicit_overrides[("1", "Generator-wind-pv_sun-Total System Cost ($000)-Total")] = "Wind/Solar Total Costs ($)"
explicit_overrides[("1", "Reserve-all-Shortage Cost-Total")] = "Shortfall Energy Cost ($)"
explicit_overrides[("29", "Reserve-all-Shortage-Total")] = "Shortfall Energy (MWH)"
explicit_overrides[("29", "Reserve-all-Shortage Hours-Total")] = "Shortfall Hours"
explicit_overrides[("29", "Reserve-all-Shortage Cost-Total")] = "Shortfall Energy Cost ($)"
explicit_overrides[("102", "APS")] = "Supplemental Dump Value ($)"

#Effluent name spelling map (PLEXOS -> RT Sim), compared on the text after " -- "
effluent_spelling = {"so2": "sox", "nox": "nox", "co2": "cox"}   #CO, VOC, HG, PM10 are already identical

#Ratings: PLEXOS section names that are the same section as an RT Sim section under another name (compared with letters / digits only, lower case)
rating_section_aliases = {"allperiods": "atc"}

#Table types in the Standard report CSV
sub_row_tables = ["3", "6"]                        #plant row + indented fuel rows
effluent_group_tables = ["14", "73"]               #effluent type heading + plant rows
day_tables = ["27", "28"]                          #contract heading + 31 day rows
effluent_total_table = "32"                        #effluent type heading + one total row
contract_tables = ["23", "24", "25", "26", "27", "28"]   #units in parentheses differ between the two tools
heading_tables = effluent_group_tables + day_tables + [effluent_total_table]

norm_steps = [(r"_", " "), (r"&", " and "), (r"[-]+", " "), (r"[():,$/]+", " "), (r"\s+", " "),
              (r"\b1000\s+mmbtu\b", "mmbtu"), (r"\blbs\b", "lb"), (r"\bhours\b", ""), (r"\s+", " ")]
month_re = re.compile(r"^\s*(?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec|Tot)-\d{4}\s*$", re.IGNORECASE)
title_re = re.compile(r"^\s*\(\s*(\d+[a-z]?)\s*\)\s*(.*)$", re.IGNORECASE)
non_value_columns = ["day_id", "index"]

#the alias table, normalized the same way every label is normalized (normalize_label with norm_steps; the shared normalize_text of functions_and_imports.py is no longer replaced by a lambda here)
alias_norm = {normalize_label(alias_key, norm_steps): normalize_label(alias_value, norm_steps) for alias_key, alias_value in alias_map.items()}
override_raw = {(table_key, normalize_label(label_text, norm_steps)): override_value for (table_key, label_text), override_value in explicit_overrides.items()}
title_key_full = lambda text: alias_label(text, norm_steps, alias_norm)
title_key_plain = lambda text: alias_label(text, norm_steps, alias_norm, True)

#result of every label, as it is called in the console and in the workbook
ST_SAME = "Already identical"
ST_FORMAT = "Changed: padding / spacing / case / punctuation"
ST_UNITS = "Changed: units text"
ST_RENAMED = "Changed: renamed / different text"
ST_KEPT = "Kept: no RT Sim row"
ST_NOTABLE = "Kept: table / section not in RT Sim"
ST_OFF = "Not changed: option off / not selected"
status_columns = [ST_SAME, ST_FORMAT, ST_UNITS, ST_RENAMED, ST_KEPT, ST_NOTABLE, ST_OFF]


#3. THE VERIFICATION WORKBOOK (created here, filled report by report)

fill_header = PatternFill("solid", fgColor="1F3864")
fill_same = PatternFill("solid", fgColor="C6EFCE")
fill_format = PatternFill("solid", fgColor="DDEBF7")
fill_units = PatternFill("solid", fgColor="FFF2CC")
fill_renamed = PatternFill("solid", fgColor="F8CBAD")
fill_kept = PatternFill("solid", fgColor="D9D9D9")
fill_bad = PatternFill("solid", fgColor="FF9999")
status_fill = {ST_SAME: fill_same, ST_FORMAT: fill_format, ST_UNITS: fill_units, ST_RENAMED: fill_renamed, ST_KEPT: fill_kept, ST_NOTABLE: fill_kept, ST_OFF: fill_kept}
font_header = Font(bold=True, color="FFFFFF")

wb = openpyxl.Workbook()
summary_ws = wb.active
summary_ws.title = "Summary"
summary_ws.append(["PLEXOS Standard and Ratings Reports: sections mapped to KeyedData, first column changed to the RT Sim labels, and verified (PlexOS-4.01 / PlexOS-4.04)"])
summary_ws["A1"].font = Font(bold=True, size=13)
summary_ws.append(["Only the first column (label column) was changed in each report; no number and no other column was changed. The only lines added are the blank lines at the top (TOP_BLANK_LINES)."])
formats_rows = []                          #rows of the 'Label Formats' sheet, collected report by report
all_reports_pass = OrderedDict()
test_step_tables = OrderedDict()           #test case -> list of (step, check, status), for the final summary
output_paths = OrderedDict()
verify_folder = ""
overall_labels_rows = OrderedDict()
verify_path = os.path.join(OUTPUT_FOLDER, VERIFY_NAME)
prepare_results_folder(OUTPUT_FOLDER)


#4. THE REPORT LOOP: every step below is done to the Standard report (PlexOS-4.01) and then to the Ratings report (PlexOS-4.04)

for report in RUN_REPORTS:
    is_ratings = report == "rat"
    report_label = "Ratings" if is_ratings else "Standard"
    unit_word = "section" if is_ratings else "table"
    TC_ID, TC_TITLE = TEST_CASES[report]
    TC_NAME = f"{TC_ID} {TC_TITLE}"
    top_blank = int(TOP_BLANK_LINES.get(report, 0))          #blank lines added at the top of this report's output
    step_results = {}                                        #step number -> True / False, for THIS test case
    if is_ratings:
        #the Ratings report has no numbered tables, so none of the Standard-only table rules apply
        r_sub = []
        r_effgroup = []
        r_day = []
        r_efftotal = "~none~"
        r_contract = []
        r_heading = []
        convert_names = RATING_SECTIONS_TO_CONVERT
        verify_names = RATING_SECTIONS_TO_VERIFY
    else:
        r_sub = sub_row_tables
        r_effgroup = effluent_group_tables
        r_day = day_tables
        r_efftotal = effluent_total_table
        r_contract = contract_tables
        r_heading = heading_tables
        convert_names = TABLES_TO_CONVERT
        verify_names = TABLES_TO_VERIFY
    convert_set = lowercase_set_or_none(convert_names)
    verify_set = lowercase_set_or_none(verify_names)
    print_banner(f"#{TC_NAME.upper()}", "#", blank_line=True)

    #=================================================================
    #4.1 THE TWO INPUT FILES (from the inputs block), CHECK WHICH IS WHICH
    #=================================================================
    p_path = plexos4_inputs[report + "_P"]
    c_path = plexos4_inputs[report + "_C"]
    same_input_file = os.path.abspath(p_path) == os.path.abspath(c_path)
    p_rows = read_csv_rows(p_path)
    c_rows = read_csv_rows(c_path)
    #---- which file is the PLEXOS export and which one is the RT Sim file? PLEXOS numbers are written with full float precision,
    #PLEXOS Standard titles are '( 1 )', PLEXOS labels look like 'Generator-...' / '...-all-...', PLEXOS rating sections are 'All Periods' / 'Off-Peak' / 'On-Peak';
    #RT Sim numbers have 4 decimals, titles ' ( 1)', rating sections 'ATC Mon-Sun' / '... Mon-Sat' / 'Sunday On-peak' ----
    role_score = {}
    for role_key, role_rows in [("P", p_rows), ("C", c_rows)]:
        plexos_votes = 0
        rtsim_votes = 0
        long_numbers = 0
        all_numbers = 0
        for cells in role_rows:
            first = cells[0] if len(cells) > 0 else ""
            if re.match(r"^\(\s\d+[a-z]?\s\)", first):
                plexos_votes += 1
            elif re.match(r"^\s+\(\s*\d+[a-z]?\)", first) and "=" not in first:
                rtsim_votes += 1
            if "-all-" in first or first.startswith("Generator-"):
                plexos_votes += 1
            if len([c for c in cells if c.strip() != ""]) == 1:
                section_text = first.strip().lower()
                if re.fullmatch(r"all periods|off-?peak|on-?peak", section_text):
                    plexos_votes += 1
                elif re.search(r"mon-sun|mon-sat|sunday on-peak", section_text):
                    rtsim_votes += 1
            if all_numbers < 30000:
                for c in cells[1:]:
                    if re.fullmatch(r"-?\d+\.\d+", c.strip()):
                        all_numbers += 1
                        if len(c.strip().split(".")[1]) >= 10:
                            long_numbers += 1
        precision = long_numbers / all_numbers if all_numbers else 0.0
        vote_ratio = (plexos_votes - rtsim_votes) / max(1, plexos_votes + rtsim_votes)
        role_score[role_key] = {"precision": precision, "votes": vote_ratio, "score": precision * 100 + vote_ratio * 10}
    print_banner(f"{report_label.upper()}: INPUT FILES")
    for role_key, role_path in [("P", p_path), ("C", c_path)]:
        print(f"{role_key} = {FILE_NAMES[report + '_' + role_key]}")
        print(f"      {role_path}")
        print(f"      PLEXOS-likeness: numbers with 10+ decimals {role_score[role_key]['precision']:.0%}, title / label / section style vote {role_score[role_key]['votes']:+.2f}")
    roles_swapped = False
    if AUTO_DETECT_ROLES and role_score["C"]["score"] > role_score["P"]["score"] + 3:
        print("NOTE: the file given as the PLEXOS report looks like an RT Sim file and the file given as the RT Sim report looks like the PLEXOS export")
        print("      (full-precision numbers / '( 1 )' titles / 'Generator-...' names / 'All Periods' sections). The two were SWAPPED, so the PLEXOS export is the one that gets its labels changed.")
        p_path, c_path = c_path, p_path
        p_rows, c_rows = c_rows, p_rows
        roles_swapped = True
        print(f"      PLEXOS file used : {p_path}")
        print(f"      RT Sim file used : {c_path}")
    with open(p_path, "rb") as f:
        raw_head = f.read(200000)
    has_bom = raw_head.startswith(b"\xef\xbb\xbf")
    line_end = "\r\n" if b"\r\n" in raw_head else "\n"

    #blank lines the PLEXOS file already has at its top (counted so that the check below knows on which line the first section has to end up)
    input_leading_blank = count_leading_blank_rows(p_rows)
    if top_blank > 0 and input_leading_blank > 0:
        print(f"NOTE: the PLEXOS {report_label} file already starts with {input_leading_blank} blank line(s); {top_blank} more are added, so {top_blank + input_leading_blank} blank lines come before the first {unit_word}.")

    output_folder = OUTPUT_FOLDER.strip().strip('"')
    if output_folder == "":
        output_folder = os.path.dirname(p_path) or "."
    prepare_results_folder(output_folder)
    verify_folder = output_folder
    output_path = os.path.join(output_folder, OUTPUT_NAMES[report])
    output_paths[report] = output_path
    if os.path.abspath(output_path) in [os.path.abspath(p_path), os.path.abspath(c_path)]:
        raise ValueError("The output file would overwrite one of the input files. Change the results folder (input 5) or OUTPUT_NAMES.")

    #=================================================================
    #4.2 THE TWO PASSES
    #pass 1: PLEXOS file + RT Sim (KeyedData) file -> plan every label change -> write THE ONE OUTPUT CSV (with the blank lines at the top)
    #pass 2: the OUTPUT CSV + RT Sim file, run through exactly the same matching -> proves the output equals RT Sim and a re-run changes nothing
    #(pass 1 line numbers are PLEXOS-file lines; pass 2 line numbers are OUTPUT-file lines = PLEXOS line + top_blank)
    #=================================================================
    parsed_rt = None
    converted_names = set()
    verified_names = set()
    cands_by_pass = {}
    rt_only_by_pass = {}
    rt_only_tables_by_pass = {}
    section_maps = {}
    section_methods = {}
    px_tables_by_pass = {}
    backup_rows = {}
    for pass_number in [1, 2]:
        if pass_number == 1:
            source_rows = p_rows
        else:
            source_rows = read_csv_rows(output_path)
        backup_rows[pass_number] = source_rows
        sides = [("px", source_rows)]
        if pass_number == 1:
            sides.append(("rt", c_rows))
        parsed = OrderedDict()
        if pass_number == 2:
            parsed["rt"] = parsed_rt

        #------------------------------------------------------------------
        #4.2a parse the file(s) into tables / sections -> title / month-header rows / group headings / data rows, with the position of the label cell
        #------------------------------------------------------------------
        for side, rows in sides:
            #month header rows: the position of the first month cell, when only blanks / numbers / "index" come before it
            header_at = [month_header_position(cells, month_re) for cells in rows]

            tables = OrderedDict()
            current = None
            offset = None
            names = []
            last_offset = None
            last_names = []
            group = ""
            parent = ""
            day_position = 0
            for i, cells in enumerate(rows):
                non_empty = [c for c in cells if c.strip() != ""]
                first = cells[0] if len(cells) > 0 else ""
                #---- table title: Standard '( n ) text' (the '(1) Change Case File = ...' lines at the top of an RT Sim file contain '=' and are not titles);
                #Ratings: a one-cell line directly above a month-header row is a section title ----
                is_title = False
                if len(non_empty) == 1 and len(cells) > 0 and cells[0].strip() != "":
                    if is_ratings:
                        is_title = i + 1 < len(rows) and header_at[i + 1] is not None
                    else:
                        is_title = bool(title_re.match(first)) and "=" not in first
                if is_title:
                    if is_ratings:
                        key = first.strip()
                    else:
                        key = title_re.match(first).group(1).lower()
                    if key in tables:
                        current = None                                     #a second table / section with the same name: its lines are left alone
                        continue
                    tables[key] = {"key": key, "title": {"row": i, "col": 0, "raw": first}, "headers": [], "items": [], "headings": []}
                    current = key
                    offset = None
                    names = []
                    if last_offset is not None and not (i + 1 < len(rows) and header_at[i + 1] is not None) and not (i + 2 < len(rows) and header_at[i + 2] is not None):
                        offset = last_offset                           #the table has no month-header row of its own
                        names = list(last_names)
                    group = ""
                    parent = ""
                    day_position = 0
                    continue
                if current is None:
                    continue
                #---- month header row ----
                if header_at[i] is not None:
                    offset = header_at[i]
                    names = [c.strip() for c in cells[offset:]]
                    while names and names[-1] == "":
                        names.pop()
                    tables[current]["headers"].append({"tab": current, "row": i, "col": offset - 1 if offset >= 1 else 0, "label": cells[offset - 1] if offset >= 1 else "", "kind": "header",
                                                       "names": [n for n in names if n.lower() not in non_value_columns]})
                    if offset == 1 and len([n for n in names if n.lower() not in non_value_columns]) >= len(last_names) - 1:
                        last_offset = offset
                        last_names = list(names)
                    continue
                if offset is None:
                    #RT Sim writes the first group heading of a table before the table's month-header row
                    if first.strip() != "" and len(non_empty) == 1:
                        group = first
                        parent = ""
                        day_position = 0
                        tables[current]["headings"].append(heading_record(current, i, 0, first))
                    continue
                label_col = offset - 1 if offset >= 1 else 0
                cell_label = cells[label_col] if len(cells) > label_col else ""
                #a DATA row has something in a value column; a heading / spacer line does not. Only "is there something" is looked at.
                has_values = False
                for k, name in enumerate(names):
                    if name.lower() in non_value_columns:
                        continue
                    if offset + k < len(cells) and cells[offset + k].strip() != "":
                        has_values = True
                        break
                if cell_label.strip() == "" and not has_values:
                    continue                                              #spacer row
                next_is_header = i + 1 < len(rows) and header_at[i + 1] is not None
                if cell_label.strip() != "" and not has_values and (current in r_heading or next_is_header):
                    group = cell_label                                    #group heading (effluent type / contract / plant)
                    parent = ""
                    day_position = 0
                    tables[current]["headings"].append(heading_record(current, i, label_col, cell_label))
                    continue
                kind = "row"
                item_group = group
                if current in r_sub:
                    if cell_label[:5].strip() == "":
                        kind = "sub"
                        item_group = parent
                    else:
                        parent = cell_label
                        item_group = ""
                item = {"tab": current, "row": i, "col": label_col, "label": cell_label, "group": item_group, "kind": kind, "pos": day_position}
                if current in r_day:
                    day_position += 1
                tables[current]["items"].append(item)
            parsed[side] = tables
        parsed_rt = parsed["rt"]
        px_tables = parsed["px"]
        rt_tables = parsed["rt"]
        px_tables_by_pass[pass_number] = px_tables

        #------------------------------------------------------------------
        #4.2b Ratings only: map every PLEXOS section onto an RT Sim section (dynamic - no hard-coded section names). The METHOD of every match is kept:
        #exact name / alias / name prefix are approved KeyedData categories; an approximate name similarity is flagged (PlexOS-4.04 step 3).
        #------------------------------------------------------------------
        section_map = {}
        section_method = {}
        if is_ratings:
            used = set()
            for px_name in px_tables:
                px_plain = re.sub(r"[^a-z0-9]", "", px_name.lower())
                px_compact = rating_section_aliases.get(px_plain, px_plain)
                best_name = None
                best_score = 0.0
                best_method = ""
                for rt_name in rt_tables:
                    if rt_name in used:
                        continue
                    rt_compact = re.sub(r"[^a-z0-9]", "", rt_name.lower())
                    if rt_compact == px_plain:
                        score = 4.0
                        method = "exact name"
                    elif rt_compact == px_compact:
                        score = 3.0
                        method = "alias (rating_section_aliases)"
                    elif px_compact != "" and (rt_compact.startswith(px_compact) or px_compact.startswith(rt_compact)):
                        score = 2.0 + min(len(px_compact), len(rt_compact)) / max(len(px_compact), len(rt_compact))
                        method = "alias + name prefix" if px_compact != px_plain else "name prefix"
                    else:
                        ratio = SequenceMatcher(None, px_compact, rt_compact).ratio()
                        score = ratio if ratio >= 0.75 else 0.0
                        method = f"name similarity {ratio:.2f} (approximate)"
                    if score > best_score:
                        best_score = score
                        best_name = rt_name
                        best_method = method
                if best_name is not None:
                    section_map[px_name] = best_name
                    section_method[px_name] = (best_method, best_score)
                    used.add(best_name)
            if pass_number == 1:
                print("Ratings sections (PLEXOS -> RT Sim):")
                for px_name in px_tables:
                    print(f"   {px_name}  ->  {section_map.get(px_name, 'no RT Sim section (kept under its own name)')}")
                for rt_name in rt_tables:
                    if rt_name not in used:
                        print(f"   (RT Sim only, not added)  {rt_name}")
        section_maps[pass_number] = section_map
        section_methods[pass_number] = section_method

        #------------------------------------------------------------------
        #4.2c normalize, match PLEXOS rows to RT Sim rows, decide the new text of every label cell
        #------------------------------------------------------------------
        candidates = []
        rt_only = []
        for px_key, px_table in px_tables.items():
            if is_ratings:
                rt_key = section_map.get(px_key)
            else:
                rt_key = px_key
            rt_table = rt_tables.get(rt_key) if rt_key is not None else None
            table_name = rt_key if (is_ratings and rt_key is not None) else px_key         #the name shown for this table / section (the RT Sim section name when it has one)
            in_convert = convert_set is None or px_key.lower() in convert_set or table_name.lower() in convert_set
            in_verify = verify_set is None or px_key.lower() in verify_set or table_name.lower() in verify_set
            #pass 2 reads the OUTPUT file, where a renamed rating section carries its RT Sim name and would no longer be found in the selection lists:
            #it must convert / verify exactly what pass 1 converted / verified
            if pass_number == 1:
                if in_convert:
                    converted_names.add(table_name)
                if in_verify:
                    verified_names.add(table_name)
            else:
                in_convert = table_name in converted_names
                in_verify = table_name in verified_names
            rt_items = rt_table["items"] if rt_table is not None else []
            rt_heads = rt_table["headings"] if rt_table is not None else []

            #---- keys: group / label normalized the same way on both sides ----
            for key_side, key_items in [("px", px_table["items"] + px_table["headings"]), ("rt", rt_items + rt_heads)]:
                for item in key_items:
                    tab = item["tab"]
                    group_text = item["group"]
                    label_text = item["label"]
                    if tab in r_effgroup or tab == r_efftotal:
                        group_text = group_text.split("--")[-1]
                    if tab == r_efftotal and item["kind"] != "heading":
                        label_text = ""                                #PLEXOS names this row "None-all-Generation Production-Total", RT Sim leaves it blank
                    contract = tab in r_contract
                    base_text = ""
                    unit_text = ""
                    if is_ratings and ":" in label_text and item["kind"] != "heading":
                        base_text, _, unit_text = label_text.rpartition(":")     #Ratings labels are "<resource>:<unit>"
                    override_label = None
                    if key_side == "px" and not is_ratings and item["kind"] != "heading":
                        raw_key = (tab, " ".join(label_text.lower().split()))
                        if raw_key in override_raw:
                            override_label = override_raw[raw_key]
                    texts_to_normalize = [(group_text, contract), (label_text, contract), (override_label if override_label is not None else "", False), (base_text, False), (unit_text, False)]
                    normalized_texts = [normalize_label(text, norm_steps, strip_parentheses) for text, strip_parentheses in texts_to_normalize]
                    group_key = normalized_texts[0]
                    label_key = normalized_texts[1]
                    if tab in r_effgroup or tab == r_efftotal:
                        group_key = effluent_spelling.get(group_key, group_key)
                    if tab in r_day:
                        label_key = "day%02d" % (item["pos"] + 1)
                    via = ""
                    if alias_norm.get(label_key, label_key) != label_key:
                        via = "alias"
                    group_key = alias_norm.get(group_key, group_key)
                    label_key = alias_norm.get(label_key, label_key)
                    if is_ratings and base_text != "":
                        #the alias table applies to the resource name of a "<resource>:<unit>" label
                        base_key = normalized_texts[3]
                        if alias_norm.get(base_key, base_key) != base_key:
                            via = "alias"
                        label_key = alias_norm.get(base_key, base_key) + " " + normalized_texts[4]
                    if override_label is not None:
                        label_key = normalized_texts[2]
                        via = "explicit override"
                    item["g"] = group_key
                    item["l"] = label_key
                    item["via"] = via
                    item["override"] = override_label
                    item["match"] = None
                    item["twin"] = None
                    item["fuzzy"] = False

            #---- match 1: exact normalized match ----
            lookup = defaultdict(list)
            for item in rt_items:
                lookup[(item["g"], item["l"])].append(item)
            for item in px_table["items"]:
                for candidate_item in lookup.get((item["g"], item["l"]), []):
                    if candidate_item["match"] is None:
                        candidate_item["match"] = item
                        item["match"] = candidate_item
                        break
            #---- match 2: high-confidence fuzzy match inside the same group (digits must be identical) ----
            unmatched_rt = defaultdict(list)
            for item in rt_items:
                if item["match"] is None:
                    unmatched_rt[item["g"]].append(item)
            for item in px_table["items"]:
                if item["match"] is not None or item["l"] == "" or px_key in r_day:
                    continue
                best_item = None
                best_score = 0.0
                for candidate_item in unmatched_rt.get(item["g"], []):
                    if candidate_item["match"] is not None or re.findall(r"\d+", candidate_item["l"]) != re.findall(r"\d+", item["l"]):
                        continue
                    score = SequenceMatcher(None, item["l"], candidate_item["l"]).ratio()
                    if score > best_score:
                        best_score = score
                        best_item = candidate_item
                if best_item is not None and best_score >= FUZZY_THRESHOLD:
                    best_item["match"] = item
                    item["match"] = best_item
                    item["fuzzy"] = True
            #---- match 3: a PLEXOS row repeated with the same label (e.g. table 1 'Total System Cost ($000)') gets the RT Sim label of its first twin ----
            first_matched_by_key = {}
            for item in px_table["items"]:
                if item["match"] is not None and (item["g"], item["l"]) not in first_matched_by_key:
                    first_matched_by_key[(item["g"], item["l"])] = item
            for item in px_table["items"]:
                if item["match"] is None and (item["g"], item["l"]) in first_matched_by_key and px_key not in r_day:
                    item["twin"] = first_matched_by_key[(item["g"], item["l"])]

            #---- the RT Sim group headings ----
            rt_heading_by_g = OrderedDict()
            for head in rt_heads:
                if head["g"] not in rt_heading_by_g:
                    rt_heading_by_g[head["g"]] = {"label": head["label"], "row": head["row"]}
            for item in rt_items:
                if item["kind"] != "sub" and item["group"].strip() != "" and item["g"] not in rt_heading_by_g:
                    rt_heading_by_g[item["g"]] = {"label": item["group"], "row": item["row"]}
            rt_first_header = rt_table["headers"][0] if rt_table is not None and len(rt_table["headers"]) > 0 else None

            #---- hints for the rows that stay as they are ----
            rt_unmatched_labels = {}
            px_unmatched_labels = {}
            if in_verify and rt_table is not None:
                for item in rt_items:
                    if item["match"] is None and item["l"] != "":
                        rt_unmatched_labels.setdefault(item["l"], item)
                for item in px_table["items"]:
                    if item["match"] is None and item["twin"] is None and item["l"] != "":
                        px_unmatched_labels.setdefault(item["l"], item)

            table_cands = []
            #---- the table / section title ----
            title_cand = new_label_candidate(table_name, "title", px_table["title"]["row"], 0, px_table["title"]["raw"], px_key)
            if rt_table is not None:
                title_cand["rt_row"] = rt_table["title"]["row"] + 1
                title_cand["rt_label"] = rt_table["title"]["raw"]
                if CHANGE_TABLE_TITLES:
                    title_cand["new"] = rt_table["title"]["raw"]
                    title_cand["state"] = "matched"
                else:
                    title_cand["state"] = "option_off"
            table_cands.append(title_cand)
            #---- the month-header rows ----
            for head in px_table["headers"]:
                head_cand = new_label_candidate(table_name, "header", head["row"], head["col"], head["label"], px_key)
                if rt_table is not None and rt_first_header is None:
                    head_cand["state"] = "no_rt_header"
                elif rt_table is not None:
                    head_cand["rt_row"] = rt_first_header["row"] + 1
                    head_cand["rt_label"] = rt_first_header["label"]
                    if CHANGE_HEADER_FIRST_CELL:
                        head_cand["new"] = rt_first_header["label"]
                        head_cand["state"] = "matched"
                    else:
                        head_cand["state"] = "option_off"
                table_cands.append(head_cand)
            #---- the group headings ----
            for head in px_table["headings"]:
                head_cand = new_label_candidate(table_name, "heading", head["row"], head["col"], head["label"], px_key, l=head["g"])
                if rt_table is not None:
                    rt_head = rt_heading_by_g.get(head["g"])
                    if rt_head is None:
                        head_cand["state"] = "kept"
                    else:
                        head_cand["rt_row"] = rt_head["row"] + 1
                        head_cand["rt_label"] = rt_head["label"]
                        if CHANGE_GROUP_HEADINGS:
                            head_cand["new"] = rt_head["label"]
                            head_cand["state"] = "matched"
                        else:
                            head_cand["state"] = "option_off"
                table_cands.append(head_cand)
            #---- the data rows ----
            for item in px_table["items"]:
                item_cand = new_label_candidate(table_name, item["kind"], item["row"], item["col"], item["label"], px_key, via=item["via"], fuzzy=item["fuzzy"], group=item["group"].strip(), l=item["l"])
                if rt_table is not None:
                    if item["match"] is not None:
                        item_cand["new"] = item["match"]["label"]
                        item_cand["rt_row"] = item["match"]["row"] + 1
                        item_cand["rt_label"] = item["match"]["label"]
                        item_cand["state"] = "matched"
                    elif item["twin"] is not None:
                        item_cand["new"] = item["twin"]["match"]["label"]
                        item_cand["rt_row"] = item["twin"]["match"]["row"] + 1
                        item_cand["rt_label"] = item["twin"]["match"]["label"]
                        item_cand["state"] = "twin"
                    elif item["override"] is not None:
                        item_cand["new"] = item["override"]
                        item_cand["state"] = "override"
                    else:
                        item_cand["state"] = "kept"
                        if item["l"] in px_unmatched_labels and len(rt_unmatched_labels) > 0:
                            close = get_close_matches(item["l"], list(rt_unmatched_labels.keys()), n=1, cutoff=0.75)
                            if len(close) > 0:
                                item_cand["hint"] = f"closest RT Sim-only label: '{rt_unmatched_labels[close[0]]['label'].strip()}' (same row under another name? add it to alias_map)"
                table_cands.append(item_cand)
            for cand in table_cands:
                cand["in_verify"] = in_verify
                if not in_convert:
                    cand["new"] = cand["old"]
                    cand["state"] = "not_converted"
            candidates.extend(table_cands)

            #---- RT Sim rows that no PLEXOS row matched ----
            if in_verify and rt_table is not None:
                for item in rt_items:
                    if item["match"] is None:
                        hint = ""
                        if item["l"] != "" and len(px_unmatched_labels) > 0:
                            close = get_close_matches(item["l"], list(px_unmatched_labels.keys()), n=1, cutoff=0.75)
                            if len(close) > 0:
                                hint = f"closest PLEXOS-only label (kept as it is): '{px_unmatched_labels[close[0]]['label'].strip()}'"
                        rt_only.append({"table": table_name, "row": item["row"] + 1, "label": item["label"], "group": item["group"].strip(), "hint": hint})
        #---- RT Sim tables / sections that no PLEXOS table / section matched ----
        mapped_rt_keys = set(section_map.values()) if is_ratings else set(px_tables.keys())
        rt_only_tables = []
        for rt_key2 in rt_tables:
            if rt_key2 not in mapped_rt_keys and (verify_set is None or rt_key2.lower() in verify_set):
                rt_only_tables.append({"table": rt_key2, "row": rt_tables[rt_key2]["title"]["row"] + 1, "label": rt_tables[rt_key2]["title"]["raw"], "n_rows": len(rt_tables[rt_key2]["items"])})

        #------------------------------------------------------------------
        #4.2d status and reason for every label cell
        #------------------------------------------------------------------
        for cand in candidates:
            old = cand["old"]
            new = cand["new"]
            state = cand["state"]
            cand["why_short"] = ""
            if state == "no_table":
                cand["status"] = ST_NOTABLE
                cand["why"] = f"RT Sim has no {unit_word} matching this one; the PLEXOS text is left as it is."
                cand["why_short"] = f"{unit_word} not in RT Sim"
            elif state == "no_rt_header":
                cand["status"] = ST_KEPT
                cand["why"] = f"The RT Sim {unit_word} has no month-header row to copy; left as it is."
                cand["why_short"] = f"RT Sim {unit_word} has no month-header row"
            elif state == "kept":
                cand["status"] = ST_KEPT
                cand["why"] = "RT Sim has no matching row; the PLEXOS label is kept exactly as it is." + (" " + cand["hint"] if cand["hint"] != "" else "")
                cand["why_short"] = "no RT Sim row"
            elif state == "not_converted":
                cand["status"] = ST_OFF
                cand["why"] = "This table / section is not in TABLES_TO_CONVERT / RATING_SECTIONS_TO_CONVERT."
                cand["why_short"] = f"{unit_word} not selected"
            elif state == "option_off":
                cand["status"] = ST_OFF
                cand["why"] = "The setting for this kind of label (CHANGE_TABLE_TITLES / CHANGE_GROUP_HEADINGS / CHANGE_HEADER_FIRST_CELL) is False."
                cand["why_short"] = "option off"
            elif old == new:
                cand["status"] = ST_SAME
                cand["why"] = "The PLEXOS text already equals the RT Sim text."
                cand["why_short"] = "already identical"
            else:
                #the ladder: the FIRST step at which the old and the new text become equal is the reason they differ
                o1 = old.strip()
                n1 = new.strip()
                o2 = " ".join(old.split())
                n2 = " ".join(new.split())
                o3 = o2.lower()
                n3 = n2.lower()
                o4 = normalize_label(old, norm_steps)
                n4 = normalize_label(new, norm_steps)
                o5 = normalize_label(old, norm_steps, True)
                n5 = normalize_label(new, norm_steps, True)
                stage = 0
                if o1 == n1:
                    stage = 1
                elif o2 == n2:
                    stage = 2
                elif o3 == n3:
                    stage = 3
                elif o4 == n4:
                    stage = 4
                elif o5 == n5:
                    stage = 5
                prefix = ""
                if state == "twin":
                    prefix = "Repeated PLEXOS row (same label as an earlier row): gets the same RT Sim label. "
                if stage == 1:
                    lead_old, trail_old = count_padding(old)
                    lead_new, trail_new = count_padding(new)
                    cand["status"] = ST_FORMAT
                    cand["why"] = prefix + f"Padding only: {lead_old} leading / {trail_old} trailing spaces -> {lead_new} / {trail_new} (RT Sim indents and right-pads its labels)."
                    cand["why_short"] = "padding only (leading / trailing spaces)"
                elif stage == 2:
                    cand["status"] = ST_FORMAT
                    cand["why"] = prefix + "Spacing inside the text (e.g. a double space) differs."
                    cand["why_short"] = "spacing inside the text"
                elif stage == 3:
                    cand["status"] = ST_FORMAT
                    cand["why"] = prefix + "Upper / lower case differs."
                    cand["why_short"] = "upper / lower case"
                elif stage == 4:
                    cand["status"] = ST_FORMAT
                    cand["why"] = prefix + "Punctuation or the spacing around ( ) - _ & / , differs."
                    cand["why_short"] = "punctuation / spacing around ( ) - _ & / ,"
                elif stage == 5:
                    cand["status"] = ST_UNITS
                    cand["why"] = prefix + "Only the text inside parentheses (units / suffix) differs."
                    cand["why_short"] = "text inside parentheses (units) differs"
                else:
                    cand["status"] = ST_RENAMED
                    if state == "override":
                        cand["why"] = "RT Sim has no row for this label; it is written as your explicit_overrides table says."
                        cand["why_short"] = "label from explicit_overrides (no RT Sim row)"
                    elif cand["kind"] == "title" and is_ratings:
                        cand["why"] = f"Rating section name: the PLEXOS section is matched to the RT Sim section by name ('{old.strip()}' -> '{new.strip()}')."
                        cand["why_short"] = "rating section renamed to the RT Sim section"
                    elif cand["via"] != "":
                        cand["why"] = prefix + f"Different name, matched through {cand['via']}: '{old.strip()}' -> '{new.strip()}'."
                        cand["why_short"] = f"renamed through {cand['via']}"
                    elif cand["fuzzy"]:
                        cand["why"] = prefix + f"Different name, matched by name similarity: '{old.strip()}' -> '{new.strip()}'."
                        cand["why_short"] = "renamed, matched by name similarity"
                    elif cand["tab"] in r_day and cand["kind"] == "row":
                        cand["why"] = "Contract day row: the label is taken from the RT Sim day row in the same position (PLEXOS leaves the first day row unlabeled)."
                        cand["why_short"] = "contract day row, label by position"
                    elif cand["tab"] == r_efftotal and cand["kind"] == "row":
                        cand["why"] = "Table 32 total row: RT Sim leaves the label blank (PLEXOS writes a generated name)."
                        cand["why_short"] = "table 32 total row label"
                    elif cand["kind"] == "heading" and (cand["tab"] in r_effgroup or cand["tab"] == r_efftotal):
                        cand["why"] = "Effluent heading: PLEXOS spells the types SO2 / NOX / CO2, RT Sim SOx / NOx / COx."
                        cand["why_short"] = "effluent heading spelling"
                    else:
                        cand["why"] = prefix + f"Different text: '{old.strip()}' -> '{new.strip()}'."
                        cand["why_short"] = "different text"
        cands_by_pass[pass_number] = candidates
        rt_only_by_pass[pass_number] = rt_only
        rt_only_tables_by_pass[pass_number] = rt_only_tables

        #------------------------------------------------------------------
        #4.2e pass 1 only: write THE ONE OUTPUT CSV = the PLEXOS file with the planned label cells replaced,
        #and top_blank blank lines in front of everything (before the first section)
        #------------------------------------------------------------------
        if pass_number == 1:
            out_rows = [list(r) for r in source_rows]
            for cand in candidates:
                if cand["new"] != cand["old"]:
                    while len(out_rows[cand["row"]]) <= cand["col"]:
                        out_rows[cand["row"]].append("")
                    out_rows[cand["row"]][cand["col"]] = cand["new"]
            out_rows = [[] for blank_index in range(top_blank)] + out_rows       #an empty list is written as an empty line
            with open(output_path, "w", newline="", encoding="utf-8-sig" if has_bom else "utf-8") as f:
                csv.writer(f, lineterminator=line_end).writerows(out_rows)
            print()
            print(f"Pass 1 done: {sum(1 for c in candidates if c['new'] != c['old'])} label cells changed, {top_blank} blank lines added at the top -> {output_path}")


    cands1 = cands_by_pass[1]
    cands2 = cands_by_pass[2]
    b_rows = backup_rows[2]                       #the output file, read back from disk
    pos2 = {}
    for cand in cands2:
        pos2[(cand["row"], cand["col"])] = cand                      #pass 2 line numbers are OUTPUT-file lines
    rt_tables_final = parsed_rt

    #=================================================================
    #4.3 VERIFY THE OUTPUT FILE   (PLEXOS line r is output line r + top_blank)      [computed here, reported in Step 6]
    #=================================================================
    checks = []
    changed_positions = {}
    for cand in cands1:
        if cand["new"] != cand["old"]:
            changed_positions[(cand["row"], cand["col"])] = cand["new"]
    #check 1: the number of lines = the PLEXOS lines + the blank lines added at the top
    checks.append(("Same number of lines as the PLEXOS input plus the blank lines added at the top", len(p_rows) + top_blank == len(b_rows),
                   f"{len(p_rows)} lines in the input + {top_blank} blank lines added = {len(p_rows) + top_blank} expected, {len(b_rows)} in the output"))
    #check 1b: the first top_blank lines are blank, and the first line with content (the first section) comes right after them
    if top_blank > 0:
        blanks_ok = len(b_rows) >= top_blank and all(all(c.strip() == "" for c in b_rows[index]) for index in range(top_blank))
        first_content_in = count_leading_blank_rows(p_rows)
        first_content_out = count_leading_blank_rows(b_rows)
        first_text = b_rows[first_content_out][0].strip() if first_content_out < len(b_rows) and len(b_rows[first_content_out]) > 0 else ""
        first_header_line = None
        for index, cells in enumerate(b_rows):
            if month_header_position(cells, month_re) is not None:
                first_header_line = index + 1
                break
        checks.append((f"The output starts with {top_blank} blank lines added before the first {unit_word}, which follows right after them",
                       blanks_ok and first_content_out == first_content_in + top_blank,
                       f"lines 1-{top_blank} blank: {'yes' if blanks_ok else 'NO'}; first line with content = line {first_content_out + 1} ('{first_text}'); first month-header row = line {first_header_line}"))
    #check 2: every cell outside the changed label cells is identical (a count only - no value is listed)
    if VERIFY_OTHER_CELLS_UNCHANGED:
        cells_checked = 0
        cells_differ = 0
        rows_length_differ = 0
        for r in range(min(len(p_rows), len(b_rows) - top_blank)):
            a_row = p_rows[r]
            b_row = b_rows[r + top_blank]
            if len(a_row) != len(b_row):
                rows_length_differ += 1
                continue
            for c in range(len(a_row)):
                if (r, c) in changed_positions:
                    continue
                cells_checked += 1
                if a_row[c] != b_row[c]:
                    cells_differ += 1
        checks.append(("Every cell outside the changed label cells is identical to the PLEXOS input (numbers and all other columns untouched)", cells_differ == 0 and rows_length_differ == 0,
                       f"{cells_checked} cells checked, {cells_differ} differ, {rows_length_differ} lines with a different number of cells"))
    #check 3: every planned change is in the file
    planned_present = 0
    for position in changed_positions:
        out_line = position[0] + top_blank
        if out_line < len(b_rows) and position[1] < len(b_rows[out_line]) and b_rows[out_line][position[1]] == changed_positions[position]:
            planned_present += 1
    checks.append(("Every planned label change is in the output file", planned_present == len(changed_positions), f"{planned_present} of {len(changed_positions)} changed cells found"))
    #check 4: every label that has an RT Sim counterpart now equals the RT Sim label exactly (independent re-match of the output file)
    matched_total = 0
    matched_equal = 0
    not_equal_examples = []
    for cand in cands2:
        if cand["state"] in ("matched", "twin"):
            matched_total += 1
            if cand["old"] == cand["new"]:
                matched_equal += 1
            elif len(not_equal_examples) < 5:
                not_equal_examples.append(f"{cand['table']} line {cand['row'] + 1}: '{cand['old']}' vs RT Sim '{cand['new']}'")
    matched_pass1 = sum(1 for c in cands1 if c["state"] in ("matched", "twin"))
    checks.append(("Every label that has an RT Sim counterpart equals the RT Sim label exactly (output re-matched against the RT Sim file)", matched_total == matched_equal and matched_total == matched_pass1,
                   f"{matched_equal} of {matched_total} labels identical to RT Sim ({matched_pass1} were matched when the file was written)" + ("; e.g. " + " | ".join(not_equal_examples) if not_equal_examples else "")))
    #check 5: a second run changes nothing
    second_run_changes = sum(1 for c in cands2 if c["new"] != c["old"])
    checks.append(("Running the conversion again on the output changes nothing", second_run_changes == 0, f"{second_run_changes} label cells would still change"))
    #check 6 (Ratings only): every PLEXOS section that was matched to an RT Sim section is still matched to the SAME RT Sim section when the output file is read back.
    #(When the section title was changed, the output section is named exactly like the RT Sim section; when it was not changed - option off / section not selected - it keeps its PLEXOS name.)
    if is_ratings:
        title_in_output = {}
        for cand in cands1:
            if cand["kind"] == "title":
                title_in_output[cand["tab"]] = cand["new"].strip()
        sections_ok = 0
        sections_bad = []
        for px_name in section_maps[1]:
            output_name = title_in_output.get(px_name, px_name)
            if section_maps[2].get(output_name) == section_maps[1][px_name]:
                sections_ok += 1
            else:
                sections_bad.append(f"'{px_name}' -> written as '{output_name}' -> matched to '{section_maps[2].get(output_name)}' instead of '{section_maps[1][px_name]}'")
        checks.append(("Every PLEXOS section keeps its RT Sim section match when the output file is read back (and is named exactly like it when its title was changed)", sections_ok == len(section_maps[1]),
                       f"{sections_ok} of {len(section_maps[1])} sections" + ("; e.g. " + " | ".join(sections_bad[:3]) if sections_bad else "")))
    label_checks_pass = all(passed for name, passed, detail in checks)

    #=================================================================
    #4.4 NUMBERS FOR THE CONSOLE AND THE WORKBOOK
    #=================================================================
    summary = OrderedDict()
    for cand in cands1:
        if cand["in_verify"]:
            summary.setdefault(cand["table"], Counter())[cand["status"]] += 1
    rt_only_count = Counter()
    for item in rt_only_by_pass[2]:
        rt_only_count[item["table"]] += 1
    problem_count = Counter()
    for cand in cands2:
        if cand["in_verify"] and cand["state"] in ("matched", "twin") and cand["old"] != cand["new"]:
            problem_count[cand["table"]] += 1
    if is_ratings or verify_set is None:
        verify_tables_present = list(summary.keys())
        for item in rt_only_tables_by_pass[2]:
            if item["table"] not in verify_tables_present:
                verify_tables_present.append(item["table"])
    else:
        verify_tables_present = list(TABLES_TO_VERIFY)
    all_status = Counter()
    kind_changed = Counter()
    why_changed = Counter()
    for cand in cands1:
        all_status[cand["status"]] += 1
        if cand["new"] != cand["old"]:
            kind_changed[cand["kind"]] += 1
            why_changed[cand["why_short"]] += 1

    #=================================================================
    #4.5 THIS REPORT'S PART OF THE VERIFICATION WORKBOOK
    #=================================================================
    ws = summary_ws
    ws.append([])
    append_bold_row(ws, [f"==================== {TC_NAME.upper()} ===================="], size=12)
    ws.append(["PLEXOS input", p_path])
    ws.append(["RT Sim file (KeyedData baseline; labels copied from it)", c_path])
    ws.append(["THE ONE OUTPUT CSV", output_path])
    ws.append(["Blank lines added at the top of the output (before the first section)", top_blank])
    ws.append([])
    ws.append(["Verification check", "Result", "Detail"])
    style_cells(ws[ws.max_row], fill_header, font_header)
    for name, passed, detail in checks:
        ws.append([name, "PASS" if passed else "FAIL", detail])
        color_pass_fail(ws, ws.max_row, 2, passed, fill_same, fill_bad)
    ws.append([])
    ws.append([unit_word.capitalize(), "Labels"] + status_columns + ["RT Sim-only rows (not in the output)", "Output differs from RT Sim (must be 0)"])
    style_cells(ws[ws.max_row], fill_header, font_header, True)
    totals = Counter()
    for table_number in summary:
        ws.append([table_number, sum(summary[table_number].values())] + [summary[table_number][s] for s in status_columns] + [rt_only_count[table_number], problem_count[table_number]])
        totals.update(summary[table_number])
    ws.append(["TOTAL", sum(totals.values())] + [totals[s] for s in status_columns] + [sum(rt_only_count.values()), sum(problem_count.values())])
    style_cells(ws[ws.max_row], font=Font(bold=True))
    ws.append([])
    append_bold_row(ws, [f"Why {report_label} labels were changed", "count"])
    for reason, n in why_changed.most_common(20):
        ws.append([reason, n])

    #---------------- Labels sheet of this report ----------------
    label_headers = [unit_word.capitalize(), "Kind", "PLEXOS line", "PLEXOS label (before)", "Output label (after, read back from the file)", "RT Sim label (target)", "RT Sim line", "Result", "What changed and why", "Output = RT Sim?", "Output line"]
    ws = add_header_sheet(wb, f"{report_label} Labels", label_headers, fill_header, font_header, True)
    labels_rows = 0
    for cand in cands1:
        if not cand["in_verify"]:
            continue
        out_line = cand["row"] + top_blank                    #the same line in the output file (0-based)
        cand2 = pos2.get((out_line, cand["col"]))
        after = b_rows[out_line][cand["col"]] if out_line < len(b_rows) and cand["col"] < len(b_rows[out_line]) else "(missing)"
        if cand2 is not None and cand2["state"] in ("matched", "twin"):
            equal_text = "Yes" if cand2["old"] == cand2["new"] else "NO"
        else:
            equal_text = "n/a (no RT Sim row)"
        shown = []
        for text in [cand["old"], after, cand["rt_label"] if cand["rt_label"] != "" or cand["state"] in ("matched", "twin") else ""]:
            shown.append(show_spaces(text) if text != "" else ("(empty)" if cand["state"] in ("matched", "twin") or cand["kind"] in ("header", "title") else ""))
        rt_shown = shown[2] if cand["rt_row"] != "" else "(no RT Sim row)"
        ws.append([cand["table"], cand["kind"], cand["row"] + 1, shown[0] if shown[0] != "" else "(empty)", shown[1] if shown[1] != "" else "(empty)", rt_shown, cand["rt_row"], cand["status"], cand["why"], equal_text, out_line + 1])
        labels_rows += 1
        excel_row = ws.max_row
        for column_index in range(1, len(label_headers) + 1):
            ws.cell(row=excel_row, column=column_index).number_format = "@"
        ws.cell(row=excel_row, column=8).fill = status_fill[cand["status"]]
        if equal_text == "NO":
            ws.cell(row=excel_row, column=5).fill = fill_bad
            ws.cell(row=excel_row, column=10).fill = fill_bad
        elif equal_text == "Yes":
            ws.cell(row=excel_row, column=10).fill = fill_same
    overall_labels_rows[report_label] = labels_rows
    set_openpyxl_column_widths(ws, [18, 9, 10, 42, 42, 42, 10, 40, 90, 20, 12])
    ws.freeze_panes = "E2"
    ws.auto_filter.ref = ws.dimensions

    #---------------- RT Sim only sheet of this report ----------------
    ws = add_header_sheet(wb, f"{report_label} RT Sim only", [unit_word.capitalize(), "RT Sim line", "RT Sim label", "Group", "Why it is not in the output", "Closest PLEXOS-only label"], fill_header, font_header)
    for item in rt_only_tables_by_pass[2]:
        ws.append([item["table"], item["row"], item["label"], "", f"The whole {unit_word} ({item['n_rows']} rows) exists only in RT Sim; it is not added.", ""])
    written_rt_only = 0
    for item in rt_only_by_pass[2]:
        if written_rt_only >= MAX_RT_ONLY_ROWS_IN_WORKBOOK:
            ws.append([f"(stopped at {MAX_RT_ONLY_ROWS_IN_WORKBOOK} rows - raise MAX_RT_ONLY_ROWS_IN_WORKBOOK to list more; {len(rt_only_by_pass[2])} RT Sim-only rows in total)"])
            break
        visible = show_spaces(item["label"])
        ws.append([item["table"], item["row"], visible if visible != "" else "(empty)", item["group"], "RT Sim has this row, the PLEXOS file does not. The output keeps the PLEXOS rows only, so it is not added.", item["hint"]])
        written_rt_only += 1
    set_openpyxl_column_widths(ws, [18, 11, 42, 30, 80, 70])
    ws.freeze_panes = "A2"
    ws.auto_filter.ref = ws.dimensions

    #---------------- label format patterns of this report (the 'Label Formats' sheet is written after the loop) ----------------
    format_sets = OrderedDict()
    format_sets["PLEXOS input (before)"] = [cand["old"] for cand in cands1 if cand["in_verify"] and cand["kind"] in ("row", "sub")]
    format_sets["Output (after)"] = [b_rows[cand["row"] + top_blank][cand["col"]] for cand in cands1 if cand["in_verify"] and cand["kind"] in ("row", "sub") and cand["row"] + top_blank < len(b_rows) and cand["col"] < len(b_rows[cand["row"] + top_blank])]
    rt_labels_all = []
    for table_number in rt_tables_final:
        if verify_set is None or table_number.lower() in verify_set or (is_ratings and table_number in section_maps[1].values()):
            for item in rt_tables_final[table_number]["items"]:
                rt_labels_all.append(item["label"])
    format_sets["RT Sim (target)"] = rt_labels_all
    for set_name in format_sets:
        pattern_counts = Counter()
        for text in format_sets[set_name]:
            lead, trail = count_padding(text)
            if trail > 0:
                pattern_counts[f"{lead} leading spaces, right-padded to {len(text)} characters"] += 1
            else:
                pattern_counts[f"{lead} leading spaces, no padding"] += 1
        for pattern, n in pattern_counts.most_common(12):
            formats_rows.append([report_label, set_name, "row / sub-row labels", pattern, n])

    #=================================================================
    #4.6 THE SECTION MAPPING RECORDS: one per {unit_word} in the verification scope. Each record answers the questions of the test case:
    #is the section mapped to a KeyedData structure, is that structure valid, is the destination defined, is the category valid / approved, is the section ID traceable to a KeyedData record.
    #=================================================================
    pass1_px_tables = px_tables_by_pass[1]
    pass2_px_tables = px_tables_by_pass[2]
    title_cands = {}
    rows_by_tab = defaultdict(list)
    for cand in cands1:
        if cand["kind"] == "title":
            title_cands[cand["tab"]] = cand
        elif cand["kind"] in ("row", "sub"):
            rows_by_tab[cand["tab"]].append(cand)
    section_records = []
    sections_outside_scope = []
    for px_key, px_table in pass1_px_tables.items():
        rt_key = section_maps[1].get(px_key) if is_ratings else (px_key if px_key in rt_tables_final else None)
        table_name = rt_key if (is_ratings and rt_key is not None) else px_key
        if table_name not in verified_names:
            sections_outside_scope.append(px_key)
            continue
        rt_table = rt_tables_final.get(rt_key) if rt_key is not None else None
        record = {"id": px_key, "px_title": px_table["title"]["raw"].strip(), "px_line": px_table["title"]["row"] + 1, "kd_key": rt_key if rt_table is not None else None,
                  "kd_title": "", "kd_line": None, "method": "", "destination_defined": rt_table is not None, "structure_valid": False, "structure_detail": "",
                  "category_valid": False, "category_detail": "", "traceable": False, "trace_detail": "", "rows": len(rows_by_tab[px_key]),
                  "rows_traced": sum(1 for c in rows_by_tab[px_key] if c["state"] in ("matched", "twin"))}
        if rt_table is None:
            record["method"] = "none"
            record["structure_detail"] = f"no KeyedData {unit_word} found for this {unit_word}"
            record["category_detail"] = "no KeyedData destination"
            record["trace_detail"] = "no KeyedData record"
        else:
            record["kd_title"] = rt_table["title"]["raw"].strip()
            record["kd_line"] = rt_table["title"]["row"] + 1
            #---- valid KeyedData structure: the KeyedData section holds records, and (when both have a month-header row) the same number of value columns as the PLEXOS section ----
            kd_record_rows = len(rt_table["items"])
            px_header_names = px_table["headers"][0]["names"] if len(px_table["headers"]) > 0 else None
            kd_header_names = rt_table["headers"][0]["names"] if len(rt_table["headers"]) > 0 else None
            if px_header_names is not None and kd_header_names is not None:
                columns_ok = (len(px_header_names) == len(kd_header_names)) or (not REQUIRE_SAME_COLUMN_COUNT)
                names_differ = sum(1 for a, b in zip(px_header_names, kd_header_names) if a.strip().lower() != b.strip().lower())
                column_text = f"value columns PLEXOS {len(px_header_names)} / KeyedData {len(kd_header_names)}" + (f"; {names_differ} column name(s) differ" if names_differ > 0 else "")
            else:
                columns_ok = True
                column_text = "value columns not compared (a month-header row is missing on one side)"
            record["structure_valid"] = kd_record_rows > 0 and columns_ok
            record["structure_detail"] = f"{kd_record_rows} KeyedData record row(s); {column_text}"
            #---- valid / approved category ----
            if is_ratings:
                method_text, method_score = section_methods[1].get(px_key, ("", 0.0))
                record["method"] = method_text
                record["category_valid"] = method_text in RATING_APPROVED_METHODS
                record["category_detail"] = f"PLEXOS section '{px_key}' -> KeyedData section '{rt_key}' by {method_text}" + ("" if record["category_valid"] else ": an approximate name match is not an approved category, review it")
            else:
                px_title_text = title_re.match(px_table["title"]["raw"]).group(2)
                kd_title_text = title_re.match(rt_table["title"]["raw"]).group(2)
                same_after_rules = title_key_full(px_title_text) == title_key_full(kd_title_text) or title_key_plain(px_title_text) == title_key_plain(kd_title_text)
                title_similarity = SequenceMatcher(None, title_key_plain(px_title_text), title_key_plain(kd_title_text)).ratio()
                if " ".join(px_title_text.lower().split()) == " ".join(kd_title_text.lower().split()):
                    record["method"] = "same section number, identical title"
                    record["category_valid"] = True
                elif same_after_rules:
                    record["method"] = "same section number, same title after normalisation / alias / units"
                    record["category_valid"] = True
                elif title_similarity >= TITLE_SIMILARITY_MIN:
                    record["method"] = f"same section number, similar title ({title_similarity:.2f})"
                    record["category_valid"] = True
                else:
                    record["method"] = f"same section number, DIFFERENT title ({title_similarity:.2f})"
                    record["category_valid"] = False
                record["category_detail"] = f"PLEXOS '{px_title_text.strip()}' vs KeyedData '{kd_title_text.strip()}' ({record['method']})"
        #---- traceable: the section ID has a KeyedData record location AND the exported output file holds the section at the expected line, titled like its KeyedData section ----
        title_cand = title_cands.get(px_key)
        if rt_table is not None and title_cand is not None:
            output_key = title_cand["new"].strip() if is_ratings else px_key
            output_table = pass2_px_tables.get(output_key)
            output_line_expected = px_table["title"]["row"] + top_blank + 1
            found_in_output = output_table is not None and output_table["title"]["row"] + 1 == output_line_expected and output_table["title"]["raw"] == title_cand["new"]
            record["traceable"] = found_in_output
            record["trace_detail"] = f"KeyedData line {record['kd_line']}; output line {output_line_expected}" + ("" if found_in_output else " (NOT found there in the exported file)")
        record["section_pass"] = record["destination_defined"] and record["structure_valid"] and record["category_valid"] and record["traceable"]
        section_records.append(record)
    n_sections = len(section_records)
    kd_destination_counts = Counter(r_["kd_key"] for r_ in section_records if r_["kd_key"] is not None)
    #Standard: the sections the test case expects (TABLES_TO_VERIFY) that are not in the PLEXOS file at all
    missing_expected = []
    if (not is_ratings) and verify_set is not None:
        present_ids = set(k.lower() for k in pass1_px_tables)
        missing_expected = [t for t in TABLES_TO_VERIFY if t.lower() not in present_ids]

    #---- the exception report: every section that is not mapped / not valid / not traceable ----
    exception_rows = []
    for record in section_records:
        if not record["destination_defined"]:
            exception_rows.append(section_exception(record["id"], record["px_title"], record["px_line"], "Unmapped section: no KeyedData destination", "1-2", record["structure_detail"]))
            continue
        if not record["structure_valid"]:
            exception_rows.append(section_exception(record["id"], record["px_title"], record["px_line"], "KeyedData structure not valid", "1", record["structure_detail"]))
        if not record["category_valid"]:
            exception_rows.append(section_exception(record["id"], record["px_title"], record["px_line"], "KeyedData category not valid / not approved", "3", record["category_detail"]))
        if not record["traceable"]:
            exception_rows.append(section_exception(record["id"], record["px_title"], record["px_line"], "Section ID not traceable to a KeyedData record", "4", record["trace_detail"]))
    for missing_id in missing_expected:
        exception_rows.append(section_exception(missing_id, "", "", "Expected section missing from the PLEXOS report", "1", f"{unit_word} {missing_id} is in the verification list but the PLEXOS file has no such {unit_word}"))
    for exception_number, exception_row in enumerate(exception_rows, start=1):
        exception_row["Exception Id"] = f"EX-{exception_number:04d}"
        exception_row["Test Case"] = TC_ID
        exception_row["Report"] = f"{report_label} Report"

    #---- automatic exports ----
    tc_file_id = TC_ID.replace("-", "_").replace(".", "_")
    mapping_results_path = os.path.join(output_folder, f"{tc_file_id}_Section_Mapping_Results.csv")
    exceptions_path = os.path.join(output_folder, f"{tc_file_id}_Section_Mapping_Exceptions.csv")
    step_summary_path = os.path.join(output_folder, f"{tc_file_id}_Step_Summary.csv")
    mapping_columns = ["Test Case", "Report", "Section", "PLEXOS Title", "PLEXOS Line", "KeyedData Section", "KeyedData Title", "KeyedData Line", "Mapping Method", "Destination Defined",
                       "Structure Valid", "Structure Detail", "Category Valid", "Category Detail", "Traceable", "Trace Detail", "Labels In Section", "Labels Traced To KeyedData Rows", "Section Status"]
    write_csv_file(mapping_results_path, mapping_columns, [section_mapping_row(record, TC_ID, report_label, status_text) for record in section_records])
    exception_columns = ["Exception Id", "Test Case", "Report", "Section", "PLEXOS Title", "PLEXOS Line", "Exception Type", "Step", "Detail"]
    write_csv_file(exceptions_path, exception_columns, [[exception_row[c] for c in exception_columns] for exception_row in exception_rows])

    #the Section Mapping sheet of the verification workbook, then the workbook is saved (the Label Formats sheet and the colour key are added when the last report is done)
    ws = add_header_sheet(wb, f"{report_label} Section Mapping", mapping_columns, fill_header, font_header, True)
    for record in section_records:
        ws.append(section_mapping_row(record, TC_ID, report_label, status_text))
        for column_index in (10, 11, 13, 15, 19):
            color_pass_fail(ws, ws.max_row, column_index, ws.cell(row=ws.max_row, column=column_index).value == "PASS", fill_same, fill_bad)
    set_openpyxl_column_widths(ws, [12, 18, 10, 40, 10, 28, 40, 10, 46, 12, 12, 60, 12, 70, 12, 40, 10, 12, 12])
    ws.freeze_panes = "D2"
    wb.save(verify_path)
    workbook_sheet_names, workbook_problems = read_workbook_sheet_names(verify_path)
    workbook_readable = len(workbook_problems) == 0

    #=================================================================
    #4.7 THE SIX TEST STEPS: every check is a (name, PASS / FAIL / INFO, detail). The step passes when none of its checks fails. After every step the status of the whole test case is shown.
    #=================================================================
    list_ids = lambda items: ", ".join(str(x) for x in items[:12]) + (f" ... and {len(items) - 12} more" if len(items) > 12 else "")
    mapped_records = [r_ for r_ in section_records if r_["destination_defined"]]
    unmapped_records = [r_ for r_ in section_records if not r_["destination_defined"]]
    structure_bad = [r_ for r_ in mapped_records if not r_["structure_valid"]]
    category_bad = [r_ for r_ in mapped_records if not r_["category_valid"]]
    trace_bad = [r_ for r_ in section_records if not r_["traceable"]]
    destination_pct = 100.0 * len(mapped_records) / n_sections if n_sections > 0 else 0.0
    step_defs = []

    #---- Step 1: all sections are mapped to valid KeyedData structures ----
    step1_checks = []
    step1_checks.append(("The PLEXOS file and the KeyedData (RT Sim) file are two different files", not same_input_file, "the same file was given for both" if same_input_file else "two different files"
                         + ("; the two were swapped automatically (the file given as PLEXOS looked like RT Sim)" if roles_swapped else "")))
    step1_checks.append((f"PLEXOS {report_label} file read and {unit_word}s found", len(pass1_px_tables) > 0, f"{len(p_rows)} lines, {len(pass1_px_tables)} {unit_word}s found, {n_sections} in the verification scope"
                         + (f" ({len(sections_outside_scope)} outside the scope are not counted)" if len(sections_outside_scope) > 0 else "")))
    step1_checks.append((f"KeyedData (RT Sim) {report_label} file read and {unit_word}s found", len(rt_tables_final) > 0, f"{len(c_rows)} lines, {len(rt_tables_final)} KeyedData {unit_word}s found"))
    step1_checks.append((f"At least one {unit_word} is in the verification scope", n_sections > 0, f"{n_sections} {unit_word}(s) are verified"))
    if (not is_ratings) and verify_set is not None:
        step1_checks.append((f"Every expected {unit_word} is present in the PLEXOS file ({len(TABLES_TO_VERIFY)} expected)", len(missing_expected) == 0, "all present" if len(missing_expected) == 0 else f"missing from the PLEXOS file: {list_ids(missing_expected)}"))
    step1_checks.append((f"Every {unit_word} is mapped to a KeyedData structure", n_sections > 0 and len(unmapped_records) == 0, f"{len(mapped_records)} of {n_sections} mapped" + ("" if len(unmapped_records) == 0 else f"; unmapped: {list_ids([r_['id'] for r_ in unmapped_records])}")))
    step1_checks.append((f"Every mapped KeyedData structure is valid (it holds records and has the same value columns as the PLEXOS {unit_word})", len(mapped_records) > 0 and len(structure_bad) == 0,
                         f"{len(mapped_records) - len(structure_bad)} of {len(mapped_records)} valid" + ("" if len(structure_bad) == 0 else "; invalid: " + "; ".join(f"{r_['id']} ({r_['structure_detail']})" for r_ in structure_bad[:6]))))
    step_defs.append((1, f"Verify that all {report_label} Report sections are mapped to valid KeyedData structures", step1_checks))

    #---- Step 2: 100% of the sections have a defined KeyedData destination ----
    step2_checks = []
    step2_checks.append((f"KeyedData destination defined for {len(mapped_records)} of {n_sections} {unit_word}s = {destination_pct:.2f}% (required: 100%)", n_sections > 0 and len(mapped_records) == n_sections,
                         "every section has a destination" if len(unmapped_records) == 0 and n_sections > 0 else f"no destination: {list_ids([r_['id'] for r_ in unmapped_records])}"))
    step2_checks.append((f"Every destination has a line in the KeyedData file", n_sections > 0 and all(r_["kd_line"] is not None for r_ in mapped_records), "; ".join(f"{r_['id']} -> '{r_['kd_title']}' (line {r_['kd_line']})" for r_ in mapped_records[:5]) + (" ..." if len(mapped_records) > 5 else "")))
    duplicate_destinations = [k for k, v in kd_destination_counts.items() if v > 1]
    step2_checks.append((f"Each KeyedData {unit_word} is the destination of only one PLEXOS {unit_word}", len(duplicate_destinations) == 0, "one-to-one" if len(duplicate_destinations) == 0 else f"received by more than one {unit_word}: {list_ids(duplicate_destinations)}"))
    step_defs.append((2, f"Verify: 100% of {report_label} Report sections have a defined KeyedData destination", step2_checks))

    #---- Step 3: each section maps to a valid / approved KeyedData category ----
    step3_checks = []
    method_counts = Counter(r_["method"].split("(")[0].strip() for r_ in mapped_records)
    step3_checks.append((f"Each {report_label} Report section maps to {'a valid' if not is_ratings else 'an approved'} KeyedData category", n_sections > 0 and len(unmapped_records) == 0 and len(category_bad) == 0,
                         f"{len(mapped_records) - len(category_bad)} of {n_sections} valid" + ("" if len(category_bad) == 0 else "; not valid: " + " | ".join(f"{r_['id']}: {r_['category_detail']}" for r_ in category_bad[:5]))))
    step3_checks.append(("How the categories were established (informational)", None, "; ".join(f"{count} x {method}" for method, count in method_counts.most_common()) if method_counts else "no mapped sections"))
    step_defs.append((3, f"Verify: Each {report_label} Report section maps to {'a valid' if not is_ratings else 'an approved'} KeyedData category", step3_checks))

    #---- Step 4: all section identifiers are traceable to KeyedData records ----
    step4_checks = []
    step4_checks.append((f"Every {unit_word} ID has a KeyedData record location (line number in the KeyedData file)", n_sections > 0 and all(r_["kd_line"] is not None for r_ in section_records),
                         f"{sum(1 for r_ in section_records if r_['kd_line'] is not None)} of {n_sections} located"))
    step4_checks.append((f"Every {unit_word} ID is found in the exported output file at the expected line, titled like its KeyedData {unit_word}", n_sections > 0 and len(trace_bad) == 0,
                         f"{n_sections - len(trace_bad)} of {n_sections} traced" + ("" if len(trace_bad) == 0 else "; not traced: " + " | ".join(f"{r_['id']} ({r_['trace_detail']})" for r_ in trace_bad[:5]))))
    labels_in_scope = sum(r_["rows"] for r_ in section_records)
    labels_traced = sum(r_["rows_traced"] for r_ in section_records)
    step4_checks.append((f"Row-level traceability (informational; rows with no KeyedData row keep their PLEXOS label)", None, f"{labels_traced} of {labels_in_scope} labels trace to a KeyedData row"))
    step_defs.append((4, f"Verify: All {report_label} Report section identifiers are traceable to KeyedData records", step4_checks))

    #---- Step 5: unmapped sections generate exceptions ----
    exception_back = read_csv_rows(exceptions_path, as_dicts=True)
    unmapped_expected = [r_["id"] for r_ in unmapped_records]
    unmapped_in_report = [row_["Section"] for row_ in exception_back if row_["Exception Type"].startswith("Unmapped section")]
    #an independent scan of the KeyedData file (not the parser above): the section titles it really contains
    independent_ids = set()
    for index, cells in enumerate(c_rows):
        non_blank_cells = [c for c in cells if c.strip() != ""]
        if len(non_blank_cells) == 1 and len(cells) > 0 and cells[0].strip() != "":
            if is_ratings:
                next_cells = c_rows[index + 1] if index + 1 < len(c_rows) else []
                if any(month_re.match(c) for c in next_cells):
                    independent_ids.add(cells[0].strip())
            elif "=" not in cells[0] and title_re.match(cells[0]):
                independent_ids.add(title_re.match(cells[0]).group(1).lower())
    step5_checks = []
    step5_checks.append((f"{len(unmapped_expected)} unmapped {unit_word}(s) found", None, "none: the exception report is still written (without unmapped rows)" if len(unmapped_expected) == 0 else list_ids(unmapped_expected)))
    step5_checks.append((f"Every unmapped {unit_word} has an exception row in the exception report (report read back from disk)", sorted(unmapped_expected) == sorted(unmapped_in_report),
                         f"{len(unmapped_in_report)} unmapped exception row(s) in the file for {len(unmapped_expected)} unmapped {unit_word}(s)"))
    step5_checks.append((f"The exception report holds exactly the failed {unit_word} checks (unmapped / structure / category / traceability / expected {unit_word} missing)", len(exception_back) == len(exception_rows),
                         f"{len(exception_back)} exception row(s) in the file, {len(exception_rows)} expected"))
    step5_checks.append(("Every exception row identifies the test case, report, section and exception type", all(row_["Test Case"] != "" and row_["Report"] != "" and row_["Section"] != "" and row_["Exception Type"] != "" for row_ in exception_back),
                         f"{len(exception_back)} row(s) checked"))
    if is_ratings:
        destinations_in_file = all(r_["kd_key"] in independent_ids for r_ in mapped_records)
        step5_checks.append((f"Every KeyedData destination really exists in the KeyedData file (independent line scan found {len(independent_ids)} sections)", destinations_in_file,
                             "all destinations found" if destinations_in_file else "a destination is not in the file: " + list_ids([r_["kd_key"] for r_ in mapped_records if r_["kd_key"] not in independent_ids])))
    else:
        independent_unmapped = sorted(r_["id"] for r_ in section_records if r_["id"].lower() not in independent_ids)
        step5_checks.append((f"The unmapped {unit_word}s agree with an independent scan of the KeyedData file's titles ({len(independent_ids)} {unit_word}s found by the scan)", independent_unmapped == sorted(unmapped_expected),
                             f"scan: {len(independent_unmapped)} {unit_word}(s) not in the KeyedData file; test: {len(unmapped_expected)}"))
    step_defs.append((5, f"Verify: Unmapped {report_label} Report sections generate exceptions (written to an exception report)", step5_checks))

    #---- Step 6: results are exported automatically ----
    mapping_back = read_csv_rows(mapping_results_path, as_dicts=True)
    run_started_limit = plexos4_run_started - 2.0
    exports_fresh = all(os.path.isfile(p_) and os.path.getmtime(p_) >= run_started_limit for p_ in (mapping_results_path, exceptions_path, output_path, verify_path))
    output_lines_back = sum(1 for _ in open(output_path, "r", encoding="utf-8-sig", errors="replace"))
    step6_checks = []
    step6_checks.append((f"{TC_ID} section mapping results exported and read back ({n_sections} {unit_word}s)", len(mapping_back) == n_sections and n_sections > 0, f"{len(mapping_back)} row(s) in {mapping_results_path}"))
    step6_checks.append(("Exception report exported and read back", len(exception_back) == len(exception_rows), f"{len(exception_back)} exception row(s) in {exceptions_path}"))
    step6_checks.append((f"The converted {report_label} Report (THE ONE OUTPUT CSV) is exported", os.path.isfile(output_path) and output_lines_back == len(b_rows), f"{output_lines_back} lines in {output_path} ({top_blank} blank lines at the top)"))
    step6_checks.append(("The verification workbook is saved and opens (with this report's Labels, RT Sim only and Section Mapping sheets)",
                         workbook_readable and all(f"{report_label} {s_}" in workbook_sheet_names for s_ in ("Labels", "RT Sim only", "Section Mapping")), f"{verify_path}: {len(workbook_sheet_names)} sheet(s)"))
    step6_checks.append(("All of these files were produced automatically in this run (no manual step; only the inputs block was answered)", exports_fresh, "every export is newer than the start of the run" if exports_fresh else "an export is older than the run"))
    for check_name, check_passed, check_detail in checks:
        step6_checks.append((f"Output file verification: {check_name}", check_passed, check_detail))
    step_defs.append((6, f"Verify: {report_label} Report section mapping validation results are exported automatically", step6_checks))

    #=================================================================
    #4.8 PRINT THE STEPS (PASS / FAIL for every check, for every step, and for the whole test case after every step)
    #=================================================================
    print_banner(f"{TC_NAME.upper()}: STEP BY STEP", blank_line=True)
    print(f"{n_sections} {unit_word}(s) in the verification scope ({len(sections_outside_scope)} {unit_word}(s) of the PLEXOS file outside the scope are not counted: {list_ids(sections_outside_scope) if sections_outside_scope else 'none'})")
    step_summary_rows = []
    for step_number, step_title, step_check_list in step_defs:
        print()
        print("-" * 100)
        print(f"Step {step_number}: {step_title}")
        for check_name, check_status, check_detail in step_check_list:
            status_word = "INFO" if check_status is None else status_text[bool(check_status)]
            print(f"  [{status_word}] {check_name}")
            if check_detail != "":
                print(f"          {check_detail}")
        step_pass = all(bool(s_) for n_, s_, d_ in step_check_list if s_ is not None)
        step_summary_rows.append((step_number, step_title, status_text[step_pass]))
        record_step_result(step_number, step_pass, step_results, TC_NAME, status_text)
    test_case_pass = all(step_results.values())
    all_reports_pass[f"{TC_ID} {report_label}"] = test_case_pass
    test_step_tables[TC_NAME] = step_summary_rows
    write_csv_file(step_summary_path, ["Test Case", "Step", "Check", "Status"], [[TC_ID, step_number, step_title, step_status] for step_number, step_title, step_status in step_summary_rows] + [[TC_ID, "Overall", TC_TITLE, status_text[test_case_pass]]])
    print()
    print(f"STEP SUMMARY: {step_summary_path}")
    print(f"Section mapping results: {mapping_results_path}")
    print(f"Exception report:        {exceptions_path}")
    print()
    if test_case_pass:
        print(f"PASSED: {TC_NAME}")
    else:
        failed_steps = [s_ for s_ in sorted(step_results) if not step_results[s_]]
        print(f"Reason: step(s) {failed_steps} did not pass.")
        print(f"FAILED: {TC_NAME}")
    ws = summary_ws
    ws.append([])
    append_bold_row(ws, [f"{TC_NAME}: step results"])
    for step_number, step_title, step_status in step_summary_rows:
        ws.append([f"Step {step_number}: {step_title}", step_status])
        color_pass_fail(ws, ws.max_row, 2, step_status == "PASS", fill_same, fill_bad)
    append_bold_row(ws, [f"{TC_NAME}: OVERALL", status_text[test_case_pass]])
    color_pass_fail(ws, ws.max_row, 2, test_case_pass, fill_same, fill_bad)

    #=================================================================
    #4.9 SUPPORTING DETAIL (what was changed in the first column) - the same information as before, after the result
    #=================================================================
    print_banner(f"{report_label.upper()}: WHAT WAS CHANGED (every {unit_word} in the PLEXOS file)", blank_line=True)
    print(f"blank lines added at the top: {top_blank}")
    print(f"label cells in the first column looked at: {len(cands1)} | changed: {len(changed_positions)} | " + " | ".join(f"{kind_name}: {kind_changed[kind_name]}" for kind_name in ["title", "header", "heading", "row", "sub"] if kind_changed[kind_name] > 0))
    for status_name in status_columns:
        if all_status[status_name] > 0:
            print(f"   {all_status[status_name]:>7}  {status_name}")
    print("\nWhy labels were changed (most frequent first):")
    for reason, n in why_changed.most_common(12):
        print(f"   [{n:>6}] {reason}")
    print()
    print(f"{report_label.upper()}: RESULT PER VERIFICATION {unit_word.upper()}")
    for table_number in verify_tables_present:
        if table_number not in summary:
            if not any(t["table"] == table_number for t in rt_only_tables_by_pass[2]):
                print(f"{unit_word} {table_number}: not in the PLEXOS file or the RT Sim file")
            continue                                                  #a table / section that only RT Sim has is printed below
        parts = [f"{summary[table_number][status_name]} {status_name}" for status_name in status_columns if summary[table_number][status_name] > 0]
        print(f"{unit_word} {table_number}: {sum(summary[table_number].values())} labels")
        for part in parts:
            print(f"      {part}")
        print(f"      RT Sim-only rows not in the output: {rt_only_count[table_number]} | output differs from RT Sim: {problem_count[table_number]}")
    for item in rt_only_tables_by_pass[2]:
        print(f"{unit_word} {item['table']}: exists only in RT Sim ({item['n_rows']} rows) - not added")


#5. FINISH THE VERIFICATION WORKBOOK AND SAVE IT

ws = summary_ws
ws.append([])
append_bold_row(ws, ["Colors on the Labels sheets"])
for color_name, fill_value, meaning in [("green", fill_same, "already identical to RT Sim"), ("blue", fill_format, "changed: padding / spacing / case / punctuation"), ("yellow", fill_units, "changed: only the text in parentheses (units)"),
                                        ("orange", fill_renamed, "changed: different name / text (override, alias, similarity, day row, effluent heading, rating section)"), ("grey", fill_kept, "kept as in PLEXOS (no RT Sim row / table or section not in RT Sim / option off)"),
                                        ("red", fill_bad, "output label differs from the RT Sim label it was matched to (should never appear)")]:
    ws.append([color_name, meaning])
    ws.cell(row=ws.max_row, column=1).fill = fill_value
set_openpyxl_column_widths(ws, [70, 60, 70] + [20] * 10)

ws = add_header_sheet(wb, "Label Formats", ["Report", "File", "Kind of label", "How it is written", "Count"], fill_header, font_header)
for row_values in formats_rows:
    ws.append(row_values)
set_openpyxl_column_widths(ws, [12, 26, 24, 60, 12])

wb.save(verify_path)
final_workbook_sheets, final_workbook_problems = read_workbook_sheet_names(verify_path)
final_workbook_ok = len(final_workbook_sheets) > 0

print_banner("OVERALL (PlexOS-4 section mapping tests)", blank_line=True)
for test_name, test_steps in test_step_tables.items():
    print(f"{test_name}")
    for step_number, step_title, step_status in test_steps:
        print(f"   Step {step_number}: {step_status}  {step_title}")
    print(f"   -> {'PASSED' if all(s_ == 'PASS' for n_, t_, s_ in test_steps) else 'FAILED'}")
print()
for report_name in all_reports_pass:
    print(f"{report_name}: " + ("PASSED (all 6 steps)" if all_reports_pass[report_name] else "!! FAILED - at least one step did not pass"))
print()
print("FILES")
for report in output_paths:
    print(f"THE ONE OUTPUT CSV ({'Ratings' if report == 'rat' else 'Standard'}) : {output_paths[report]}   ({int(TOP_BLANK_LINES.get(report, 0))} blank lines at the top)")
print(f"Verification workbook (both reports) : {verify_path}   (" + ", ".join(f"{k}: {v} labels" for k, v in overall_labels_rows.items()) + f"; spaces shown as dots; saved and opens: {status_text[final_workbook_ok]})")
print(f"Results folder: {Path(OUTPUT_FOLDER).resolve()}")
```

# PlexOS-4.01 (Standard Report) and 4.04 (Ratings Report) Section Mapping (2)

```python
#PURPOSE
#1. SETTINGS

KEYEDDATA_WORKBOOK_PATH = r""          #the OpRep_01 .xlsm you created KeyedData with. Leave "" to be asked for it when the script runs.
STANDARD_OUTPUT_PATH = r""             #"" = the Standard output written by the previous cell (variable output_paths), else the file in OUTPUT_FOLDER, else you are asked
RATINGS_OUTPUT_PATH = r""              #"" = same for the Ratings output
KEYEDDATA_SHEET_NAME = "KeyedData"
KEYEDDATA_KEY_COLUMN = ""              #"" = detected (the column that holds the most Std_ / Rat_ keys, normally C = PrimaryKey). Or a column letter, e.g. "C"
KEYEDDATA_MAX_COLUMNS = 12             #only the first columns of KeyedData are read (the key column is in there; the month values are not needed)
MACRO_EFFLUENT_COUNT = None            #None = no limit assumed (a table 32 block is only reported when KeyedData lacks it). Or the "Number of Effluent Types" of Control, e.g. 8
MACRO_FIRST_READ_ROW = 26              #the first row of the report file the macro reads (rows above it are never keyed). Taken from the shape of your KeyedData
MACRO_RATING_SECTION_NAMES = ["On-peak Mon-Sat", "Off-peak 23-6 Mon-Sun", "Sunday On-peak", "ATC Mon-Sun"]   #the section names the macro recognises
MAX_SUGGESTIONS = 500                  #"closest KeyedData key" is looked up for this many missing rows (it is the slow part)
REPORT_FOLDER = ""                     #"" = the folder of the Standard output (or of the Ratings output when only that one is checked)
REPORT_NAME = "KeyedData_Row_Check.xlsx"
MAX_KEYEDDATA_ONLY_ROWS = 50000        #rows of the 'KeyedData only' sheet
LOOSE_MATCHING = True                  #True = after the exact tiers, also try normalized / contained / other-table / elsewhere-in-the-workbook matches (see the header)
SEARCH_OTHER_SHEETS = True             #True = the 'elsewhere in file' search also reads every other tab of the .xlsm (slower; set False if the workbook is huge)
SEARCH_MAX_COLUMNS = 40                #columns read on the KeyedData tab and on the other tabs for that search
SEARCH_MAX_ROWS = 30000                #rows read per other tab
MIN_LOOSE_LENGTH = 6                   #a row value needs at least this many letters / digits to be matched loosely by its text
CONTEXT_MAX_COLUMNS = 4                #the first columns of the other tabs (A:D) that are read for the context search (table title / contract heading / day number / label)
CONTEXT_MAX_DAY_ROWS = 40              #a day row is looked for this many rows below its contract heading

month_re = re.compile(r"^\s*(?:Jan|Feb|Mar|Apr|May|Jun|Jul|Aug|Sep|Oct|Nov|Dec|Tot)-\d{4}\s*$", re.IGNORECASE)
title_re = re.compile(r"^\s*\(\s*(\d+[a-z]?)\s*\)\s*(.*)$", re.IGNORECASE)
norm_re = re.compile(r"[^0-9a-z]+")                  #everything that is not a letter or a digit (the full normalisation with the NFKD / 'Â' handling is normalize_key_text; the whitespace squeeze is squeeze_text)
report_labels = {"std": "Standard", "rat": "Ratings"}
sub_row_tables = ["3", "6"]                         #plant row + indented fuel rows
effluent_group_tables = ["14", "73"]                #effluent type heading + plant rows
day_tables = ["27", "28"]                           #contract heading + day rows
effluent_total_table = "32"                         #one block per effluent type
heading_tables = effluent_group_tables + day_tables + [effluent_total_table]
status_names = ["Found", "Found - text differs", "Found - other section no.", "Found - normalized", "Found - contained in key", "Found - other table", "Found - elsewhere in file", "Missing", "No key"]
not_exact_statuses = ["Found - text differs", "Found - other section no.", "Found - normalized", "Found - contained in key", "Found - other table", "Found - elsewhere in file"]
wrong_place_statuses = ["Found - other section no.", "Found - other table", "Found - elsewhere in file"]
unconfirmed_statuses = ["Found - other table", "Found - elsewhere in file"]   #the label exists, but not under the key the output needs: it is reported, not counted as a row of this table


#2. THE TWO OUTPUT REPORTS OF THE PREVIOUS CELL

report_paths = OrderedDict()
previous_paths = globals().get("output_paths", {})
for report in ["std", "rat"]:
    report_label = report_labels[report]
    answer = STANDARD_OUTPUT_PATH if report == "std" else RATINGS_OUTPUT_PATH
    if answer.strip() == "" and report in previous_paths:
        answer = previous_paths[report]
    if answer.strip() == "" and "OUTPUT_FOLDER" in globals() and "OUTPUT_NAMES" in globals():
        answer = os.path.join(OUTPUT_FOLDER, OUTPUT_NAMES[report])
    while True:
        answer = strip_path_text(answer)
        if answer.lower() == "skip" or os.path.isfile(answer):
            break
        if answer != "":
            print(f"The {report_label} output report was not found: {answer}")
        answer = input(f"Paste the full path of the OUTPUT {report_label} report (.csv) that was loaded into OpRep_01, or type SKIP to leave the {report_label} report out: ")
    if answer.lower() != "skip":
        report_paths[report] = answer
if len(report_paths) == 0:
    raise ValueError("No output report to check.")


#3. ASK FOR THE .xlsm THAT HOLDS THE KeyedData, READ ITS KeyedData TAB

workbook_answer = KEYEDDATA_WORKBOOK_PATH
while True:
    if workbook_answer.strip() == "":
        workbook_answer = input("\nPaste the full path of the OpRep_01 .xlsm you used to create KeyedData from the output reports above: ")
    workbook_answer = strip_path_text(workbook_answer)
    if not os.path.isfile(workbook_answer):
        print(f"   File not found: {workbook_answer}" if workbook_answer != "" else "   Nothing was entered.")
        workbook_answer = ""
        continue
    if os.path.splitext(workbook_answer)[1].lower() not in (".xlsm", ".xlsx"):
        print("   Only .xlsm / .xlsx workbooks can be read (an .xlsb must be saved as .xlsm first).")
        workbook_answer = ""
        continue
    try:
        workbook_kd = openpyxl.load_workbook(workbook_answer, read_only=True, data_only=True)
    except PermissionError:
        print("   The workbook is open in Excel and cannot be read. Close it (or save a copy) and paste the path again.")
        workbook_answer = ""
        continue
    except Exception as error:
        print(f"   The workbook could not be opened ({error}).")
        workbook_answer = ""
        continue
    sheet_match = [name for name in workbook_kd.sheetnames if name.strip().lower() == KEYEDDATA_SHEET_NAME.lower()]
    if len(sheet_match) == 0:
        print(f"   The workbook has no '{KEYEDDATA_SHEET_NAME}' tab. Its tabs: {', '.join(workbook_kd.sheetnames)}")
        workbook_kd.close()
        workbook_answer = ""
        continue
    break
keyeddata_path = workbook_answer
print(f"\nReading the '{sheet_match[0]}' tab of {keyeddata_path} ...")
kd_rows = []
for r_index, row in enumerate(workbook_kd[sheet_match[0]].iter_rows(min_row=1, max_col=max(KEYEDDATA_MAX_COLUMNS, SEARCH_MAX_COLUMNS), values_only=True), start=1):
    kd_rows.append((r_index, row))
other_cells = []                         #(tab, cell address, text) of every text cell outside the key column (other columns of the KeyedData tab are added below)
ctx_cells = defaultdict(list)            #tab -> (row, column index, text) of the first columns, numbers included (for the context search)
sheets_scanned = []
if LOOSE_MATCHING and SEARCH_OTHER_SHEETS:
    for sheet_name in workbook_kd.sheetnames:
        if sheet_name == sheet_match[0]:
            continue
        cell_count = 0
        try:
            for r_index, row in enumerate(workbook_kd[sheet_name].iter_rows(min_row=1, max_row=SEARCH_MAX_ROWS, max_col=SEARCH_MAX_COLUMNS, values_only=True), start=1):
                for c_index, value in enumerate(row):
                    if c_index < CONTEXT_MAX_COLUMNS and value is not None and not isinstance(value, bool):
                        #the context search needs numbers too: Excel stores a day label such as 31 as a number, not as text
                        if isinstance(value, (int, float)):
                            ctx_text = str(int(value)) if float(value).is_integer() else str(value)
                        else:
                            ctx_text = str(value)
                        if ctx_text.strip() != "":
                            ctx_cells[sheet_name].append((r_index, c_index, ctx_text))
                    if isinstance(value, str) and value.strip() != "":
                        other_cells.append((sheet_name, get_column_letter(c_index + 1) + str(r_index), value))
                        cell_count += 1
        except Exception as error:
            print(f"   (tab '{sheet_name}' could not be searched: {error})")
        sheets_scanned.append((sheet_name, cell_count))
workbook_kd.close()

#the key column = the column that holds the most Std_ / Rat_ keys (normally C = PrimaryKey)
column_counts = Counter()
for r_index, row in kd_rows:
    for c_index, value in enumerate(row):
        if isinstance(value, str) and (value.strip().lower().startswith("std_") or value.strip().lower().startswith("rat_")):
            column_counts[c_index] += 1
if KEYEDDATA_KEY_COLUMN.strip() != "":
    key_index = column_index_from_string(KEYEDDATA_KEY_COLUMN.strip().upper()) - 1
elif len(column_counts) > 0:
    key_index = column_counts.most_common(1)[0][0]
else:
    raise ValueError("The KeyedData tab holds no Std_ / Rat_ keys in its first columns. Was the macro run? (If the keys are formulas, open the workbook in Excel, save it, and run this cell again.)")
key_letter = get_column_letter(key_index + 1)
key_header = ""
for r_index, row in kd_rows[:3]:
    if key_index < len(row) and isinstance(row[key_index], str) and not (row[key_index].strip().lower().startswith("std_") or row[key_index].strip().lower().startswith("rat_")) and row[key_index].strip() != "":
        key_header = row[key_index].strip()
        break

if LOOSE_MATCHING:
    for r_index, row in kd_rows:
        for c_index, value in enumerate(row):
            if c_index != key_index and isinstance(value, str) and value.strip() != "":
                other_cells.append((sheet_match[0], get_column_letter(c_index + 1) + str(r_index), value))
other_norm = [normalize_key_text(text) for tab_name, address, text in other_cells]
ctx_rows = {}                            #tab -> {row: [(column index, text)]}
ctx_row_list = {}                        #tab -> sorted rows
ctx_titles = {}                          #tab -> [(row, column index, table number)] of the table titles '( 27) ...'
for tab_name in ctx_cells:
    rows_here = defaultdict(list)
    titles_here = []
    for r_index, c_index, text in ctx_cells[tab_name]:
        rows_here[r_index].append((c_index, text))
        title_found = title_re.match(text)
        if title_found and "=" not in text:
            titles_here.append((r_index, c_index, title_found.group(1).lower()))
    ctx_rows[tab_name] = rows_here
    ctx_row_list[tab_name] = sorted(rows_here)
    ctx_titles[tab_name] = sorted(titles_here)

#every KeyedData key, split the way it is built:  <Std|Rat> _ <section number> _ <row value>
kd_keys = []                                        #(KeyedData row, key text) of every row that has a key
for r_index, row in kd_rows:
    value = row[key_index] if key_index < len(row) else None
    if value is None or str(value).strip() == "":
        continue
    kd_keys.append((r_index, str(value)))
kd_records = []
kd_nfull = defaultdict(list)                        #normalized full key -> records
kd_by_section = defaultdict(list)                   #(prefix, section) -> records
kd_exact = defaultdict(list)                        #full key, lower case (like XMATCH) -> KeyedData records
kd_sq = defaultdict(list)                           #(prefix, section, row value without any space) -> records
kd_sections = Counter()                             #(prefix, section) -> number of KeyedData rows
kd_pool = defaultdict(list)                         #(prefix, section) -> distinct key texts (for the 'closest key' hint)
kd_pool_seen = defaultdict(set)
kd_days = {}                                        #(prefix, section, contract without spaces) -> highest Day number
kd_kinds = Counter()
for r_index, text in kd_keys:
    kd_kinds[text.strip().split("_")[0].lower() + "_"] += 1
    parts = text.split("_", 2)
    if len(parts) < 3 or parts[0].strip().lower() not in ("std", "rat"):
        continue
    record = {"row": r_index, "text": text, "prefix": parts[0].strip().lower(), "section": parts[1], "section_l": parts[1].strip().lower(), "rest": parts[2],
              "sq": squeeze_text(parts[2]), "used": False}
    record["nrest"] = normalize_key_text(parts[2])
    record["nfull"] = normalize_key_text(text)
    kd_nfull[record["nfull"]].append(record)
    kd_by_section[(record["prefix"], record["section_l"])].append(record)
    kd_records.append(record)
    kd_exact[text.lower()].append(record)
    kd_sq[(record["prefix"], record["section_l"], record["sq"])].append(record)
    kd_sections[(record["prefix"], record["section_l"])] += 1
    pool_key = (record["prefix"], record["section_l"])
    if text not in kd_pool_seen[pool_key]:
        kd_pool_seen[pool_key].add(text)
        kd_pool[pool_key].append(text)
    day_match = re.match(r"^(.*)_(?:day|d)?0*(\d{1,2})$", record["sq"])
    if day_match:
        day_key = (record["prefix"], record["section_l"], day_match.group(1))
        kd_days[day_key] = max(kd_days.get(day_key, 0), int(day_match.group(2)))
if LOOSE_MATCHING and sheets_scanned:
    print("   other tabs searched: " + ", ".join(f"{name} ({n} text cells)" for name, n in sheets_scanned))
print(f"   key column: {key_letter}" + (f" ('{key_header}')" if key_header != "" else "") + f" | rows with a key: {len(kd_keys)} | " + " | ".join(f"{kind} {n}" for kind, n in kd_kinds.most_common(6)))


#4. READ THE OUTPUT REPORTS: every data row -> the key the macro writes into KeyedData
#(the macro trims ordinary spaces from the row value; table 32 is the exception and keeps the raw cell text)

records = OrderedDict()                  #report -> list of rows
structure_issues = OrderedDict()         #report -> list of lines the macro would not recognise
file_lines = OrderedDict()
first_title_line = OrderedDict()         #report -> line of the first table / section title
for report in report_paths:
    is_ratings = report == "rat"
    report_label = report_labels[report]
    prefix_text = "Rat" if is_ratings else "Std"
    rows = read_csv_rows(report_paths[report])
    file_lines[report] = len(rows)
    first_title_line[report] = None
    header_at = [month_header_position(cells, month_re) for cells in rows]       #position of the first month cell of a month-header row (only blanks / numbers / "index" before it)

    report_records = []
    issues = []
    current = None
    table_id = ""
    offset = None
    group = ""
    plant = ""
    block_number = 0
    block_lines = []
    sub_plant = ""
    sub_child = False
    for i, cells in enumerate(rows):
        non_empty = [c for c in cells if c.strip() != ""]
        if len(non_empty) == 0:
            continue
        first = cells[0]
        #---- table title (Standard '( n ) text', Ratings: a one-cell line directly above a month-header row) ----
        is_title = False
        if len(non_empty) == 1 and first.strip() != "":
            if is_ratings:
                is_title = i + 1 < len(rows) and header_at[i + 1] is not None
            else:
                is_title = bool(title_re.match(first)) and "=" not in first
        if is_title:
            if first_title_line[report] is None:
                first_title_line[report] = i + 1
            if is_ratings:
                current = first.strip()
                table_id = current
                if current not in MACRO_RATING_SECTION_NAMES:
                    issues.append({"line": i + 1, "text": first, "problem": f"section title '{current}' is not one of the section names the macro recognises ({', '.join(MACRO_RATING_SECTION_NAMES)}): the macro keys its rows under the previous section"})
            else:
                current = title_re.match(first).group(1)
                table_id = current.lower()
                if first.find("(") + 1 != 2:
                    issues.append({"line": i + 1, "text": first, "problem": f"table title of table {current} does not have its '(' as the 2nd character (' ( 1) ...'): the macro does not recognise it as a table title and keys the rows under the previous table number"})
            offset = None
            group = ""
            plant = ""
            block_number = 0
            sub_plant = ""
            sub_child = False
            continue
        if current is None:
            continue
        if header_at[i] is not None:                      #month-header row
            offset = header_at[i]
            continue
        #---- the label cell: the cell just left of the first month column (the PLEXOS export puts an index number in column A of tables 3 / 14 / 32) ----
        label_col = offset - 1 if offset is not None and offset >= 1 else 0
        if len(non_empty) == 1 and non_empty[0].strip().isdigit():
            continue                                      #an index-only line of the PLEXOS export
        label_text = cells[label_col] if label_col < len(cells) else ""
        label_position = label_col
        values_right = any(c.strip() != "" for c in cells[label_col + 1:])
        if label_text.strip() == "" and len(non_empty) == 1:
            label_text = non_empty[0]                     #a lone cell that is not in the label column (a heading written in column A)
            label_position = cells.index(label_text)
            values_right = False
        label_only = label_text.strip() != "" and not values_right
        label_trim = label_text.strip(" ")                #what the macro's Trim keeps: ordinary spaces go, a non-breaking space stays
        label_clean = label_text.strip()                  #only used to tell a blank label from a real one
        shown = show_spaces(label_text.replace("\u00a0", "[nbsp]"))
        row_record = {"report": report_label, "prefix": prefix_text, "section": current, "table": current, "line": i + 1, "label": label_text, "shown": shown if shown != "" else "(empty)",
                      "label_col": label_position, "rest": None, "alts": [], "kind": "row", "block": 0, "note": "", "contract": "", "day": 0, "group_bad": False}

        if not is_ratings and table_id in heading_tables and label_only:
            #---- a heading line: table 32 effluent block / table 14, 73 effluent heading / table 27, 28 contract ----
            if table_id == effluent_total_table:
                block_number += 1
                block_lines.append(i + 1)
                row_record["kind"] = "table 32 block"
                row_record["block"] = block_number
                row_record["rest"] = label_text                #table 32: the macro keeps the raw cell text (a leading space is part of the key)
                report_records.append(row_record)
            elif table_id in effluent_group_tables:
                group = label_trim
                heading_ok, heading_name = effluent_heading_check(label_text, table_id)
                if not heading_ok:
                    issues.append({"line": i + 1, "text": label_text, "problem": f"table {current} heading '{label_text.strip()}' does not start with '{heading_name}': the macro takes it for a plant row instead of an effluent heading"})
            else:
                plant = label_trim
            continue
        if not is_ratings and table_id == effluent_total_table:
            continue                                      #the value line of a table 32 block: the macro keys the heading, not this line
        if label_clean == "":
            row_record["note"] = "blank label: the macro skips blank-label lines, so this line has no key"
        elif is_ratings:
            row_record["rest"] = label_trim
        elif table_id in sub_row_tables:
            if label_text[:5].strip() == "":
                row_record["kind"] = "fuel sub-row"
                row_record["rest"] = plant + "_" + label_trim
            else:
                plant = label_trim
                row_record["kind"] = "plant row"
                row_record["rest"] = plant + "_Total"
        elif table_id in effluent_group_tables:
            row_record["rest"] = group + "_" + label_trim
        elif table_id in day_tables:
            row_record["kind"] = "day row"
            row_record["rest"] = plant + "_Day" + label_trim.zfill(2)
            row_record["contract"] = plant
            row_record["day"] = int(label_trim) if label_trim.isdigit() else 0
        else:
            row_record["rest"] = label_trim
            if re.sub(r"[a-z]+$", "", table_id) in sub_row_tables:
                #tables 3a / 3b (a plant line, indented fuel lines, a block 'Total'): KeyedData holds them as  <plant> - <label>_Total,  <plant>_<label>  or  <label>_Total
                indented = len(label_text) - len(label_text.lstrip(" ")) >= 2
                if indented or (label_trim.lower() == "total" and sub_child and sub_plant != ""):
                    row_record["alts"] = [sub_plant + " - " + label_trim + "_Total", sub_plant + "_" + label_trim, label_trim + "_Total"]
                    row_record["kind"] = "fuel / block line"
                    sub_child = True
                else:
                    sub_plant = label_trim
                    row_record["alts"] = [label_trim + "_Total"]
                    row_record["kind"] = "plant line"
                    sub_child = False
        if not is_ratings and table_id in effluent_group_tables and row_record["rest"] is not None and group != "":
            if not effluent_heading_check(group, table_id)[0]:
                row_record["group_bad"] = True
        report_records.append(row_record)
    if len(block_lines) > 0:
        if MACRO_EFFLUENT_COUNT is not None and len(block_lines) > MACRO_EFFLUENT_COUNT:
            issues.append({"line": block_lines[MACRO_EFFLUENT_COUNT], "text": "", "problem": f"table 32 has {len(block_lines)} effluent blocks but the macro reads only {MACRO_EFFLUENT_COUNT} (Control: Number of Effluent Types); block {MACRO_EFFLUENT_COUNT + 1} onward cannot be keyed"})
        steps = [block_lines[k + 1] - block_lines[k] for k in range(len(block_lines) - 1)]
        if any(step != 4 for step in steps):
            issues.append({"line": block_lines[0], "text": "", "problem": f"table 32 blocks are not exactly 4 lines apart (steps: {sorted(set(steps))}); the macro steps 4 lines per effluent block"})
    records[report] = report_records
    structure_issues[report] = issues
    print(f"{report_label}: {file_lines[report]} lines read, {len(report_records)} data rows, {sum(1 for r in report_records if r['rest'] is not None)} with a key | first title on line {first_title_line[report]}")


#5. LOOK EVERY KEY UP IN KeyedData

blank_owner = {}                         #prefix -> the one output table / section that owns the blank-section rows of KeyedData
for report in records:
    out_count = Counter()
    for rec in records[report]:
        if rec["rest"] is not None:
            rec["sq"] = squeeze_text(rec["rest"])
            out_count[(rec["section"].strip().lower(), rec["sq"])] += 1
    for rec in records[report]:
        rec["kd_rows"] = []
        rec["out_count"] = 0
        rec["kd_count"] = 0
        rec["reason"] = ""
        rec["closest"] = ""
        rec["matched_key"] = ""
        rec["kd_section"] = None
        rec["tier"] = 0
        if rec["rest"] is None:
            rec["key"] = None
            rec["status"] = "No key"
            continue
        rec["key"] = build_key(rec["prefix"], rec["section"], rec["rest"])
        prefix_l = rec["prefix"].lower()
        own_l = rec["section"].strip().lower()
        digits_l = re.sub(r"[a-z]+$", "", own_l) if prefix_l == "std" else own_l
        own_has_rows = kd_sections.get((prefix_l, own_l), 0) > 0
        candidates = [rec["rest"]] + rec["alts"]
        hit_list = None
        tier = 0
        matched_candidate = rec["rest"]
        #tier 1: the key exactly as the macro writes it
        found_list, found_candidate = find_first_hit(candidates, lambda c: build_key(rec["prefix"], rec["section"], c).lower(), kd_exact)
        if found_list is not None:
            hit_list, tier, matched_candidate = found_list, 1, found_candidate
        #tier 2: the CSV was read as ANSI instead of UTF-8 (a non-breaking space becomes 'Â' + a space)
        if hit_list is None:
            found_list, found_candidate = find_first_hit(candidates, lambda c: build_key(rec["prefix"], rec["section"], c).encode("utf-8").decode("cp1252", "replace").lower(), kd_exact)
            if found_list is not None:
                hit_list, tier, matched_candidate = found_list, 2, found_candidate
        #tier 3: the same row value once every space / hidden character is ignored
        if hit_list is None:
            found_list, found_candidate = find_first_hit(candidates, lambda c: (prefix_l, own_l, squeeze_text(c)), kd_sq)
            if found_list is not None:
                hit_list, tier, matched_candidate = found_list, 3, found_candidate
        #tier 4: the same row value under another section number - only when KeyedData has NO row at all under this section number,
        #and only a KeyedData row that no other output row has used
        if hit_list is None and not own_has_rows:
            sections_to_try = [""]
            if digits_l != own_l and digits_l != "":
                sections_to_try.append(digits_l)
            for section_try in sections_to_try:
                if section_try == "" and blank_owner.get(prefix_l, own_l) != own_l:
                    continue                                   #all blank-section rows belong to ONE table: the one the macro was in when it started reading
                for candidate in candidates:
                    free = [r for r in kd_sq.get((prefix_l, section_try, squeeze_text(candidate)), []) if not r["used"]]
                    if len(free) > 0:
                        hit_list = free
                        tier = 4
                        matched_candidate = candidate
                        break
                if hit_list is not None:
                    if section_try == "":
                        blank_owner[prefix_l] = own_l
                    break
                #tier 4b: a DAY row (tables 27 / 28) found by contract + day NUMBER, however KeyedData spells the day
        #("Day31", "Day 31", "Day031", "D31", "31"), in its own table first, then under the number without the letter / a blank number
        if hit_list is None and rec["kind"] == "day row" and rec["day"] > 0:
            day_contract_norm = normalize_key_text(rec["contract"])
            day_pattern = re.compile(r"^(?:day|d)?0*" + str(rec["day"]) + r"$")
            if day_contract_norm != "":
                for section_try in [own_l, digits_l, ""]:
                    free = [r for r in kd_by_section.get((prefix_l, section_try), [])
                            if r["nrest"].startswith(day_contract_norm) and day_pattern.match(r["nrest"][len(day_contract_norm):]) and not r["used"]]
                    if len(free) > 0:
                        hit_list = free
                        tier = 3 if section_try == own_l else 4      #3 = "Found - text differs", 4 = "Found - other section no."
                        matched_candidate = rec["rest"]
                        break
        #---- loose tiers 5 - 8: only when the exact tiers found nothing ----
        where_text = ""
        loose_status = ""
        if hit_list is None and LOOSE_MATCHING:
            ncands = [normalize_key_text(c) for c in candidates]
            #tier 5: the same key once only letters and digits are compared
            found_list, found_candidate = find_first_hit(candidates, lambda c: normalize_key_text(build_key(rec["prefix"], rec["section"], c)), kd_nfull)
            if found_list is not None:
                hit_list, tier, matched_candidate = found_list, 5, found_candidate
            #tier 6: the row value is part of a key of the same table / section (own number, the number without the letter, or the blank number)
            if hit_list is None:
                for k, ncand in enumerate(ncands):
                    if len(ncand) < MIN_LOOSE_LENGTH:
                        continue
                    for section_try in [own_l, digits_l, ""]:
                        found_here = [r for r in kd_by_section.get((prefix_l, section_try), []) if ncand in r["nrest"]]
                        if len(found_here) > 0:
                            hit_list = found_here
                            tier = 6
                            matched_candidate = candidates[k]
                            break
                    if hit_list is not None:
                        break
            #tier 7: the row value is part of a key of ANOTHER table / section (same Std / Rat family first)
            if hit_list is None:
                for k, ncand in enumerate(ncands):
                    if len(ncand) < MIN_LOOSE_LENGTH:
                        continue
                    found_any = [r for r in kd_records if r["prefix"] == prefix_l and ncand in r["nrest"]] or [r for r in kd_records if ncand in r["nrest"]]
                    if len(found_any) > 0:
                        hit_list = found_any
                        tier = 7
                        matched_candidate = candidates[k]
                        sections_seen = sorted(set((r["section"].strip() or "(blank)") for r in found_any))
                        where_text = f"{len(found_any)} KeyedData key(s) hold it, in section(s): " + ", ".join(sections_seen[:8]) + (" ..." if len(sections_seen) > 8 else "")
                        break
            #tier 8: the row value is in another column of the KeyedData tab or on another tab of the workbook
            if hit_list is None:
                for k, ncand in enumerate(ncands):
                    if len(ncand) < MIN_LOOSE_LENGTH:
                        continue
                    cell_hits = [(other_cells[n][0], other_cells[n][1], other_cells[n][2]) for n in range(len(other_cells)) if ncand in other_norm[n]]
                    if len(cell_hits) > 0:
                        tier = 8
                        matched_candidate = candidates[k]
                        where_text = f"{len(cell_hits)} cell(s) hold it, e.g. " + " | ".join(f"{tab_name}!{address}: '{text.strip()[:60]}'" for tab_name, address, text in cell_hits[:3])
                        loose_status = "Found - elsewhere in file"
                        break
            #tier 9: a SHORT label ('31', 'Total') or any row that tier 8 could not place: find it by its CONTEXT in the report copy of the workbook
            #(table title -> for a day row the contract heading and the day number below it, otherwise the label inside the table)
            if hit_list is None and tier != 8 and prefix_l == "std" and len(ctx_rows) > 0:
                ctx_label_norm = normalize_key_text(rec["label"])
                ctx_contract_norm = normalize_key_text(rec["contract"])
                ctx_found = ""
                for tab_name in ctx_rows:
                    title_list = ctx_titles[tab_name]
                    for t_index in range(len(title_list)):
                        t_row, t_col, t_id = title_list[t_index]
                        if t_id != own_l or ctx_found != "":
                            continue
                        end_row = title_list[t_index + 1][0] if t_index + 1 < len(title_list) else 10 ** 9
                        scan_rows = [x for x in ctx_row_list[tab_name] if t_row < x < end_row]
                        if rec["kind"] == "day row":
                            if ctx_contract_norm == "":
                                continue
                            heading_row = None
                            heading_col = 0
                            for x in scan_rows:
                                for c_, text_ in ctx_rows[tab_name][x]:
                                    if normalize_key_text(text_) == ctx_contract_norm:
                                        heading_row = x
                                        heading_col = c_
                                if heading_row is not None:
                                    break
                            if heading_row is None:
                                continue
                            for x in scan_rows:
                                if x <= heading_row:
                                    continue
                                if x > heading_row + CONTEXT_MAX_DAY_ROWS:
                                    break
                                next_heading = False
                                for c_, text_ in ctx_rows[tab_name][x]:
                                    if c_ == heading_col:
                                        try:
                                            float(text_)
                                            is_number_cell = True
                                        except ValueError:
                                            is_number_cell = False
                                        if not is_number_cell and text_.strip() != "" and not month_re.match(text_):
                                            next_heading = True
                                        elif is_number_cell and text_.strip() == str(rec["day"]):
                                            ctx_found = f"{tab_name}!{get_column_letter(c_ + 1)}{x}"
                                if next_heading or ctx_found != "":
                                    break
                        else:
                            if ctx_label_norm == "":
                                continue
                            for x in scan_rows:
                                for c_, text_ in ctx_rows[tab_name][x]:
                                    if ctx_found == "" and normalize_key_text(text_) == ctx_label_norm:
                                        ctx_found = f"{tab_name}!{get_column_letter(c_ + 1)}{x}"
                                if ctx_found != "":
                                    break
                if ctx_found != "":
                    tier = 8
                    loose_status = "Found - elsewhere in file"
                    matched_candidate = rec["rest"]
                    if rec["kind"] == "day row":
                        day_top = kd_days.get((prefix_l, own_l, squeeze_text(rec["contract"])), 0)
                        where_text = (f"the report copy in the workbook has it: {ctx_found} = day {rec['day']} of contract '{rec['contract']}' in table {rec['section'].strip()} (Day{rec['day']:02d})"
                                      + (f"; the KeyedData key column keys this contract only up to Day{day_top:02d}, so there is no key for Day{rec['day']:02d}" if day_top > 0 and rec["day"] > day_top else ""))
                    else:
                        where_text = f"the report copy in the workbook has it: {ctx_found} = '{rec['label'].strip()}' inside table {rec['section'].strip()}; the KeyedData key column has no key {rec['key']}"
        if tier == 8:
            rec["status"] = loose_status
            rec["tier"] = 8
            rec["matched_key"] = where_text[:300]
            rec["reason"] = "The row is not in a KeyedData key of its own table, but it is in another place of the workbook: " + where_text
            rec["out_count"] = out_count[(own_l, rec["sq"])]
            continue
        if hit_list is None:
            rec["status"] = "Missing"
            continue
        unused = [r for r in hit_list if not r["used"]]
        target = unused[0] if len(unused) > 0 else hit_list[0]
        target["used"] = True
        rec["tier"] = tier
        rec["kd_rows"] = [r["row"] for r in hit_list]
        rec["matched_key"] = target["text"]
        rec["kd_section"] = target["section"]
        rec["out_count"] = out_count[(own_l, rec["sq"])]
        rec["kd_count"] = len(hit_list)
        if tier == 1:
            rec["status"] = "Found"
        elif tier == 5:
            rec["status"] = "Found - normalized"
            rec["reason"] = "Same key once only letters and digits are compared. KeyedData: '" + show_hidden_chars(target["text"], True) + "'  output: '" + show_hidden_chars(build_key(rec["prefix"], rec["section"], matched_candidate), True) + "'"
        elif tier == 6:
            rec["status"] = "Found - contained in key"
            rec["reason"] = "The row value is part of a longer key of the same table: " + show_hidden_chars(target["text"], True) + (f"  ({len(hit_list)} such keys)" if len(hit_list) > 1 else "")
        elif tier == 7:
            rec["status"] = "Found - other table"
            rec["reason"] = "The row value is NOT keyed under " + rec["prefix"] + "_" + rec["section"].strip() + "; " + where_text
        elif tier == 4:
            rec["status"] = "Found - other section no."
            if target["section_l"] == "":
                rec["reason"] = (f"KeyedData holds this row value under a BLANK section number (KeyedData row {target['row']}): the macro never saw the title of this {'section' if prefix_l == 'rat' else 'table'}, "
                                 f"so the key reads {target['prefix'].capitalize()}__<row> instead of {rec['prefix']}_{rec['section'].strip()}_<row>")
            else:
                rec["reason"] = (f"KeyedData holds this row value under section '{target['section']}' (KeyedData row {target['row']}) instead of '{rec['section'].strip()}': the title of this table is not in the form the macro recognises, "
                                 f"so the rows are keyed under the previous table number")
        else:
            rec["status"] = "Found - text differs"
            if tier == 2:
                rec["reason"] = "KeyedData holds 'Â' in front of the non-breaking space: the macro read the CSV as ANSI, not UTF-8. The key text differs from the output (and from a workbook key typed with a real non-breaking space)."
            else:
                rec["reason"] = "Same row value once spaces / hidden characters are ignored. KeyedData: '" + show_hidden_chars(target["rest"], True) + "'  output: '" + show_hidden_chars(matched_candidate, True) + "'"

#KeyedData rows (Std_ / Rat_) that no output row used - for the report types that are being checked
checked_prefixes = []
if "std" in records:
    checked_prefixes.append("std")
if "rat" in records:
    checked_prefixes.append("rat")
kd_only = [r for r in kd_records if not r["used"] and r["prefix"] in checked_prefixes]
kd_only_by_table = Counter()
orphan_by_rest = defaultdict(list)
for r in kd_only:
    kd_only_by_table[r["prefix"].capitalize() + "_" + r["section"].strip()] += 1
    orphan_by_rest[(r["prefix"], r["sq"])].append(r)

#why a row is missing, and what KeyedData holds instead
suggestions_done = 0
for report in records:
    for rec in records[report]:
        if rec["status"] != "Missing":
            continue
        prefix_l = rec["prefix"].lower()
        own_l = rec["section"].strip().lower()
        reasons = []
        if rec["kind"] == "table 32 block" and MACRO_EFFLUENT_COUNT is not None and rec["block"] > MACRO_EFFLUENT_COUNT:
            reasons.append(f"table 32 block {rec['block']}: the macro reads only the first {MACRO_EFFLUENT_COUNT} effluent blocks")
        if rec["kind"] == "day row":
            day_top = kd_days.get((prefix_l, own_l, squeeze_text(rec["contract"])), 0)
            if day_top > 0 and rec["day"] > day_top:
                reasons.append(f"KeyedData holds this contract only up to Day{day_top:02d}: the macro keys {day_top} day rows per contract, the output has day {rec['day']}")
        if rec["group_bad"]:
            reasons.append("the effluent heading above this row is not in the form the macro recognises (see the 'Macro structure' sheet), so the macro keys the row under another heading")
        if rec["kind"] == "plant line":
            reasons.append("this is the plant line of a 3a / 3b block: KeyedData has no row of its own for it (only the fuel lines and the block Total, keyed as '<plant> - <label>')")
        orphans = orphan_by_rest.get((prefix_l, rec["sq"]), [])
        if len(orphans) > 0:
            reasons.append(f"KeyedData holds this row value only as a row nobody else uses: section '{orphans[0]['section'].strip() or '(blank)'}' (KeyedData row {orphans[0]['row']}). The macro read this row under another table")
        if kd_sections.get((prefix_l, own_l), 0) == 0:
            reasons.append(f"KeyedData has no '{rec['prefix']}_{rec['section'].strip()}' rows at all: the macro did not read this {'section' if prefix_l == 'rat' else 'table'} (title not recognised, or above the macro's first row, or KeyedData was built from another file)")
        if rec["label_col"] > 0:
            reasons.append(f"the label sits in column {get_column_letter(rec['label_col'] + 1)}, behind an index number in column A of this table; if the macro reads column A it cannot build this key")
        pool = kd_pool.get((prefix_l, own_l), [])
        if len(pool) > 0 and suggestions_done < MAX_SUGGESTIONS:
            close = get_close_matches(rec["key"], pool, n=1, cutoff=0.75)
            suggestions_done += 1
            if len(close) > 0:
                rec["closest"] = close[0]
                reasons.append("the closest key KeyedData has for this table is listed in 'Closest KeyedData key'")
        if len(reasons) == 0:
            reasons.append("KeyedData has this table, but not this row")
        if LOOSE_MATCHING:
            reasons.append(f"not found anywhere: searched every KeyedData key (normalized), the other columns of the KeyedData tab and {len(sheets_scanned)} other tab(s) = {len(other_cells)} text cells"
                           if len(rec["rest"]) > 0 and len(norm_re.sub("", rec["rest"].lower())) >= MIN_LOOSE_LENGTH else
                           "the row value is shorter than MIN_LOOSE_LENGTH letters / digits, so only exact / normalized matching was tried")
        rec["reason"] = "; ".join(reasons)


#6. THE CHECKS

checks = OrderedDict()
per_table = OrderedDict()
for report in records:
    report_label = report_labels[report]
    keyed_rows = [r for r in records[report] if r["key"] is not None]
    counts_by_status = Counter(r["status"] for r in records[report])
    n_missing = counts_by_status["Missing"]
    n_nokey = counts_by_status["No key"]
    missing_examples = [f"line {r['line']}: {r['key']}" for r in keyed_rows if r["status"] == "Missing"][:3]
    report_checks = []
    n_unconfirmed = sum(counts_by_status[name] for name in unconfirmed_statuses)
    report_checks.append((f"Every data row of the {report_label} output has a row in KeyedData (a label that only exists under another table / elsewhere in the file does not count)", n_missing == 0 and n_unconfirmed == 0,
                          f"{len(keyed_rows)} rows with a key: " + ", ".join(f"{counts_by_status[name]} {name.lower()}" for name in status_names if name not in ("No key",) and counts_by_status[name] > 0)
                          + (f" ({n_nokey} further line(s) with a blank label have no key)" if n_nokey else "") + ("; e.g. " + " | ".join(missing_examples) if missing_examples else "")))
    report_checks.append((f"Every table title / section title / effluent heading of the {report_label} output is in the form the macro recognises", len(structure_issues[report]) == 0,
                          f"{len(structure_issues[report])} line(s) the macro would not recognise" + ("; e.g. line " + str(structure_issues[report][0]["line"]) + ": " + structure_issues[report][0]["problem"][:170] if structure_issues[report] else "")))
    short = []
    seen_short = set()
    for r in keyed_rows:
        if r["status"] in ("Found", "Found - text differs") and r["out_count"] > r["kd_count"] and (r["section"], r["sq"]) not in seen_short:
            seen_short.add((r["section"], r["sq"]))
            short.append(f"{r['key']} (output {r['out_count']}x, KeyedData {r['kd_count']}x)")
    report_checks.append((f"A key repeated N times in the {report_label} output appears at least N times in KeyedData", len(short) == 0,
                          f"{len(short)} repeated key(s) have fewer rows in KeyedData than in the output" + ("; e.g. " + " | ".join(short[:2]) if short else "")))
    other_rows = [r for r in keyed_rows if r["status"] in wrong_place_statuses]
    other_counter = Counter()
    for r in other_rows:
        other_counter[f"{r['section'].strip()} -> {r['kd_section'].strip() if r['kd_section'] is not None and r['kd_section'].strip() else ('(blank)' if r['kd_section'] is not None else 'elsewhere in the file')}"] += 1
    report_checks.append((f"Every row of the {report_label} output that was found is keyed under its OWN section number", len(other_rows) == 0,
                          f"{len(other_rows)} row(s) are not keyed under their own section number" + (": " + ", ".join(f"{k} x{v}" for k, v in other_counter.most_common(5)) if other_rows else "")))
    differ_rows = [r for r in keyed_rows if r["status"] in ("Found - text differs", "Found - normalized", "Found - contained in key")]
    report_checks.append((f"Every row of the {report_label} output that was found matches KeyedData character for character (what XMATCH needs)", len(differ_rows) == 0 and len(other_rows) == 0,
                          f"{len(differ_rows)} row(s) differ in text, {len(other_rows)} are keyed elsewhere" + ("; e.g. " + show_hidden_chars(differ_rows[0]["key"]) if differ_rows else "")))
    checks[report_label] = report_checks
    per_table[report_label] = OrderedDict()
    for r in records[report]:
        per_table[report_label].setdefault(r["table"], Counter())[r["status"]] += 1

#where does the macro start reading? (rows above MACRO_FIRST_READ_ROW are never keyed)
reading_notes = []
for report in records:
    report_label = report_labels[report]
    prefix_l = "std" if report == "std" else "rat"
    blank_rows = kd_sections.get((prefix_l, ""), 0)
    title_line = first_title_line[report]
    if title_line is not None and title_line <= MACRO_FIRST_READ_ROW:
        reading_notes.append(f"{report_label}: the first {'section' if report == 'rat' else 'table'} title is on line {title_line}, but the macro starts reading at row {MACRO_FIRST_READ_ROW}. Everything above row {MACRO_FIRST_READ_ROW} is never keyed"
                             + (f" and KeyedData holds {blank_rows} rows with a BLANK section number ({prefix_l.capitalize()}__<row>)" if blank_rows else "")
                             + f". Add blank lines at the top of this output (TOP_BLANK_LINES['{report}'] = 27 in the conversion cell) so that the first title is below row {MACRO_FIRST_READ_ROW}, then load it and run the macro again.")
    elif blank_rows > 0:
        reading_notes.append(f"{report_label}: KeyedData holds {blank_rows} rows with a BLANK section number ({prefix_l.capitalize()}__<row>) although the first title of this output is on line {title_line} (below row {MACRO_FIRST_READ_ROW}). "
                             f"KeyedData was probably built from an earlier version of this report - load the current output into OpRep_01 and run the macro again.")

all_pass = all(passed for report_label in checks for name, passed, detail in checks[report_label])

print_banner("KEYEDDATA ROW CHECK", blank_line=True)
print(f"KeyedData: {keyeddata_path}")
for report in records:
    print(f"{report_labels[report]} output: {report_paths[report]}")
for report_label in checks:
    print()
    print(f"--- {report_label.upper()} ---")
    for name, passed, detail in checks[report_label]:
        print(f"[{'PASS' if passed else 'FAIL'}] {name}")
        print(f"        {detail}")
print()
print("ALL CHECKS PASSED: every output row has a row in KeyedData, under its own section number" if all_pass else "!! AT LEAST ONE CHECK FAILED - see the lines above and the 'Missing rows' / 'Found but not exact' sheets of the workbook")
print(f"(information) {len(kd_only)} Std_ / Rat_ row(s) of KeyedData were not used by any output row" + (": " + ", ".join(f"{t} {n}" for t, n in kd_only_by_table.most_common(6)) if kd_only else ""))
if reading_notes:
    print_banner("WHERE THE MACRO STARTS READING", blank_line=True)
    for note in reading_notes:
        print(note)

print_banner("RESULT PER TABLE / SECTION", blank_line=True)
for report_label in per_table:
    print(f"{report_label}:")
    for table_name in per_table[report_label]:
        counts = per_table[report_label][table_name]
        total = sum(counts.values())
        parts = [f"{counts[s]} {s.lower()}" for s in status_names if counts[s] > 0]
        print(f"   {str(table_name)[:26].ljust(26)} {str(total).rjust(6)} rows | " + " | ".join(parts))
missing_all = [r for report in records for r in records[report] if r["status"] == "Missing"]
if len(missing_all) > 0:
    why_counter = Counter()
    for r in missing_all:
        why_counter[(r["report"], r["reason"].split(";")[0][:150])] += 1
    print_banner(f"WHY ROWS ARE MISSING (all {len(missing_all)} are on the 'Missing rows' sheet)", blank_line=True)
    for (report_name, why_text), n in why_counter.most_common(12):
        print(f"[{n:>5}] {report_name}: {why_text}")
    print()
    for r in missing_all[:10]:
        print(f"{r['report']} line {r['line']}: {r['key']}")
        print(f"      {r['reason'][:300]}")
        if r["closest"] != "":
            print(f"      closest KeyedData key: {r['closest']}")

    day_missing = [r for r in missing_all if r["kind"] == "day row"]
    if len(day_missing) > 0:
        print_banner(f"DAY ROWS NOT IN KeyedData: what KeyedData holds for these contracts ({len(day_missing)} rows)", blank_line=True)
        day_seen = set()
        for r in day_missing:
            contract_norm = normalize_key_text(r["contract"])
            if (r["prefix"], r["section"], contract_norm) in day_seen or contract_norm == "":
                continue
            day_seen.add((r["prefix"], r["section"], contract_norm))
            own_keys = [x for x in kd_records if x["prefix"] == r["prefix"].lower() and x["section_l"] == r["section"].strip().lower() and x["nrest"].startswith(contract_norm)]
            any_keys = [x for x in kd_records if x["prefix"] == r["prefix"].lower() and x["nrest"].startswith(contract_norm)]
            print(f"{r['report']} table {r['section']} | contract {r['contract'].strip()!r} | output has day {r['day']} (line {r['line']})")
            print(f"      KeyedData rows for this contract in table {r['section']}: {len(own_keys)}; in any table: {len(any_keys)}")
            for x in own_keys[-3:]:
                print(f"      last keys: row {x['row']}: {x['text']!r}")
            if len(own_keys) == 0:
                print("      none - KeyedData has no row that starts with this contract name")


#7. THE EXCEL WORKBOOK

folder = REPORT_FOLDER.strip().strip('"')
if folder == "":
    folder = os.path.dirname(list(report_paths.values())[0]) or "."
prepare_results_folder(folder)
report_path = os.path.join(folder, REPORT_NAME)

fill_header = PatternFill("solid", fgColor="1F3864")
fill_found = PatternFill("solid", fgColor="C6EFCE")
fill_text = PatternFill("solid", fgColor="FFF2CC")
fill_other = PatternFill("solid", fgColor="F8CBAD")
fill_missing = PatternFill("solid", fgColor="FF9999")
fill_nokey = PatternFill("solid", fgColor="D9D9D9")
status_fill = {"Found": fill_found, "Found - text differs": fill_text, "Found - other section no.": fill_other, "Found - normalized": fill_text, "Found - contained in key": fill_text,
               "Found - other table": fill_other, "Found - elsewhere in file": fill_other, "Missing": fill_missing, "No key": fill_nokey}
font_header = Font(bold=True, color="FFFFFF")

wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Summary"
append_bold_row(ws, ["KeyedData row check: does every row of the output Standard / Ratings report have a row in KeyedData?"], size=13)
ws.append([])
ws.append(["KeyedData workbook", keyeddata_path])
ws.append(["KeyedData key column", key_letter + (f" ('{key_header}')" if key_header != "" else "")])
ws.append(["KeyedData rows with a key", len(kd_keys)])
for report in records:
    ws.append([report_labels[report] + " output report", report_paths[report]])
ws.append([])
ws.append(["Check", "Result", "Detail"])
style_cells(ws[ws.max_row], fill_header, font_header)
for report_label in checks:
    for name, passed, detail in checks[report_label]:
        ws.append([name, "PASS" if passed else "FAIL", detail])
        ws.cell(row=ws.max_row, column=2).fill = fill_found if passed else fill_missing
if reading_notes:
    ws.append([])
    append_bold_row(ws, ["Where the macro starts reading"])
    for note in reading_notes:
        ws.append([note])
ws.append([])
ws.append(["Table / section", "Rows"] + status_names)
style_cells(ws[ws.max_row], fill_header, font_header)
for report_label in per_table:
    append_bold_row(ws, [f"--- {report_label} ---"])
    for table_name in per_table[report_label]:
        counts = per_table[report_label][table_name]
        ws.append([table_name, sum(counts.values())] + [counts[s] for s in status_names])
ws.append([])
ws.append([f"(information) Std_ / Rat_ rows of KeyedData that no output row used: {len(kd_only)}"])
for table_name, n in kd_only_by_table.most_common(30):
    ws.append([table_name, n])
set_openpyxl_column_widths(ws, [85, 40, 110, 24, 24, 24, 24])

row_headers = ["Report", "Table / section", "Output line", "Kind", "Label (spaces as dots)", "Key looked up in KeyedData", "Status", "KeyedData row(s)", "Rows in output", "Rows in KeyedData", "Matched KeyedData key", "Why / note", "Closest KeyedData key"]
row_widths = [10, 16, 11, 15, 40, 60, 26, 16, 11, 12, 60, 110, 60]

for report in records:
    for rec in records[report]:
        kd_text = ""
        if len(rec["kd_rows"]) > 0:
            kd_text = str(rec["kd_rows"][0]) + (f" (+{len(rec['kd_rows']) - 1} more)" if len(rec["kd_rows"]) > 1 else "")
        rec["sheet_row"] = [rec["report"], rec["table"], rec["line"], rec["kind"], rec["shown"], show_hidden_chars(rec["key"]) if rec["key"] is not None else "(no key)", rec["status"], kd_text,
                            rec["out_count"] if rec["key"] is not None else "", rec["kd_count"] if rec["key"] is not None else "", show_hidden_chars(rec["matched_key"]),
                            rec["reason"] if rec["reason"] != "" else rec["note"], rec["closest"]]

#---- Missing rows / Found but not exact / one sheet per report with every row (the status column is colored by its status) ----
not_exact_all = [r for report in records for r in records[report] if r["status"] in not_exact_statuses]
add_table_sheet(wb, "Missing rows", row_headers, [r["sheet_row"] for r in missing_all], row_widths, fill_header, font_header, [status_fill[r["status"]] for r in missing_all], 7)
add_table_sheet(wb, "Found but not exact", row_headers, [r["sheet_row"] for r in not_exact_all], row_widths, fill_header, font_header, [status_fill[r["status"]] for r in not_exact_all], 7)
for report in records:
    add_table_sheet(wb, f"{report_labels[report]} rows", row_headers, [r["sheet_row"] for r in records[report]], row_widths, fill_header, font_header, [status_fill[r["status"]] for r in records[report]], 7)

#---- Macro structure ----
add_table_sheet(wb, "Macro structure", ["Report", "Output line", "Text on that line", "What the macro would do with it"],
                [[report_labels[report], issue["line"], issue["text"], issue["problem"]] for report in records for issue in structure_issues[report]], [10, 12, 50, 160], fill_header, font_header, auto_filter=False)

#---- KeyedData only ----
kd_only_rows = [[r["row"], show_hidden_chars(r["text"]), "No row of the output report(s) used this key (stale KeyedData, a stub row, or a table the macro read differently)"] for r in kd_only[:MAX_KEYEDDATA_ONLY_ROWS]]
if len(kd_only) > MAX_KEYEDDATA_ONLY_ROWS:
    kd_only_rows.append([f"(stopped at {MAX_KEYEDDATA_ONLY_ROWS} rows of {len(kd_only)})"])
add_table_sheet(wb, "KeyedData only", ["KeyedData row", "Key in KeyedData", "Why it is listed"], kd_only_rows, [14, 70, 100], fill_header, font_header, auto_filter=False)

wb.save(report_path)
print()
print("=" * 100)
print(f"Workbook written: {report_path}")
print("   sheets: Summary | Missing rows | Found but not exact | " + " | ".join(f"{report_labels[report]} rows" for report in records) + " | Macro structure | KeyedData only")
```

# PlexOS-4.02 (Standard Report) and 4.05 (Ratings Report) Field Transformation Validation

```python
#PURPOSE
#PlexOS-4.02 - Standard Report Field Transformation Validation

#1. SETTINGS
TEST_ID = "PlexOS-4.02"
TEST_TITLE = "Standard Report Field Transformation Validation"
TEST_NAME = TEST_ID + " - " + TEST_TITLE
REPORT_NAME = "Standard"

BASE_DIR = r"C:\Users\ZN1027\Downloads\Plexos_to_reporting\PlexOS_4_Test_Results"
REPORT_PATH = os.path.join(BASE_DIR, "Standard_Report_PLEXOS_RTSimLabels_mon.csv")
KEYEDDATA_PATH = os.path.join(BASE_DIR, "OpReports", "Testing", "TESTING_OpRep_01_DataModel_2026_rev1_260630.xlsm")
KEYEDDATA_SHEET = "KeyedData"
KEY_HEADER = "PrimaryKey"             #header text of the key column in KeyedData (column C)
KEY_PREFIX = "Std_"                #PrimaryKey prefix the macro writes for this report
KEYEDDATA_VALUE_FIRST_COL = None       #None = detect where the report's numbers start in KeyedData; or force a column letter, e.g. "H"
TOL_ABS = 1e-6                         #two numbers are equal if they differ by no more than max(TOL_ABS, TOL_REL * larger)
TOL_REL = 1e-9
OUTPUT_DIR = os.path.join(BASE_DIR, "Test_Results_" + TEST_ID)
EXCEPTION_CSV = os.path.join(OUTPUT_DIR, TEST_ID + "_Exception_Report.csv")
AUDIT_XLSX = os.path.join(OUTPUT_DIR, TEST_ID + "_Transformation_Audit.xlsx")
MAX_EXAMPLES = 8                       #examples printed per failed step
EXCEPTION_SHEET_LIMIT = 50000          #rows of the Exceptions sheet in the workbook (the CSV always has every exception)
ORPHAN_SHEET_LIMIT = 5000
HEADER_CELL_RE = re.compile(r"^Tot-\d{4}$", re.IGNORECASE)   #a row holding one of these cells is a month-header row

step_names = {1: 'Standard Report rows, attributes, and metrics are correctly transformed into KeyedData records', 2: '100% of required Standard Report fields populate KeyedData fields', 3: 'No required KeyedData field is left blank after transformation', 4: 'Field-level transformation success rate is 100%', 5: 'Invalid field mappings are captured in an exception report', 6: 'Transformation audit results are exported automatically'}
step_results = OrderedDict()           #step number -> (PASSED / FAILED, one-line detail)
exceptions = []                        #every invalid / failed mapping found, one dict each
print("=" * 100)
print("RUNNING: " + TEST_NAME)
print("=" * 100)


#2. LOCATE THE TWO INPUT FILES (prompt if a path is wrong)

input_files = OrderedDict()
input_files["report"] = [REPORT_PATH, "the Standard Report CSV written by the conversion script", [("CSV files", "*.csv"), ("All files", "*.*")]]
input_files["keyeddata"] = [KEYEDDATA_PATH, "the OpRep_01 workbook (.xlsm) that holds the KeyedData tab", [("Excel files", "*.xlsm *.xlsx"), ("All files", "*.*")]]
for input_key in input_files:
    input_path = input_files[input_key][0]
    if not os.path.isfile(input_path):
        print("FILE NOT FOUND: " + input_path)
        picked = ""
        try:
            import tkinter as tk
            from tkinter import filedialog
            root = tk.Tk()
            root.withdraw()
            root.attributes("-topmost", True)
            picked = filedialog.askopenfilename(parent=root, title="Select " + input_files[input_key][1], filetypes=input_files[input_key][2])
            root.destroy()
        except Exception:
            picked = input("Paste the full path to " + input_files[input_key][1] + ": ").strip().strip('"').strip("'")
        if not picked or not os.path.isfile(picked):
            print("FAILED: " + TEST_NAME + " (input file not available: " + input_path + ")")
            raise RuntimeError("Input file not available: " + input_path)
        input_files[input_key][0] = picked
REPORT_PATH = input_files["report"][0]
KEYEDDATA_PATH = input_files["keyeddata"][0]
os.makedirs(OUTPUT_DIR, exist_ok=True)
print("REPORT     : " + REPORT_PATH)
print("KEYEDDATA  : " + KEYEDDATA_PATH)
print("OUTPUT DIR : " + OUTPUT_DIR)


#3. LOAD KEYEDDATA (read-only)

try:
    wb_kd = openpyxl.load_workbook(KEYEDDATA_PATH, read_only=True, data_only=True)
except PermissionError:
    print("CANNOT OPEN THE WORKBOOK (it is locked). Close it in Excel and run this cell again.")
    print("FAILED: " + TEST_NAME)
    raise RuntimeError("KeyedData workbook is locked")
if KEYEDDATA_SHEET not in wb_kd.sheetnames:
    print("TAB NOT FOUND: '" + KEYEDDATA_SHEET + "'. Tabs in the workbook: " + ", ".join(wb_kd.sheetnames))
    print("FAILED: " + TEST_NAME)
    raise RuntimeError("KeyedData tab not found")
kd_all_rows = [tuple(r) for r in wb_kd[KEYEDDATA_SHEET].iter_rows(values_only=True)]
wb_kd.close()

key_col = None
header_row_index = None
for r_index in range(min(15, len(kd_all_rows))):
    for c_index in range(len(kd_all_rows[r_index])):
        if str(kd_all_rows[r_index][c_index]).strip().lower() == KEY_HEADER.lower():
            header_row_index = r_index
            key_col = c_index
            break
    if key_col is not None:
        break
if key_col is None:
    print("COLUMN NOT FOUND: no cell in the first 15 rows of KeyedData is '" + KEY_HEADER + "'.")
    print("FAILED: " + TEST_NAME)
    raise RuntimeError("PrimaryKey column not found")
kd_header = kd_all_rows[header_row_index]
kd_rows = kd_all_rows[header_row_index + 1:]
kd_excel_row0 = header_row_index + 2          #Excel row number of kd_rows[0]
kd_keys = []
for kd_row in kd_rows:
    key_cell = kd_row[key_col] if key_col < len(kd_row) else None
    kd_keys.append(str(key_cell).replace("\xa0", " ").strip() if key_cell is not None else "")
print("KEYEDDATA LOADED: " + str(len(kd_rows)) + " rows, key column " + get_column_letter(key_col + 1) + " ('" + KEY_HEADER + "'), header on Excel row " + str(header_row_index + 1))
print("KEYEDDATA ROWS WITH A '" + KEY_PREFIX + "' KEY: " + str(sum(1 for k in kd_keys if k.startswith(KEY_PREFIX))))


#4. READ THE REPORT AND DERIVE THE KEYEDDATA KEY OF EVERY DATA ROW

with open(REPORT_PATH, "r", newline="", encoding="utf-8-sig", errors="replace") as f:
    raw_lines = list(csv.reader(f))
report_rows = []
attribute_issues = []                 #titles / sections the macro would not recognise, rows without a section, ...
TITLE_RE = re.compile(r"^\s*\(\s*(\d+[A-Za-z]?)\s*\)")      #e.g. " ( 1) System Summary", " ( 33c) ..."
current_table = ""
current_plant = None
titles_seen = 0
for line_no, row in enumerate(raw_lines, start=1):
    label_raw = (row[0] if row else "").replace("\xa0", " ")
    cells = row[1:]
    label = label_raw.strip()
    vals = []
    for j, c in enumerate(cells):
        s = str(c).replace(",", "").replace("\xa0", " ").strip()
        if s == "":
            continue
        negative = s.startswith("(") and s.endswith(")")
        try:
            vals.append((j, -float(s[1:-1]) if negative else float(s)))
        except ValueError:
            pass
    m = TITLE_RE.match(label_raw)
    if m:                                                    #table title
        current_table = m.group(1)
        current_plant = None
        titles_seen += 1
        if label_raw.find("(") != 1:                         #the macro only recognises a title whose '(' is the 2nd character
            attribute_issues.append({"Severity": "FAIL", "Type": "TITLE_NOT_RECOGNISED", "Group": current_table, "Line": line_no, "Label": label_raw,
                                     "Detail": "table title does not have its '(' as the 2nd character, so the macro would not recognise it and would key its rows under the previous table"})
        continue
    if label == "" or any(HEADER_CELL_RE.match(str(c).strip()) for c in cells) or not vals:
        continue                                             #blank line, month-header row or group heading: no KeyedData record of its own
    if current_table == "":
        attribute_issues.append({"Severity": "FAIL", "Type": "ROW_WITHOUT_TABLE", "Group": "", "Line": line_no, "Label": label,
                                 "Detail": "data row is not under a recognised table title, so its key would start '" + KEY_PREFIX + "_'"})
    indent = len(label_raw) - len(label_raw.lstrip())
    cands = [KEY_PREFIX + current_table + "_" + label]       #key as the macro writes it: prefix + table + label
    if indent == 0 and label.lower() != "total":
        current_plant = label                                #un-indented row = plant / contract heading row
    elif current_plant:
        cands.append(KEY_PREFIX + current_table + "_" + current_plant + "_" + label)   #plant-scoped key (tables 3 / 6 style)
    report_rows.append({"line": line_no, "group": current_table, "label": label, "key": cands[0], "cands": cands, "values": vals, "is_control": False})
print("TABLE TITLES FOUND: " + str(titles_seen) + " (" + str(len([a for a in attribute_issues if a["Type"] == "TITLE_NOT_RECOGNISED"])) + " the macro would not recognise)")
data_row_count = len(report_rows)
print("REPORT LOADED: " + str(len(raw_lines)) + " lines, " + str(data_row_count) + " data rows, " + str(sum(len(r["values"]) for r in report_rows)) + " numeric fields")

#control row: a row that is DELIBERATELY not in KeyedData. It must end up in the exception report (STEP 5 proves the detector works).
report_rows.append({"line": 0, "group": "CONTROL", "label": "EXCEPTION_CONTROL_ROW", "key": KEY_PREFIX + "ZZ_EXCEPTION_CONTROL_ROW",
                    "cands": [KEY_PREFIX + "ZZ_EXCEPTION_CONTROL_ROW"], "values": [(0, 1.0)], "is_control": True})


#5. MATCH EVERY REPORT ROW TO ITS KEYEDDATA ROW (a repeated key is matched in order)

kd_index = defaultdict(list)
kd_index_collapsed = defaultdict(list)
for i, k in enumerate(kd_keys):
    if k:
        kd_index[k].append(i)
        kd_index_collapsed[re.sub(r"\s+", " ", k)].append(i)
kd_key_list = list(kd_index.keys())
kd_keys_by_group = defaultdict(list)                       #KeyedData keys per table / section (for the nearest-key hint)
for k in kd_key_list:
    if k.startswith(KEY_PREFIX):
        kd_keys_by_group[k[len(KEY_PREFIX):].split("_", 1)[0]].append(k)
kd_groups = set(kd_keys_by_group.keys())                    #tables / sections KeyedData has rows for
closest_lookups = 0
used_kd = set()
for r in report_rows:
    r["kd_row"] = None
    r["match"] = "MISSING"
    for cand in r["cands"]:
        hit = [i for i in kd_index.get(cand, []) if i not in used_kd]
        if hit:
            r["kd_row"], r["match"], r["matched_key"] = hit[0], "EXACT", cand
            break
    if r["kd_row"] is None:
        for cand in r["cands"]:
            hit = [i for i in kd_index_collapsed.get(re.sub(r"\s+", " ", cand), []) if i not in used_kd]
            if hit:
                r["kd_row"], r["match"], r["matched_key"] = hit[0], "SPACING", cand
                break
    if r["kd_row"] is not None:
        used_kd.add(r["kd_row"])


#6. FIND WHERE THE REPORT'S NUMBERS START IN KEYEDDATA

kd_vals = {}                          #kd row index -> list of parsed cells (None blank / float / text)
for r in report_rows:
    if r["kd_row"] is not None and r["kd_row"] not in kd_vals:
        parsed = []
        for raw in kd_rows[r["kd_row"]]:
            if raw is None:
                parsed.append(None)
            elif isinstance(raw, (int, float)) and not isinstance(raw, bool):
                parsed.append(float(raw))
            else:
                s = str(raw).replace(",", "").replace("\xa0", " ").strip()
                if s == "":
                    parsed.append(None)
                else:
                    try:
                        parsed.append(float(s))
                    except ValueError:
                        parsed.append(s)
        kd_vals[r["kd_row"]] = parsed

value_start = None
value_start_note = ""
if KEYEDDATA_VALUE_FIRST_COL:
    value_start = column_index_from_string(KEYEDDATA_VALUE_FIRST_COL) - 1
    value_start_note = "forced to column " + KEYEDDATA_VALUE_FIRST_COL
else:
    votes = Counter()
    sampled = 0
    for r in report_rows:
        if r["kd_row"] is None or r["is_control"] or sampled >= 500:
            continue
        first = next(((j, v) for j, v in r["values"] if v != 0), None)
        if first is None:
            continue
        sampled += 1
        j0, v0 = first
        row_vals = kd_vals[r["kd_row"]]
        for c in range(key_col + 1, len(row_vals)):
            kv = row_vals[c]
            if isinstance(kv, float) and abs(kv - v0) <= max(TOL_ABS, TOL_REL * max(abs(kv), abs(v0))):
                votes[c - j0] += 1
    if votes and sampled > 0:
        best_start, best_votes = votes.most_common(1)[0]
        if best_votes >= max(3, int(0.6 * sampled)):
            value_start = best_start
            value_start_note = "detected from " + str(sampled) + " rows (" + str(best_votes) + " agree)"
        else:
            value_start_note = "NOT detected: best candidate column " + get_column_letter(best_start + 1) + " only matched " + str(best_votes) + " of " + str(sampled) + " rows"
    else:
        value_start_note = "NOT detected: no report number could be found in the matched KeyedData rows"
if value_start is not None:
    print("KEYEDDATA VALUE COLUMNS START AT: " + get_column_letter(value_start + 1) + " (" + value_start_note + ")")
else:
    print("KEYEDDATA VALUE COLUMNS START AT: not found - " + value_start_note + " (set KEYEDDATA_VALUE_FIRST_COL at the top)")


#7. COMPARE EVERY FIELD

fields_total = fields_populated = fields_correct = 0
fields_in_missing_rows = fields_unlocated = blank_fields = mismatch_fields = 0
rows_total = rows_matched = rows_missing = rows_spacing_only = 0
group_stats = OrderedDict()
for r in report_rows:
    is_ctl = r["is_control"]
    if r["group"] not in group_stats:
        group_stats[r["group"]] = {"rows": 0, "rows_matched": 0, "fields": 0, "populated": 0, "correct": 0}
    gs = group_stats[r["group"]]
    excel_line = "" if is_ctl else r["line"]
    if not is_ctl:
        rows_total += 1
        gs["rows"] += 1
        fields_total += len(r["values"])
        gs["fields"] += len(r["values"])
    if r["kd_row"] is None:
        closest = []
        if not is_ctl and closest_lookups < 300:                     #nearest-key hint for the first 300 missing rows only (keeps the run fast)
            closest = get_close_matches(r["key"], kd_keys_by_group.get(r["group"], []), n=1, cutoff=0.8)
            closest_lookups += 1
        has_group = (r["group"] in kd_groups) if not is_ctl else True
        if not is_ctl:
            rows_missing += 1
            fields_in_missing_rows += len(r["values"])
        exceptions.append({"Severity": "FAIL", "Type": "MISSING_KEYEDDATA_ROW", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                           "Key": r["key"], "KeyedData_Row": "", "KeyedData_Column": "", "Source_Value": "", "KeyedData_Value": "",
                           "Fields_Affected": len(r["values"]),
                           "Detail": ("CONTROL ROW - deliberately invalid, expected in the exception report" if is_ctl else
                                      ("KeyedData has no row with this key" + ("" if has_group else " and no '" + KEY_PREFIX + r["group"] + "_' rows at all (the macro did not read this table / section)")
                                       + (". Closest key: " + closest[0] if closest else ""))),
                           "Control": "YES" if is_ctl else "NO"})
        continue
    rows_matched += 1
    gs["rows_matched"] += 1
    kd_excel_row = kd_excel_row0 + r["kd_row"]
    if r["match"] == "SPACING":
        rows_spacing_only += 1
        exceptions.append({"Severity": "WARN", "Type": "KEY_SPACING_DIFFERS", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                           "Key": r["key"], "KeyedData_Row": kd_excel_row, "KeyedData_Column": get_column_letter(key_col + 1), "Source_Value": r["key"],
                           "KeyedData_Value": kd_keys[r["kd_row"]], "Fields_Affected": 0, "Detail": "The key only matches after collapsing repeated spaces", "Control": "NO"})
    row_vals = kd_vals[r["kd_row"]]
    for j, v in r["values"]:
        if value_start is None:
            fields_unlocated += 1                #the value columns could not be located, so the field cannot be confirmed
            continue
        c = value_start + j
        kv = row_vals[c] if c < len(row_vals) else None
        col_letter = get_column_letter(c + 1)
        if kv is None:
            blank_fields += 1
            exceptions.append({"Severity": "FAIL", "Type": "BLANK_KEYEDDATA_FIELD", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                               "Key": r["key"], "KeyedData_Row": kd_excel_row, "KeyedData_Column": col_letter, "Source_Value": v, "KeyedData_Value": "",
                               "Fields_Affected": 1, "Detail": "KeyedData field is blank but the report has a value", "Control": "NO"})
            continue
        fields_populated += 1
        gs["populated"] += 1
        if isinstance(kv, float) and abs(kv - v) <= max(TOL_ABS, TOL_REL * max(abs(kv), abs(v))):
            fields_correct += 1
            gs["correct"] += 1
        else:
            mismatch_fields += 1
            exceptions.append({"Severity": "FAIL", "Type": "VALUE_MISMATCH", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                               "Key": r["key"], "KeyedData_Row": kd_excel_row, "KeyedData_Column": col_letter, "Source_Value": v, "KeyedData_Value": kv,
                               "Fields_Affected": 1, "Detail": "KeyedData value differs from the report value", "Control": "NO"})
fields_unpopulated = fields_total - fields_populated      #= fields in missing rows + blank KeyedData cells + unlocated fields

#KeyedData rows of this report type that no report row produced (information only) and rows with a blank key but numbers in them
kd_orphans = [i for i, k in enumerate(kd_keys) if k.startswith(KEY_PREFIX) and i not in used_kd]
keyed_idx = [i for i, k in enumerate(kd_keys) if k.startswith(KEY_PREFIX)]
blank_key_rows = []
if keyed_idx and value_start is not None:
    for i in range(keyed_idx[0], keyed_idx[-1] + 1):
        if kd_keys[i] == "" and any(c is not None and str(c).strip() != "" for c in kd_rows[i][value_start:]):
            blank_key_rows.append(i)
            exceptions.append({"Severity": "FAIL", "Type": "BLANK_PRIMARYKEY", "Group": "", "Report_Line": "", "Label": "", "Key": "", "KeyedData_Row": kd_excel_row0 + i,
                               "KeyedData_Column": get_column_letter(key_col + 1), "Source_Value": "", "KeyedData_Value": "", "Fields_Affected": 1,
                               "Detail": "KeyedData row has values but its PrimaryKey is blank", "Control": "NO"})
for a in attribute_issues:
    exceptions.append({"Severity": a["Severity"], "Type": a["Type"], "Group": a["Group"], "Report_Line": a["Line"], "Label": a["Label"], "Key": "", "KeyedData_Row": "",
                       "KeyedData_Column": "", "Source_Value": "", "KeyedData_Value": "", "Fields_Affected": 0, "Detail": a["Detail"], "Control": "NO"})
attr_fail = [a for a in attribute_issues if a["Severity"] == "FAIL"]
real_fail_exceptions = [e for e in exceptions if e["Severity"] == "FAIL" and e["Control"] == "NO"]
pct = (100.0 * fields_correct / fields_total) if fields_total else 0.0
pct_pop = (100.0 * fields_populated / fields_total) if fields_total else 0.0
print("RESULT COUNTS: rows " + str(rows_matched) + "/" + str(rows_total) + " matched | fields " + str(fields_populated) + "/" + str(fields_total) + " populated | "
      + str(fields_correct) + "/" + str(fields_total) + " correct | " + str(blank_fields) + " blank | " + str(mismatch_fields) + " different")
print("-" * 100)


#8. STEPS 1-4

#---- STEP 1
print("STEP 1: Verify that Standard Report rows, attributes, and metrics are correctly transformed into KeyedData records.")
rows_ok = rows_total > 0 and rows_missing == 0
attr_ok = len(attr_fail) == 0
metrics_ok = value_start is not None
print("   rows       : " + str(rows_matched) + " of " + str(rows_total) + " data rows have a KeyedData record (" + str(rows_missing) + " missing, " + str(rows_spacing_only) + " matched only after collapsing spaces)")
print("   attributes : " + str(len(attr_fail)) + " table title(s) / row(s) the macro would not read correctly")
print("   metrics    : KeyedData value columns " + ("start at " + get_column_letter(value_start + 1) if metrics_ok else "NOT FOUND (" + value_start_note + ")"))
for e in [x for x in real_fail_exceptions if x["Type"] in ("MISSING_KEYEDDATA_ROW", "TITLE_NOT_RECOGNISED", "ROW_WITHOUT_TABLE")][:MAX_EXAMPLES]:
    print("      e.g. " + e["Type"] + " | line " + str(e["Report_Line"]) + " | " + str(e["Label"])[:60] + " | " + str(e["Key"])[:70])
step1_pass = rows_ok and attr_ok and metrics_ok
step_results[1] = ("PASSED" if step1_pass else "FAILED", str(rows_matched) + "/" + str(rows_total) + " rows have a KeyedData record; " + str(len(attr_fail)) + " attribute issue(s); value columns " + ("found" if metrics_ok else "not found"))
print(step_results[1][0] + ": STEP 1 - Standard Report rows, attributes, and metrics are correctly transformed into KeyedData records")
print("-" * 100)

#---- STEP 2
print("STEP 2: Verify: 100% of required Standard Report fields populate KeyedData fields.")
print("   required fields: " + str(fields_total) + " | populated: " + str(fields_populated) + " (" + str(round(pct_pop, 4)) + " %)")
print("   not populated  : " + str(fields_unpopulated) + " = " + str(fields_in_missing_rows) + " in rows with no KeyedData record + " + str(blank_fields) + " blank KeyedData cells + " + str(fields_unlocated) + " that could not be located")
for e in [x for x in real_fail_exceptions if x["Type"] in ("MISSING_KEYEDDATA_ROW", "BLANK_KEYEDDATA_FIELD")][:MAX_EXAMPLES]:
    print("      e.g. " + e["Type"] + " | line " + str(e["Report_Line"]) + " | " + str(e["Label"])[:50] + " | KeyedData " + str(e["KeyedData_Column"]) + str(e["KeyedData_Row"]))
step2_pass = fields_total > 0 and fields_populated == fields_total
step_results[2] = ("PASSED" if step2_pass else "FAILED", str(fields_populated) + "/" + str(fields_total) + " required fields populated (" + str(round(pct_pop, 4)) + " %)")
print(step_results[2][0] + ": STEP 2 - 100% of required Standard Report fields populate KeyedData fields")
print("-" * 100)

#---- STEP 3
print("STEP 3: Verify: No required KeyedData field is left blank after transformation.")
blank_total = blank_fields + len(blank_key_rows)
print("   blank required KeyedData fields: " + str(blank_total) + " (" + str(blank_fields) + " value cells blank where the report has a value, " + str(len(blank_key_rows)) + " rows with numbers but a blank PrimaryKey)")
for e in [x for x in real_fail_exceptions if x["Type"] in ("BLANK_KEYEDDATA_FIELD", "BLANK_PRIMARYKEY")][:MAX_EXAMPLES]:
    print("      e.g. " + e["Type"] + " | " + str(e["Key"])[:60] + " | KeyedData " + str(e["KeyedData_Column"]) + str(e["KeyedData_Row"]))
step3_pass = blank_total == 0 and value_start is not None
step_results[3] = ("PASSED" if step3_pass else "FAILED", str(blank_total) + " blank required KeyedData field(s)" + ("" if value_start is not None else "; value columns not located"))
print(step_results[3][0] + ": STEP 3 - No required KeyedData field is left blank after transformation")
print("-" * 100)

#---- STEP 4
print("STEP 4: Verify: Field-level transformation success rate is 100%.")
print("   fields correct: " + str(fields_correct) + " of " + str(fields_total) + " = " + str(round(pct, 4)) + " %")
print("   not correct   : " + str(fields_total - fields_correct) + " = " + str(fields_in_missing_rows) + " in missing rows + " + str(blank_fields) + " blank + " + str(mismatch_fields) + " different value + " + str(fields_unlocated) + " not located")
for e in [x for x in real_fail_exceptions if x["Type"] == "VALUE_MISMATCH"][:MAX_EXAMPLES]:
    print("      e.g. VALUE_MISMATCH | line " + str(e["Report_Line"]) + " | " + str(e["Label"])[:40] + " | report " + str(e["Source_Value"]) + " vs KeyedData " + str(e["KeyedData_Value"]) + " (" + str(e["KeyedData_Column"]) + str(e["KeyedData_Row"]) + ")")
step4_pass = fields_total > 0 and fields_correct == fields_total
step_results[4] = ("PASSED" if step4_pass else "FAILED", "success rate " + str(round(pct, 4)) + " % (" + str(fields_correct) + "/" + str(fields_total) + " fields)")
print(step_results[4][0] + ": STEP 4 - Field-level transformation success rate is 100%")
print("-" * 100)


#9. STEP 5 - exception report

print("STEP 5: Verify: Invalid field mappings are captured in an exception report.")
exc_cols = ["Severity", "Type", "Group", "Report_Line", "Label", "Key", "KeyedData_Row", "KeyedData_Column", "Source_Value", "KeyedData_Value", "Fields_Affected", "Detail", "Control"]
control_caught = any(e["Control"] == "YES" for e in exceptions)
try:
    with open(EXCEPTION_CSV, "w", newline="", encoding="utf-8-sig") as f:
        w = csv.DictWriter(f, fieldnames=exc_cols)
        w.writeheader()
        w.writerows(exceptions)
    with open(EXCEPTION_CSV, "r", newline="", encoding="utf-8-sig") as f:
        reread = list(csv.DictReader(f))
    csv_ok = (len(reread) == len(exceptions)) and (sum(1 for e in reread if e["Control"] == "YES") == 1) and (sum(1 for e in reread if e["Severity"] == "FAIL" and e["Control"] == "NO") == len(real_fail_exceptions))
    csv_note = "exception report written and re-read: " + str(len(reread)) + " lines"
except PermissionError:
    csv_ok = False
    csv_note = "could not write " + EXCEPTION_CSV + " (close it if it is open in Excel)"
print("   control row (deliberately invalid mapping) captured as an exception: " + ("YES" if control_caught else "NO"))
print("   " + csv_note + " (" + str(len(real_fail_exceptions)) + " real failure(s), " + str(sum(1 for e in exceptions if e["Severity"] == "WARN")) + " warning(s), 1 control)")
print("   EXCEPTION REPORT: " + EXCEPTION_CSV)
step5_pass = control_caught and csv_ok
step_results[5] = ("PASSED" if step5_pass else "FAILED", "control row captured: " + ("yes" if control_caught else "no") + "; " + csv_note)
print(step_results[5][0] + ": STEP 5 - " + step_names[5])
print("-" * 100)


#10. STEP 6 - export the audit workbook automatically, then re-open it to prove it is usable

print("STEP 6: Verify: Transformation audit results are exported automatically.")
wb_out = openpyxl.Workbook()
ws = wb_out.active
ws.title = "Summary"
bold = Font(bold=True)
head_fill = PatternFill("solid", fgColor="1F4E78")
head_font = Font(bold=True, color="FFFFFF")
ws.append([TEST_NAME])
ws["A1"].font = Font(bold=True, size=14)
ws.append(["Report", REPORT_PATH])
ws.append(["KeyedData workbook", KEYEDDATA_PATH])
ws.append(["KeyedData value columns start at", get_column_letter(value_start + 1) if value_start is not None else "not found", value_start_note])
ws.append(["Data rows / matched / missing", rows_total, rows_matched, rows_missing])
ws.append(["Fields total / populated / correct", fields_total, fields_populated, fields_correct])
ws.append(["Field success rate", str(round(pct, 4)) + " %"])
ws.append(["KeyedData rows no report row produced (information)", len(kd_orphans)])
ws.append([])
ws.append(["Step", "What is verified", "Result", "Detail"])
for cell in ws[ws.max_row]:
    cell.font = head_font
    cell.fill = head_fill
for n in range(1, 6):
    ws.append([n, step_names[n], step_results[n][0], step_results[n][1]])
ws.column_dimensions["A"].width = 44
ws.column_dimensions["B"].width = 70
ws.column_dimensions["C"].width = 14
ws.column_dimensions["D"].width = 90

ws2 = wb_out.create_sheet("Group Summary")
ws2.append([("Table" if REPORT_NAME == "Standard" else "Section"), "Rows", "Rows matched", "Fields", "Fields populated", "Fields correct", "Success rate %"])
for cell in ws2[1]:
    cell.font = head_font
    cell.fill = head_fill
for g, s in group_stats.items():
    if g == "CONTROL":
        continue
    ws2.append([g, s["rows"], s["rows_matched"], s["fields"], s["populated"], s["correct"], round(100.0 * s["correct"] / s["fields"], 4) if s["fields"] else ""])
for col_letter, width in (("A", 28), ("B", 10), ("C", 14), ("D", 10), ("E", 18), ("F", 16), ("G", 16)):
    ws2.column_dimensions[col_letter].width = width

ws3 = wb_out.create_sheet("Exceptions")
ws3.append(exc_cols)
for cell in ws3[1]:
    cell.font = head_font
    cell.fill = head_fill
for e in exceptions[:EXCEPTION_SHEET_LIMIT]:
    ws3.append([e[c] for c in exc_cols])
if len(exceptions) > EXCEPTION_SHEET_LIMIT:
    ws3.append(["NOTE: only the first " + str(EXCEPTION_SHEET_LIMIT) + " of " + str(len(exceptions)) + " exceptions are shown here; the CSV has all of them."])
for col_letter, width in (("A", 10), ("B", 26), ("C", 14), ("D", 12), ("E", 36), ("F", 52), ("G", 14), ("H", 16), ("I", 16), ("J", 16), ("K", 16), ("L", 80), ("M", 10)):
    ws3.column_dimensions[col_letter].width = width
ws3.freeze_panes = "A2"

ws4 = wb_out.create_sheet("KeyedData Not Produced")
ws4.append(["KeyedData_Row", "PrimaryKey", "Note"])
for cell in ws4[1]:
    cell.font = head_font
    cell.fill = head_fill
for i in kd_orphans[:ORPHAN_SHEET_LIMIT]:
    ws4.append([kd_excel_row0 + i, kd_keys[i], "KeyedData row with a '" + KEY_PREFIX + "' key that no report row produced (information only, not a failure)"])
ws4.column_dimensions["B"].width = 70
ws4.column_dimensions["C"].width = 80

try:
    wb_out.save(AUDIT_XLSX)
    wb_check = openpyxl.load_workbook(AUDIT_XLSX)
    sheets_ok = all(s in wb_check.sheetnames for s in ("Summary", "Group Summary", "Exceptions", "KeyedData Not Produced"))
    exc_rows_ok = wb_check["Exceptions"].max_row >= 1 + min(len(exceptions), EXCEPTION_SHEET_LIMIT)
    wb_check.close()
    export_ok = sheets_ok and exc_rows_ok and os.path.getsize(AUDIT_XLSX) > 0 and os.path.getsize(EXCEPTION_CSV) > 0
    export_note = "audit workbook written and re-opened (" + str(os.path.getsize(AUDIT_XLSX)) + " bytes), exception CSV (" + str(os.path.getsize(EXCEPTION_CSV)) + " bytes)"
except PermissionError:
    export_ok = False
    export_note = "could not write " + AUDIT_XLSX + " (close it if it is open in Excel)"
print("   AUDIT WORKBOOK  : " + AUDIT_XLSX)
print("   EXCEPTION REPORT: " + EXCEPTION_CSV)
print("   " + export_note)
step_results[6] = ("PASSED" if export_ok else "FAILED", export_note)
print(step_results[6][0] + ": STEP 6 - " + step_names[6])
print("-" * 100)


#11. OVERALL RESULT

failed_steps = [n for n in step_results if step_results[n][0] != "PASSED"]
overall = "PASSED" if not failed_steps and len(step_results) == 6 else "FAILED"
if export_ok:
    wb_final = openpyxl.load_workbook(AUDIT_XLSX)
    ws_final = wb_final["Summary"]
    ws_final.append([6, step_names[6], step_results[6][0], step_results[6][1]])
    ws_final.append([])
    ws_final.append(["OVERALL", TEST_NAME, overall, ("all 6 steps passed" if overall == "PASSED" else "failed step(s): " + ", ".join(str(n) for n in failed_steps))])
    for cell in ws_final[ws_final.max_row]:
        cell.font = bold
    try:
        wb_final.save(AUDIT_XLSX)
    except PermissionError:
        print("   (could not add the overall result to the audit workbook - it is open in Excel)")
print("STEP SUMMARY")
for n in step_results:
    print("   STEP " + str(n) + ": " + step_results[n][0] + " - " + step_names[n])
print("=" * 100)
if overall == "PASSED":
    print("PASSED: " + TEST_NAME)
else:
    print("FAILED: " + TEST_NAME + " (failed step(s): " + ", ".join(str(n) for n in failed_steps) + ")")


#PURPOSE
#PlexOS-4.05 - Ratings Report Field Transformation Validation
#1. SETTINGS

TEST_ID = "PlexOS-4.05"
TEST_TITLE = "Ratings Report Field Transformation Validation"
TEST_NAME = TEST_ID + " - " + TEST_TITLE
REPORT_NAME = "Ratings"

BASE_DIR = r"C:\Users\ZN1027\Downloads\Plexos_to_reporting\PlexOS_4_Test_Results"
REPORT_PATH = os.path.join(BASE_DIR, "Ratings_Report_PLEXOS_RTSimLabels_rat.csv")
KEYEDDATA_PATH = os.path.join(BASE_DIR, "OpReports", "Testing", "TESTING_OpRep_01_DataModel_2026_rev1_260630.xlsm")
KEYEDDATA_SHEET = "KeyedData"
KEY_HEADER = "PrimaryKey"             #header text of the key column in KeyedData (column C)
KEY_PREFIX = "Rat_"                #PrimaryKey prefix the macro writes for this report
KEYEDDATA_VALUE_FIRST_COL = None       #None = detect where the report's numbers start in KeyedData; or force a column letter, e.g. "H"
TOL_ABS = 1e-6                         #two numbers are equal if they differ by no more than max(TOL_ABS, TOL_REL * larger)
TOL_REL = 1e-9
OUTPUT_DIR = os.path.join(BASE_DIR, "Test_Results_" + TEST_ID)
EXCEPTION_CSV = os.path.join(OUTPUT_DIR, TEST_ID + "_Exception_Report.csv")
AUDIT_XLSX = os.path.join(OUTPUT_DIR, TEST_ID + "_Transformation_Audit.xlsx")
MAX_EXAMPLES = 8                       #examples printed per failed step
EXCEPTION_SHEET_LIMIT = 50000          #rows of the Exceptions sheet in the workbook (the CSV always has every exception)
ORPHAN_SHEET_LIMIT = 5000
HEADER_CELL_RE = re.compile(r"^Tot-\d{4}$", re.IGNORECASE)   #a row holding one of these cells is a month-header row

step_names = {1: 'Ratings Report fields and attributes are correctly transformed into KeyedData records', 2: '100% of required Ratings Report fields populate KeyedData fields', 3: 'KeyedData record creation succeeds for all transformed Ratings Report values', 4: 'No required destination fields are blank', 5: 'Invalid mappings generate exceptions', 6: 'Transformation results are exportable'}
step_results = OrderedDict()           #step number -> (PASSED / FAILED, one-line detail)
exceptions = []                        #every invalid / failed mapping found, one dict each
print("=" * 100)
print("RUNNING: " + TEST_NAME)
print("=" * 100)


#2. LOCATE THE TWO INPUT FILES (prompt if a path is wrong)

input_files = OrderedDict()
input_files["report"] = [REPORT_PATH, "the Ratings Report CSV written by the conversion script", [("CSV files", "*.csv"), ("All files", "*.*")]]
input_files["keyeddata"] = [KEYEDDATA_PATH, "the OpRep_01 workbook (.xlsm) that holds the KeyedData tab", [("Excel files", "*.xlsm *.xlsx"), ("All files", "*.*")]]
for input_key in input_files:
    input_path = input_files[input_key][0]
    if not os.path.isfile(input_path):
        print("FILE NOT FOUND: " + input_path)
        picked = ""
        try:
            import tkinter as tk
            from tkinter import filedialog
            root = tk.Tk()
            root.withdraw()
            root.attributes("-topmost", True)
            picked = filedialog.askopenfilename(parent=root, title="Select " + input_files[input_key][1], filetypes=input_files[input_key][2])
            root.destroy()
        except Exception:
            picked = input("Paste the full path to " + input_files[input_key][1] + ": ").strip().strip('"').strip("'")
        if not picked or not os.path.isfile(picked):
            print("FAILED: " + TEST_NAME + " (input file not available: " + input_path + ")")
            raise RuntimeError("Input file not available: " + input_path)
        input_files[input_key][0] = picked
REPORT_PATH = input_files["report"][0]
KEYEDDATA_PATH = input_files["keyeddata"][0]
os.makedirs(OUTPUT_DIR, exist_ok=True)
print("REPORT     : " + REPORT_PATH)
print("KEYEDDATA  : " + KEYEDDATA_PATH)
print("OUTPUT DIR : " + OUTPUT_DIR)


#3. LOAD KEYEDDATA (read-only)

try:
    wb_kd = openpyxl.load_workbook(KEYEDDATA_PATH, read_only=True, data_only=True)
except PermissionError:
    print("CANNOT OPEN THE WORKBOOK (it is locked). Close it in Excel and run this cell again.")
    print("FAILED: " + TEST_NAME)
    raise RuntimeError("KeyedData workbook is locked")
if KEYEDDATA_SHEET not in wb_kd.sheetnames:
    print("TAB NOT FOUND: '" + KEYEDDATA_SHEET + "'. Tabs in the workbook: " + ", ".join(wb_kd.sheetnames))
    print("FAILED: " + TEST_NAME)
    raise RuntimeError("KeyedData tab not found")
kd_all_rows = [tuple(r) for r in wb_kd[KEYEDDATA_SHEET].iter_rows(values_only=True)]
wb_kd.close()

key_col = None
header_row_index = None
for r_index in range(min(15, len(kd_all_rows))):
    for c_index in range(len(kd_all_rows[r_index])):
        if str(kd_all_rows[r_index][c_index]).strip().lower() == KEY_HEADER.lower():
            header_row_index = r_index
            key_col = c_index
            break
    if key_col is not None:
        break
if key_col is None:
    print("COLUMN NOT FOUND: no cell in the first 15 rows of KeyedData is '" + KEY_HEADER + "'.")
    print("FAILED: " + TEST_NAME)
    raise RuntimeError("PrimaryKey column not found")
kd_header = kd_all_rows[header_row_index]
kd_rows = kd_all_rows[header_row_index + 1:]
kd_excel_row0 = header_row_index + 2          #Excel row number of kd_rows[0]
kd_keys = []
for kd_row in kd_rows:
    key_cell = kd_row[key_col] if key_col < len(kd_row) else None
    kd_keys.append(str(key_cell).replace("\xa0", " ").strip() if key_cell is not None else "")
print("KEYEDDATA LOADED: " + str(len(kd_rows)) + " rows, key column " + get_column_letter(key_col + 1) + " ('" + KEY_HEADER + "'), header on Excel row " + str(header_row_index + 1))
print("KEYEDDATA ROWS WITH A '" + KEY_PREFIX + "' KEY: " + str(sum(1 for k in kd_keys if k.startswith(KEY_PREFIX))))


#4. READ THE REPORT AND DERIVE THE KEYEDDATA KEY OF EVERY DATA ROW

with open(REPORT_PATH, "r", newline="", encoding="utf-8-sig", errors="replace") as f:
    raw_lines = list(csv.reader(f))
report_rows = []
attribute_issues = []                 #titles / sections the macro would not recognise, rows without a section, ...
current_section = ""
sections_seen = []
for line_no, row in enumerate(raw_lines, start=1):
    label_raw = (row[0] if row else "").replace("\xa0", " ")
    cells = row[1:]
    label = label_raw.strip()
    vals = []
    for j, c in enumerate(cells):
        s = str(c).replace(",", "").replace("\xa0", " ").strip()
        if s == "":
            continue
        negative = s.startswith("(") and s.endswith(")")
        try:
            vals.append((j, -float(s[1:-1]) if negative else float(s)))
        except ValueError:
            pass
    if label == "" or label.startswith("Ratings_Report_Title"):
        continue                                             #blank line, column-header line
    if ":" not in label and not vals:                        #section title, e.g. "ATC Mon-Sun"
        current_section = label
        sections_seen.append(label)
        continue
    resource, sep, metric = label.rpartition(":")
    if current_section == "":
        attribute_issues.append({"Severity": "FAIL", "Type": "ROW_WITHOUT_SECTION", "Group": "", "Line": line_no, "Label": label,
                                 "Detail": "data row is not under a section title, so its key would start '" + KEY_PREFIX + "_'"})
    elif sep == "" or resource.strip() == "" or metric.strip() == "":
        attribute_issues.append({"Severity": "WARN", "Type": "NO_METRIC_SUFFIX", "Group": current_section, "Line": line_no, "Label": label,
                                 "Detail": "label is not in the 'Resource:Metric' form; keyed by its full label"})
    key = KEY_PREFIX + current_section + "_" + label        #key as the macro writes it: prefix + section + 'Resource:Metric'
    report_rows.append({"line": line_no, "group": current_section, "label": label, "key": key, "cands": [key], "values": vals, "is_control": False})
print("SECTIONS FOUND: " + str(len(sections_seen)) + " (" + ", ".join(sections_seen) + ")")
data_row_count = len(report_rows)
print("REPORT LOADED: " + str(len(raw_lines)) + " lines, " + str(data_row_count) + " data rows, " + str(sum(len(r["values"]) for r in report_rows)) + " numeric fields")

#control row: a row that is DELIBERATELY not in KeyedData. It must end up in the exception report (STEP 5 proves the detector works).
report_rows.append({"line": 0, "group": "CONTROL", "label": "EXCEPTION_CONTROL_ROW", "key": KEY_PREFIX + "ZZ_EXCEPTION_CONTROL_ROW",
                    "cands": [KEY_PREFIX + "ZZ_EXCEPTION_CONTROL_ROW"], "values": [(0, 1.0)], "is_control": True})


#5. MATCH EVERY REPORT ROW TO ITS KEYEDDATA ROW (a repeated key is matched in order)

kd_index = defaultdict(list)
kd_index_collapsed = defaultdict(list)
for i, k in enumerate(kd_keys):
    if k:
        kd_index[k].append(i)
        kd_index_collapsed[re.sub(r"\s+", " ", k)].append(i)
kd_key_list = list(kd_index.keys())
kd_keys_by_group = defaultdict(list)                       #KeyedData keys per table / section (for the nearest-key hint)
for k in kd_key_list:
    if k.startswith(KEY_PREFIX):
        kd_keys_by_group[k[len(KEY_PREFIX):].split("_", 1)[0]].append(k)
kd_groups = set(kd_keys_by_group.keys())                    #tables / sections KeyedData has rows for
closest_lookups = 0
used_kd = set()
for r in report_rows:
    r["kd_row"] = None
    r["match"] = "MISSING"
    for cand in r["cands"]:
        hit = [i for i in kd_index.get(cand, []) if i not in used_kd]
        if hit:
            r["kd_row"], r["match"], r["matched_key"] = hit[0], "EXACT", cand
            break
    if r["kd_row"] is None:
        for cand in r["cands"]:
            hit = [i for i in kd_index_collapsed.get(re.sub(r"\s+", " ", cand), []) if i not in used_kd]
            if hit:
                r["kd_row"], r["match"], r["matched_key"] = hit[0], "SPACING", cand
                break
    if r["kd_row"] is not None:
        used_kd.add(r["kd_row"])


#6. FIND WHERE THE REPORT'S NUMBERS START IN KEYEDDATA

kd_vals = {}                          #kd row index -> list of parsed cells (None blank / float / text)
for r in report_rows:
    if r["kd_row"] is not None and r["kd_row"] not in kd_vals:
        parsed = []
        for raw in kd_rows[r["kd_row"]]:
            if raw is None:
                parsed.append(None)
            elif isinstance(raw, (int, float)) and not isinstance(raw, bool):
                parsed.append(float(raw))
            else:
                s = str(raw).replace(",", "").replace("\xa0", " ").strip()
                if s == "":
                    parsed.append(None)
                else:
                    try:
                        parsed.append(float(s))
                    except ValueError:
                        parsed.append(s)
        kd_vals[r["kd_row"]] = parsed

value_start = None
value_start_note = ""
if KEYEDDATA_VALUE_FIRST_COL:
    value_start = column_index_from_string(KEYEDDATA_VALUE_FIRST_COL) - 1
    value_start_note = "forced to column " + KEYEDDATA_VALUE_FIRST_COL
else:
    votes = Counter()
    sampled = 0
    for r in report_rows:
        if r["kd_row"] is None or r["is_control"] or sampled >= 500:
            continue
        first = next(((j, v) for j, v in r["values"] if v != 0), None)
        if first is None:
            continue
        sampled += 1
        j0, v0 = first
        row_vals = kd_vals[r["kd_row"]]
        for c in range(key_col + 1, len(row_vals)):
            kv = row_vals[c]
            if isinstance(kv, float) and abs(kv - v0) <= max(TOL_ABS, TOL_REL * max(abs(kv), abs(v0))):
                votes[c - j0] += 1
    if votes and sampled > 0:
        best_start, best_votes = votes.most_common(1)[0]
        if best_votes >= max(3, int(0.6 * sampled)):
            value_start = best_start
            value_start_note = "detected from " + str(sampled) + " rows (" + str(best_votes) + " agree)"
        else:
            value_start_note = "NOT detected: best candidate column " + get_column_letter(best_start + 1) + " only matched " + str(best_votes) + " of " + str(sampled) + " rows"
    else:
        value_start_note = "NOT detected: no report number could be found in the matched KeyedData rows"
if value_start is not None:
    print("KEYEDDATA VALUE COLUMNS START AT: " + get_column_letter(value_start + 1) + " (" + value_start_note + ")")
else:
    print("KEYEDDATA VALUE COLUMNS START AT: not found - " + value_start_note + " (set KEYEDDATA_VALUE_FIRST_COL at the top)")


#7. COMPARE EVERY FIELD

fields_total = fields_populated = fields_correct = 0
fields_in_missing_rows = fields_unlocated = blank_fields = mismatch_fields = 0
rows_total = rows_matched = rows_missing = rows_spacing_only = 0
group_stats = OrderedDict()
for r in report_rows:
    is_ctl = r["is_control"]
    if r["group"] not in group_stats:
        group_stats[r["group"]] = {"rows": 0, "rows_matched": 0, "fields": 0, "populated": 0, "correct": 0}
    gs = group_stats[r["group"]]
    excel_line = "" if is_ctl else r["line"]
    if not is_ctl:
        rows_total += 1
        gs["rows"] += 1
        fields_total += len(r["values"])
        gs["fields"] += len(r["values"])
    if r["kd_row"] is None:
        closest = []
        if not is_ctl and closest_lookups < 300:                     #nearest-key hint for the first 300 missing rows only (keeps the run fast)
            closest = get_close_matches(r["key"], kd_keys_by_group.get(r["group"], []), n=1, cutoff=0.8)
            closest_lookups += 1
        has_group = (r["group"] in kd_groups) if not is_ctl else True
        if not is_ctl:
            rows_missing += 1
            fields_in_missing_rows += len(r["values"])
        exceptions.append({"Severity": "FAIL", "Type": "MISSING_KEYEDDATA_ROW", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                           "Key": r["key"], "KeyedData_Row": "", "KeyedData_Column": "", "Source_Value": "", "KeyedData_Value": "",
                           "Fields_Affected": len(r["values"]),
                           "Detail": ("CONTROL ROW - deliberately invalid, expected in the exception report" if is_ctl else
                                      ("KeyedData has no row with this key" + ("" if has_group else " and no '" + KEY_PREFIX + r["group"] + "_' rows at all (the macro did not read this table / section)")
                                       + (". Closest key: " + closest[0] if closest else ""))),
                           "Control": "YES" if is_ctl else "NO"})
        continue
    rows_matched += 1
    gs["rows_matched"] += 1
    kd_excel_row = kd_excel_row0 + r["kd_row"]
    if r["match"] == "SPACING":
        rows_spacing_only += 1
        exceptions.append({"Severity": "WARN", "Type": "KEY_SPACING_DIFFERS", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                           "Key": r["key"], "KeyedData_Row": kd_excel_row, "KeyedData_Column": get_column_letter(key_col + 1), "Source_Value": r["key"],
                           "KeyedData_Value": kd_keys[r["kd_row"]], "Fields_Affected": 0, "Detail": "The key only matches after collapsing repeated spaces", "Control": "NO"})
    row_vals = kd_vals[r["kd_row"]]
    for j, v in r["values"]:
        if value_start is None:
            fields_unlocated += 1                #the value columns could not be located, so the field cannot be confirmed
            continue
        c = value_start + j
        kv = row_vals[c] if c < len(row_vals) else None
        col_letter = get_column_letter(c + 1)
        if kv is None:
            blank_fields += 1
            exceptions.append({"Severity": "FAIL", "Type": "BLANK_KEYEDDATA_FIELD", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                               "Key": r["key"], "KeyedData_Row": kd_excel_row, "KeyedData_Column": col_letter, "Source_Value": v, "KeyedData_Value": "",
                               "Fields_Affected": 1, "Detail": "KeyedData field is blank but the report has a value", "Control": "NO"})
            continue
        fields_populated += 1
        gs["populated"] += 1
        if isinstance(kv, float) and abs(kv - v) <= max(TOL_ABS, TOL_REL * max(abs(kv), abs(v))):
            fields_correct += 1
            gs["correct"] += 1
        else:
            mismatch_fields += 1
            exceptions.append({"Severity": "FAIL", "Type": "VALUE_MISMATCH", "Group": r["group"], "Report_Line": excel_line, "Label": r["label"],
                               "Key": r["key"], "KeyedData_Row": kd_excel_row, "KeyedData_Column": col_letter, "Source_Value": v, "KeyedData_Value": kv,
                               "Fields_Affected": 1, "Detail": "KeyedData value differs from the report value", "Control": "NO"})
fields_unpopulated = fields_total - fields_populated      #= fields in missing rows + blank KeyedData cells + unlocated fields

#KeyedData rows of this report type that no report row produced (information only) and rows with a blank key but numbers in them
kd_orphans = [i for i, k in enumerate(kd_keys) if k.startswith(KEY_PREFIX) and i not in used_kd]
keyed_idx = [i for i, k in enumerate(kd_keys) if k.startswith(KEY_PREFIX)]
blank_key_rows = []
if keyed_idx and value_start is not None:
    for i in range(keyed_idx[0], keyed_idx[-1] + 1):
        if kd_keys[i] == "" and any(c is not None and str(c).strip() != "" for c in kd_rows[i][value_start:]):
            blank_key_rows.append(i)
            exceptions.append({"Severity": "FAIL", "Type": "BLANK_PRIMARYKEY", "Group": "", "Report_Line": "", "Label": "", "Key": "", "KeyedData_Row": kd_excel_row0 + i,
                               "KeyedData_Column": get_column_letter(key_col + 1), "Source_Value": "", "KeyedData_Value": "", "Fields_Affected": 1,
                               "Detail": "KeyedData row has values but its PrimaryKey is blank", "Control": "NO"})
for a in attribute_issues:
    exceptions.append({"Severity": a["Severity"], "Type": a["Type"], "Group": a["Group"], "Report_Line": a["Line"], "Label": a["Label"], "Key": "", "KeyedData_Row": "",
                       "KeyedData_Column": "", "Source_Value": "", "KeyedData_Value": "", "Fields_Affected": 0, "Detail": a["Detail"], "Control": "NO"})
attr_fail = [a for a in attribute_issues if a["Severity"] == "FAIL"]
real_fail_exceptions = [e for e in exceptions if e["Severity"] == "FAIL" and e["Control"] == "NO"]
pct = (100.0 * fields_correct / fields_total) if fields_total else 0.0
pct_pop = (100.0 * fields_populated / fields_total) if fields_total else 0.0
print("RESULT COUNTS: rows " + str(rows_matched) + "/" + str(rows_total) + " matched | fields " + str(fields_populated) + "/" + str(fields_total) + " populated | "
      + str(fields_correct) + "/" + str(fields_total) + " correct | " + str(blank_fields) + " blank | " + str(mismatch_fields) + " different")
print("-" * 100)


#8. STEPS 1-4

#---- STEP 1
print("STEP 1: Verify that Ratings Report fields and attributes are correctly transformed into KeyedData records.")
rows_ok = rows_total > 0 and rows_missing == 0
attr_ok = len(attr_fail) == 0
metrics_ok = value_start is not None
values_ok = metrics_ok and mismatch_fields == 0
print("   rows       : " + str(rows_matched) + " of " + str(rows_total) + " data rows have a KeyedData record (" + str(rows_missing) + " missing, " + str(rows_spacing_only) + " matched only after collapsing spaces)")
print("   attributes : " + str(len(sections_seen)) + " section(s) found, " + str(len(attr_fail)) + " row(s) without a section / key attribute (" + str(len(attribute_issues) - len(attr_fail)) + " warning(s): label not in Resource:Metric form)")
print("   fields     : KeyedData value columns " + ("start at " + get_column_letter(value_start + 1) if metrics_ok else "NOT FOUND (" + value_start_note + ")") + "; " + str(mismatch_fields) + " value(s) differ from the report")
for e in [x for x in real_fail_exceptions if x["Type"] in ("MISSING_KEYEDDATA_ROW", "ROW_WITHOUT_SECTION", "VALUE_MISMATCH")][:MAX_EXAMPLES]:
    print("      e.g. " + e["Type"] + " | line " + str(e["Report_Line"]) + " | " + str(e["Group"])[:24] + " | " + str(e["Label"])[:50])
step1_pass = rows_ok and attr_ok and metrics_ok and values_ok
step_results[1] = ("PASSED" if step1_pass else "FAILED", str(rows_matched) + "/" + str(rows_total) + " rows have a KeyedData record; " + str(len(attr_fail)) + " attribute issue(s); " + str(mismatch_fields) + " value(s) differ")
print(step_results[1][0] + ": STEP 1 - Ratings Report fields and attributes are correctly transformed into KeyedData records")
print("-" * 100)

#---- STEP 2
print("STEP 2: Verify: 100% of required Ratings Report fields populate KeyedData fields.")
print("   required fields: " + str(fields_total) + " | populated: " + str(fields_populated) + " (" + str(round(pct_pop, 4)) + " %)")
print("   not populated  : " + str(fields_unpopulated) + " = " + str(fields_in_missing_rows) + " in rows with no KeyedData record + " + str(blank_fields) + " blank KeyedData cells + " + str(fields_unlocated) + " that could not be located")
for e in [x for x in real_fail_exceptions if x["Type"] in ("MISSING_KEYEDDATA_ROW", "BLANK_KEYEDDATA_FIELD")][:MAX_EXAMPLES]:
    print("      e.g. " + e["Type"] + " | line " + str(e["Report_Line"]) + " | " + str(e["Group"])[:24] + " | " + str(e["Label"])[:50])
step2_pass = fields_total > 0 and fields_populated == fields_total
step_results[2] = ("PASSED" if step2_pass else "FAILED", str(fields_populated) + "/" + str(fields_total) + " required fields populated (" + str(round(pct_pop, 4)) + " %)")
print(step_results[2][0] + ": STEP 2 - 100% of required Ratings Report fields populate KeyedData fields")
print("-" * 100)

#---- STEP 3
print("STEP 3: Verify: KeyedData record creation succeeds for all transformed Ratings Report values.")
print("   KeyedData records created for " + str(rows_matched) + " of " + str(rows_total) + " Ratings Report rows (" + str(rows_missing) + " rows have no record, carrying " + str(fields_in_missing_rows) + " values)")
for g, s in group_stats.items():
    if g != "CONTROL":
        print("      " + str(g)[:30].ljust(30) + " rows " + str(s["rows_matched"]) + "/" + str(s["rows"]) + " | values stored " + str(s["populated"]) + "/" + str(s["fields"]))
print("   information: " + str(len(kd_orphans)) + " '" + KEY_PREFIX + "' KeyedData row(s) were not produced by any Ratings Report row")
for e in [x for x in real_fail_exceptions if x["Type"] == "MISSING_KEYEDDATA_ROW"][:MAX_EXAMPLES]:
    print("      e.g. MISSING_KEYEDDATA_ROW | line " + str(e["Report_Line"]) + " | " + str(e["Key"])[:80])
step3_pass = rows_total > 0 and rows_missing == 0 and value_start is not None
step_results[3] = ("PASSED" if step3_pass else "FAILED", str(rows_matched) + "/" + str(rows_total) + " KeyedData records created" + ("" if value_start is not None else "; value columns not located"))
print(step_results[3][0] + ": STEP 3 - KeyedData record creation succeeds for all transformed Ratings Report values")
print("-" * 100)

#---- STEP 4
print("STEP 4: Verify: No required destination fields are blank.")
blank_total = blank_fields + len(blank_key_rows)
print("   blank required destination fields: " + str(blank_total) + " (" + str(blank_fields) + " value cells blank where the report has a value, " + str(len(blank_key_rows)) + " rows with numbers but a blank PrimaryKey)")
for e in [x for x in real_fail_exceptions if x["Type"] in ("BLANK_KEYEDDATA_FIELD", "BLANK_PRIMARYKEY")][:MAX_EXAMPLES]:
    print("      e.g. " + e["Type"] + " | " + str(e["Key"])[:60] + " | KeyedData " + str(e["KeyedData_Column"]) + str(e["KeyedData_Row"]))
step4_pass = blank_total == 0 and value_start is not None
step_results[4] = ("PASSED" if step4_pass else "FAILED", str(blank_total) + " blank required destination field(s)" + ("" if value_start is not None else "; value columns not located"))
print(step_results[4][0] + ": STEP 4 - No required destination fields are blank")
print("-" * 100)


#9. STEP 5 - exception report

print("STEP 5: Verify: Invalid mappings generate exceptions.")
exc_cols = ["Severity", "Type", "Group", "Report_Line", "Label", "Key", "KeyedData_Row", "KeyedData_Column", "Source_Value", "KeyedData_Value", "Fields_Affected", "Detail", "Control"]
control_caught = any(e["Control"] == "YES" for e in exceptions)
try:
    with open(EXCEPTION_CSV, "w", newline="", encoding="utf-8-sig") as f:
        w = csv.DictWriter(f, fieldnames=exc_cols)
        w.writeheader()
        w.writerows(exceptions)
    with open(EXCEPTION_CSV, "r", newline="", encoding="utf-8-sig") as f:
        reread = list(csv.DictReader(f))
    csv_ok = (len(reread) == len(exceptions)) and (sum(1 for e in reread if e["Control"] == "YES") == 1) and (sum(1 for e in reread if e["Severity"] == "FAIL" and e["Control"] == "NO") == len(real_fail_exceptions))
    csv_note = "exception report written and re-read: " + str(len(reread)) + " lines"
except PermissionError:
    csv_ok = False
    csv_note = "could not write " + EXCEPTION_CSV + " (close it if it is open in Excel)"
print("   control row (deliberately invalid mapping) captured as an exception: " + ("YES" if control_caught else "NO"))
print("   " + csv_note + " (" + str(len(real_fail_exceptions)) + " real failure(s), " + str(sum(1 for e in exceptions if e["Severity"] == "WARN")) + " warning(s), 1 control)")
print("   EXCEPTION REPORT: " + EXCEPTION_CSV)
step5_pass = control_caught and csv_ok
step_results[5] = ("PASSED" if step5_pass else "FAILED", "control row captured: " + ("yes" if control_caught else "no") + "; " + csv_note)
print(step_results[5][0] + ": STEP 5 - " + step_names[5])
print("-" * 100)


#10. STEP 6 - export the audit workbook automatically, then re-open it to prove it is usable

print("STEP 6: Verify: Transformation results are exportable.")
wb_out = openpyxl.Workbook()
ws = wb_out.active
ws.title = "Summary"
bold = Font(bold=True)
head_fill = PatternFill("solid", fgColor="1F4E78")
head_font = Font(bold=True, color="FFFFFF")
ws.append([TEST_NAME])
ws["A1"].font = Font(bold=True, size=14)
ws.append(["Report", REPORT_PATH])
ws.append(["KeyedData workbook", KEYEDDATA_PATH])
ws.append(["KeyedData value columns start at", get_column_letter(value_start + 1) if value_start is not None else "not found", value_start_note])
ws.append(["Data rows / matched / missing", rows_total, rows_matched, rows_missing])
ws.append(["Fields total / populated / correct", fields_total, fields_populated, fields_correct])
ws.append(["Field success rate", str(round(pct, 4)) + " %"])
ws.append(["KeyedData rows no report row produced (information)", len(kd_orphans)])
ws.append([])
ws.append(["Step", "What is verified", "Result", "Detail"])
for cell in ws[ws.max_row]:
    cell.font = head_font
    cell.fill = head_fill
for n in range(1, 6):
    ws.append([n, step_names[n], step_results[n][0], step_results[n][1]])
ws.column_dimensions["A"].width = 44
ws.column_dimensions["B"].width = 70
ws.column_dimensions["C"].width = 14
ws.column_dimensions["D"].width = 90

ws2 = wb_out.create_sheet("Group Summary")
ws2.append([("Table" if REPORT_NAME == "Standard" else "Section"), "Rows", "Rows matched", "Fields", "Fields populated", "Fields correct", "Success rate %"])
for cell in ws2[1]:
    cell.font = head_font
    cell.fill = head_fill
for g, s in group_stats.items():
    if g == "CONTROL":
        continue
    ws2.append([g, s["rows"], s["rows_matched"], s["fields"], s["populated"], s["correct"], round(100.0 * s["correct"] / s["fields"], 4) if s["fields"] else ""])
for col_letter, width in (("A", 28), ("B", 10), ("C", 14), ("D", 10), ("E", 18), ("F", 16), ("G", 16)):
    ws2.column_dimensions[col_letter].width = width

ws3 = wb_out.create_sheet("Exceptions")
ws3.append(exc_cols)
for cell in ws3[1]:
    cell.font = head_font
    cell.fill = head_fill
for e in exceptions[:EXCEPTION_SHEET_LIMIT]:
    ws3.append([e[c] for c in exc_cols])
if len(exceptions) > EXCEPTION_SHEET_LIMIT:
    ws3.append(["NOTE: only the first " + str(EXCEPTION_SHEET_LIMIT) + " of " + str(len(exceptions)) + " exceptions are shown here; the CSV has all of them."])
for col_letter, width in (("A", 10), ("B", 26), ("C", 14), ("D", 12), ("E", 36), ("F", 52), ("G", 14), ("H", 16), ("I", 16), ("J", 16), ("K", 16), ("L", 80), ("M", 10)):
    ws3.column_dimensions[col_letter].width = width
ws3.freeze_panes = "A2"

ws4 = wb_out.create_sheet("KeyedData Not Produced")
ws4.append(["KeyedData_Row", "PrimaryKey", "Note"])
for cell in ws4[1]:
    cell.font = head_font
    cell.fill = head_fill
for i in kd_orphans[:ORPHAN_SHEET_LIMIT]:
    ws4.append([kd_excel_row0 + i, kd_keys[i], "KeyedData row with a '" + KEY_PREFIX + "' key that no report row produced (information only, not a failure)"])
ws4.column_dimensions["B"].width = 70
ws4.column_dimensions["C"].width = 80

try:
    wb_out.save(AUDIT_XLSX)
    wb_check = openpyxl.load_workbook(AUDIT_XLSX)
    sheets_ok = all(s in wb_check.sheetnames for s in ("Summary", "Group Summary", "Exceptions", "KeyedData Not Produced"))
    exc_rows_ok = wb_check["Exceptions"].max_row >= 1 + min(len(exceptions), EXCEPTION_SHEET_LIMIT)
    wb_check.close()
    export_ok = sheets_ok and exc_rows_ok and os.path.getsize(AUDIT_XLSX) > 0 and os.path.getsize(EXCEPTION_CSV) > 0
    export_note = "audit workbook written and re-opened (" + str(os.path.getsize(AUDIT_XLSX)) + " bytes), exception CSV (" + str(os.path.getsize(EXCEPTION_CSV)) + " bytes)"
except PermissionError:
    export_ok = False
    export_note = "could not write " + AUDIT_XLSX + " (close it if it is open in Excel)"
print("   AUDIT WORKBOOK  : " + AUDIT_XLSX)
print("   EXCEPTION REPORT: " + EXCEPTION_CSV)
print("   " + export_note)
step_results[6] = ("PASSED" if export_ok else "FAILED", export_note)
print(step_results[6][0] + ": STEP 6 - " + step_names[6])
print("-" * 100)


#11. OVERALL RESULT

failed_steps = [n for n in step_results if step_results[n][0] != "PASSED"]
overall = "PASSED" if not failed_steps and len(step_results) == 6 else "FAILED"
if export_ok:
    wb_final = openpyxl.load_workbook(AUDIT_XLSX)
    ws_final = wb_final["Summary"]
    ws_final.append([6, step_names[6], step_results[6][0], step_results[6][1]])
    ws_final.append([])
    ws_final.append(["OVERALL", TEST_NAME, overall, ("all 6 steps passed" if overall == "PASSED" else "failed step(s): " + ", ".join(str(n) for n in failed_steps))])
    for cell in ws_final[ws_final.max_row]:
        cell.font = bold
    try:
        wb_final.save(AUDIT_XLSX)
    except PermissionError:
        print("   (could not add the overall result to the audit workbook - it is open in Excel)")
print("STEP SUMMARY")
for n in step_results:
    print("   STEP " + str(n) + ": " + step_results[n][0] + " - " + step_names[n])
print("=" * 100)
if overall == "PASSED":
    print("PASSED: " + TEST_NAME)
else:
    print("FAILED: " + TEST_NAME + " (failed step(s): " + ", ".join(str(n) for n in failed_steps) + ")")
```
